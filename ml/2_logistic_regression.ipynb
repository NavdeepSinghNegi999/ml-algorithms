# Logistic Regression (From Scratch)

## 1. Linear Equation

$$
z = XW + b
$$

Code

```python
z = X @ self.w + self.b
```

---

## 2. Sigmoid Function

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Code

```python
y_pred = self.sigmoid(z)
```

---

## 3. Binary Cross Entropy Loss

$$
J(W,b)
=
-\frac1n
\sum_{i=1}^{n}
\left[
y_i\log(\hat y_i)
+
(1-y_i)\log(1-\hat y_i)
\right]
$$

where

$$
\hat y=\sigma(z)
$$

---

## 4. Gradient w.r.t Weight

$$
\frac{\partial J}{\partial W}
=
\frac1n
X^T(\hat y-y)
$$

Code

```python
dw = (1/n_samples) * X.T @ (y_pred - y)
```

---

## 5. Gradient w.r.t Bias

$$
\frac{\partial J}{\partial b}
=
\frac1n
\sum_{i=1}^{n}
(\hat y_i-y_i)
$$

Code

```python
db = (1/n_samples) * np.sum(y_pred - y)
```

---

## 6. Gradient Descent Update

$$
W=W-\alpha\frac{\partial J}{\partial W}
$$

$$
b=b-\alpha\frac{\partial J}{\partial b}
$$

Code

```python
self.w -= self.lr * dw
self.b -= self.lr * db
```

---

## 7. Prediction

$$
\hat y=
\begin{cases}
1 & \text{if } \sigma(z)\ge0.5\\
0 & \text{otherwise}
\end{cases}
$$

Code

```python
return (prob >= 0.5).astype(int)
```

---

## Shapes

| Variable | Shape |
|-----------|-------|
| $X$ | $(n_{samples},n_{features})$ |
| $y$ | $(n_{samples},)$ |
| $W$ | $(n_{features},)$ |
| $b$ | Scalar |
| $z$ | $(n_{samples},)$ |
| $\hat y$ | $(n_{samples},)$ |
| $dw$ | $(n_{features},)$ |
| $db$ | Scalar |

# 1. Dataset

````python
import numpy as np
from sklearn.linear_model import LogisticRegression

# -------------------- Dataset --------------------
# X shape = (8,2)

X = np.array([
    [1,2],
    [2,1],
    [2,3],
    [3,2],
    [6,5],
    [7,6],
    [8,7],
    [9,8]
], dtype=float)

# y shape = (8,)
y = np.array([0,0,0,0,1,1,1,1])

print("X Shape:", X.shape)
print("y Shape:", y.shape)
````

# 2. Sklearn

````python
# -------------------- Sklearn --------------------

sk_model = LogisticRegression()

sk_model.fit(X, y)

sk_pred = sk_model.predict(X)
````

# 3. From Scratch

````python
class LogisticRegressionScratch:

    def __init__(self, lr=0.1, epochs=5000):
        self.lr = lr
        self.epochs = epochs

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):

        n_samples, n_features = X.shape

        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.epochs):

            # Linear Equation
            z = X @ self.w + self.b

            # Sigmoid
            y_pred = self.sigmoid(z)

            # Gradients
            dw = (1/n_samples) * X.T @ (y_pred - y)
            db = (1/n_samples) * np.sum(y_pred - y)

            # Update
            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):

        z = X @ self.w + self.b

        prob = self.sigmoid(z)

        return (prob >= 0.5).astype(int)


my_model = LogisticRegressionScratch()

my_model.fit(X,y)

my_pred = my_model.predict(X)

print("\n===== Sklearn =====")
print("Prediction:")
print(sk_pred)

print("Accuracy:", sk_model.score(X,y))


print("\n===== Scratch =====")
print("Prediction:")
print(my_pred)

print("\n===== Weight & biases =====")
print("Weights:", my_model.w)
print("Bias:", my_model.b)
````

# 4. Accuracy

````python
from sklearn.metrics import accuracy_score

print("\n===== Sklearn =====")
print("Accuracy :", accuracy_score(y, sk_pred))

print("\n===== Scratch =====")
print("Accuracy :", accuracy_score(y, my_pred))
````

# 5. Logistic Regression Gradient Derivation



## Step 1: Linear Model

$$
z = XW + b
$$

---

## Step 2: Sigmoid Function

$$
\hat{y} = \sigma(z)=\frac{1}{1+e^{-z}}
$$

---

## Step 3: Binary Cross Entropy Loss

$$
J(W,b)=
-\frac1n
\sum_{i=1}^{n}
\left[
y_i\log(\hat y_i)
+
(1-y_i)\log(1-\hat y_i)
\right]
$$

---

# Derivation

We need

$$
\frac{\partial J}{\partial W}
$$

Using the chain rule,

$$
\frac{\partial J}{\partial W}
=
\frac{\partial J}{\partial \hat y}
\cdot
\frac{\partial \hat y}{\partial z}
\cdot
\frac{\partial z}{\partial W}
$$

---

## Step 4: Derivative of BCE Loss

Differentiate the loss with respect to $\hat y$

$$
\frac{\partial J}{\partial \hat y}
=
-\frac{y}{\hat y}
+
\frac{1-y}{1-\hat y}
$$

---

## Step 5: Derivative of Sigmoid

Since

$$
\hat y=\sigma(z)
$$

its derivative is

$$
\frac{\partial \hat y}{\partial z}
=
\hat y(1-\hat y)
$$

---

## Step 6: Multiply Both

$$
\frac{\partial J}{\partial z}
=
\left(
-\frac{y}{\hat y}
+
\frac{1-y}{1-\hat y}
\right)
\hat y(1-\hat y)
$$

Expand

$$
=
-y(1-\hat y)
+
(1-y)\hat y
$$

Expand further

$$
=
-y+y\hat y+\hat y-y\hat y
$$

Cancel the common terms

$$
=
\hat y-y
$$

Hence

$$
\boxed{
\frac{\partial J}{\partial z}
=
\hat y-y
}
$$

---

## Step 7: Derivative of Linear Equation

Since

$$
z=XW+b
$$

$$
\frac{\partial z}{\partial W}=X
$$

Therefore

$$
\boxed{
\frac{\partial J}{\partial W}
=
\frac1n
X^T(\hat y-y)
}
$$

---

## Step 8: Bias Gradient

$$
\boxed{
\frac{\partial J}{\partial b}
=
\frac1n
\sum_{i=1}^{n}
(\hat y_i-y_i)
}
$$

---

## Step 9: Gradient Descent Update

Weights

$$
W
=
W-
\alpha
\frac{\partial J}{\partial W}
$$

Bias

$$
b
=
b-
\alpha
\frac{\partial J}{\partial b}
$$

---

# Final Result

$$
\boxed{
dw=
\frac1n
X^T(\hat y-y)
}
$$

$$
\boxed{
db=
\frac1n
\sum(\hat y-y)
}
$$
