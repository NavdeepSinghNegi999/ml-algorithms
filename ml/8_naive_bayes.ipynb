# 3. Scratch Code

```python
class GaussianNBScratch:

    def fit(self, X, y):

        self.classes = np.unique(y)

        self.mean = {}
        self.var = {}
        self.prior = {}

        for c in self.classes:

            X_c = X[y == c]

            self.mean[c] = X_c.mean(axis=0)
            self.var[c] = X_c.var(axis=0) + 1e-9
            self.prior[c] = len(X_c) / len(X)

    def gaussian(self, x, mean, var):

        return (1 / np.sqrt(2 * np.pi * var)) * 
            np.exp(-(x - mean) ** 2 / (2 * var))

    def predict(self, X):

        predictions = []

        for x in X:

            scores = []

            for c in self.classes:

                likelihood = self.gaussian(x, self.mean[c], self.var[c])

                # Naive Bayes:
                # P(X|C) = product of feature likelihoods
                score = (self.prior[c] * np.prod(likelihood))

                scores.append(score)

            predictions.append(self.classes[np.argmax(scores)])

        return np.array(predictions)
```

# 4. Naive Bayes — Important Formulas

## 1. Bayes Theorem

$$
P(C|X)
=
\frac{P(X|C)P(C)}
{P(X)}
$$

For classification, $P(X)$ is the same for every class, so:

$$
P(C|X)
\propto
P(X|C)P(C)
$$

---

## 2. Naive Independence Assumption

Naive Bayes assumes that features are conditionally independent given the class.

$$
P(X|C)
=
\prod_{j=1}^{m}
P(x_j|C)
$$

Therefore:

$$
P(C|X)
\propto
P(C)
\prod_{j=1}^{m}
P(x_j|C)
$$

---

## 3. Prior Probability

$$
P(C_k)
=
\frac{N_k}{N}
$$

where:

- $N_k$ = number of samples in class $k$
- $N$ = total number of samples

---

## 4. Gaussian Likelihood

For continuous features:

$$
P(x_j|C_k)
=
\frac{1}
{\sqrt{2\pi\sigma_{kj}^{2}}}
\exp
\left(
-\frac{(x_j-\mu_{kj})^2}
{2\sigma_{kj}^{2}}
\right)
$$

where:

$$
\mu_{kj} = \text{mean of feature }j\text{ in class }k
$$

$$
\sigma_{kj}^{2}
=
\text{variance of feature }j\text{ in class }k
$$

---

## 5. Final Prediction

Choose the class with the highest posterior probability:

$$
\hat{y}
=
\arg\max_k
P(C_k)
\prod_{j=1}^{m}
P(x_j|C_k)
$$

---

## 6. Log Probability

To avoid numerical underflow, we can use logarithms:

$$
\log P(C_k|X)
\propto
\log P(C_k)
+
\sum_{j=1}^{m}
\log P(x_j|C_k)
$$

Therefore:

$$
\hat{y}
=
\arg\max_k
\left[
\log P(C_k)
+
\sum_j
\log P(x_j|C_k)
\right]
$$
