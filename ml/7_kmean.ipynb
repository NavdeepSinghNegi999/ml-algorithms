# 1. Dataset

```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split

X, y = make_blobs(n_samples=200,centers=3, n_features=2, cluster_std=1.0, random_state=42)

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

# 2. Sklearn 

```python
from sklearn.cluster import KMeans

sk_model = KMeans(n_clusters=3, random_state=42,n_init=10)

sk_model.fit(X)

sk_pred = sk_model.labels_

print("Centroids:")
print(sk_model.cluster_centers_)

print("\nClusters:")
print(sk_pred)
```

# 3. Scratch Code

```python
import numpy as np

class KMeansScratch:

    def __init__(self, k=3, max_iters=100):
        self.k = k
        self.max_iters = max_iters

    def fit(self, X):

        # STEP 1: Initialize centroids
        idx = np.random.choice(len(X), self.k, replace=False)
        self.centroids = X[idx]

        for _ in range(self.max_iters):

            # STEP 2: Assign points
            labels = []

            for x in X:
                distances = []

                for centroid in self.centroids:
                    distance = np.linalg.norm(x - centroid)
                    distances.append(distance)

                labels.append(np.argmin(distances))

            labels = np.array(labels)

            # STEP 3: Update centroids
            new_centroids = []

            for i in range(self.k):
                points = X[labels == i]
                new_centroids.append(points.mean(axis=0))

            new_centroids = np.array(new_centroids)

            # STEP 4: Stop if unchanged
            if np.allclose(self.centroids, new_centroids):
                break

            self.centroids = new_centroids

        self.labels = labels

    def predict(self, X):

        predictions = []

        for x in X:
            distances = []

            for centroid in self.centroids:
                distance = np.linalg.norm(x - centroid)
                distances.append(distance)

            predictions.append(np.argmin(distances))

        return np.array(predictions)
```

```python
from sklearn.metrics import accuracy_score

my_model = KMeansScratch(k=3)

my_model.fit(X)

my_pred = my_model.predict(X)

print("Centroids:")
print(my_model.centroids)

print("\nClusters:")
print(my_pred)
```

```python
from sklearn.metrics import adjusted_rand_score

print("Sklearn ARI:", adjusted_rand_score(y, sk_pred))

print("Scratch ARI:", adjusted_rand_score(y, my_pred))
```

# 4. K-Means — Important Formulas

## 1. Euclidean Distance

Distance between a point and centroid:

$$
d(x_i,c_k)
=
\sqrt{
\sum_{j=1}^{m}
(x_{ij}-c_{kj})^2
}
$$

---

## 2. Cluster Assignment

Assign each point to the nearest centroid:

$$
z_i
=
\arg\min_k
d(x_i,c_k)
$$

---

## 3. Centroid Update

Calculate the mean of all points belonging to a cluster:

$$
c_k
=
\frac{1}{N_k}
\sum_{x_i\in C_k}
x_i
$$

---

## 4. K-Means Objective Function

K-Means minimizes the within-cluster sum of squared distances:

$$
J
=
\sum_{k=1}^{K}
\sum_{x_i\in C_k}
\|x_i-c_k\|^2
$$

---

## 5. Algorithm

### Step 1 — Initialize

Choose $K$ centroids.

### Step 2 — Assignment

Assign every point to its nearest centroid.

$$
z_i=\arg\min_k d(x_i,c_k)
$$

### Step 3 — Update

Recalculate each centroid.

$$
c_k
=
\frac{1}{N_k}
\sum_{x_i\in C_k}x_i
$$

### Step 4 — Repeat

Repeat assignment and update until convergence.

### Step 5 — Stop

Stop when:

$$
c_k^{(new)} \approx c_k^{(old)}
$$

or the maximum number of iterations is reached.
