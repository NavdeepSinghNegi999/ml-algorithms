# Linear Regression (From Scratch)

## 1. Prediction

The prediction equation is

$$
\hat{y} = XW + b
$$

where

- $X$ : Feature matrix of shape $(n_{\text{samples}}, n_{\text{features}})$
- $W$ : Weight vector of shape $(n_{\text{features}},)$
- $b$ : Bias (scalar)

Code:

```python
y_pred = X @ self.w + self.b
```

---

## 2. Cost Function (Mean Squared Error)

The objective is to minimize

$$
J(W,b)=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Code:

```python
# Not explicitly calculated because we directly compute gradients
```

---

## 3. Gradient with respect to Weights

$$
\frac{\partial J}{\partial W}
=
\frac{2}{n}
X^T(\hat{y}-y)
$$

Code:

```python
dw = (2 / n_samples) * X.T @ (y_pred - y)
```

---

## 4. Gradient with respect to Bias

$$
\frac{\partial J}{\partial b}
=
\frac{2}{n}
\sum_{i=1}^{n}(\hat{y}_i-y_i)
$$

Code:

```python
db = (2 / n_samples) * np.sum(y_pred - y)
```

---

## 5. Gradient Descent Update

Update the parameters as

$$
W = W - \alpha \frac{\partial J}{\partial W}
$$

$$
b = b - \alpha \frac{\partial J}{\partial b}
$$

where $\alpha$ is the learning rate.

Code:

```python
self.w -= self.lr * dw
self.b -= self.lr * db
```

---

## Shapes

| Variable | Shape |
|-----------|-------|
| $X$ | $(n_{\text{samples}}, n_{\text{features}})$ |
| $y$ | $(n_{\text{samples}},)$ |
| $W$ | $(n_{\text{features}},)$ |
| $b$ | Scalar |
| $\hat{y}$ | $(n_{\text{samples}},)$ |
| $dw$ | $(n_{\text{features}},)$ |
| $db$ | Scalar |

# 1. Dataset

````python
import numpy as np
from sklearn.linear_model import LinearRegression

# -------------------- Dataset --------------------
# X shape = (8, 2)
X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3],
    [5, 5],
    [6, 7],
    [7, 6],
    [8, 9]
], dtype=float)

# y shape = (8,)
y = np.array([5, 5, 11, 11, 15, 20, 20, 26], dtype=float)

print("X Shape:", X.shape)
print("y Shape:", y.shape)
````

# 2. Sklearn

````python
# -------------------- Sklearn --------------------
sk_model = LinearRegression()
sk_model.fit(X, y)

print("\nSklearn Prediction:")
print(sk_model.predict(X))
````

# 3. From Scratch

````python
# -------------------- From Scratch --------------------
class LinearRegressionScratch:
    def __init__(self, lr=0.01, epochs=5000):
        self.lr = lr
        self.epochs = epochs

    def fit(self, X, y):
        n_samples, n_features = X.shape

        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.epochs):

            # Prediction
            y_pred = X @ self.w + self.b

            # Gradients
            dw = (2 / n_samples) * X.T @ (y_pred - y)
            db = (2 / n_samples) * np.sum(y_pred - y)

            # Update
            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        return X @ self.w + self.b


my_model = LinearRegressionScratch()
my_model.fit(X, y)

print("\nScratch Prediction:")
print(my_model.predict(X))

print("\nWeights:", my_model.w)
print("Bias:", my_model.b)
````

# 4. Accuracy

````python
from sklearn.metrics import r2_score, mean_squared_error

# -------------------- Sklearn --------------------
sk_pred = sk_model.predict(X)

print("\n===== Sklearn =====")
print("R2 Score :", r2_score(y, sk_pred))
print("MSE      :", mean_squared_error(y, sk_pred))
print("RMSE     :", np.sqrt(mean_squared_error(y, sk_pred)))

# -------------------- Scratch --------------------
my_pred = my_model.predict(X)

print("\n===== Scratch =====")
print("R2 Score :", r2_score(y, my_pred))
print("MSE      :", mean_squared_error(y, my_pred))
print("RMSE     :", np.sqrt(mean_squared_error(y, my_pred)))
````

# 5. Linear Regression Gradient Derivation



## Step 1: Linear Model

$$
\hat{y}=XW+b
$$

---

## Step 2: Mean Squared Error (MSE) Loss

$$
J(W,b)
=
\frac1n
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
$$

---

# Derivation

We need

$$
\frac{\partial J}{\partial W}
$$

Using the chain rule,

$$
\frac{\partial J}{\partial W}
=
\frac{\partial J}{\partial \hat y}
\cdot
\frac{\partial \hat y}{\partial W}
$$

---

## Step 3: Derivative of MSE Loss

Differentiate with respect to $\hat y$

$$
J
=
\frac1n
(y-\hat y)^2
$$

Using

$$
\frac{d}{dx}(x^2)=2x
$$

we get

$$
\frac{\partial J}{\partial \hat y}
=
\frac1n
\cdot
2(y-\hat y)
\cdot
(-1)
$$

since

$$
\frac{\partial (y-\hat y)}{\partial \hat y}
=
-1
$$

Therefore

$$
\boxed{
\frac{\partial J}{\partial \hat y}
=
\frac{2}{n}
(\hat y-y)
}
$$

---

## Step 4: Derivative of Prediction

Since

$$
\hat y=XW+b
$$

$$
\frac{\partial \hat y}{\partial W}
=
X
$$

---

## Step 5: Final Gradient

Multiply both terms

$$
\frac{\partial J}{\partial W}
=
\frac{2}{n}
(\hat y-y)
X
$$

Writing in matrix form

$$
\boxed{
\frac{\partial J}{\partial W}
=
\frac{2}{n}
X^T(\hat y-y)
}
$$

---

## Step 6: Bias Gradient

Since

$$
\frac{\partial \hat y}{\partial b}=1
$$

we get

$$
\boxed{
\frac{\partial J}{\partial b}
=
\frac{2}{n}
\sum_{i=1}^{n}
(\hat y_i-y_i)
}
$$

---

## Step 7: Gradient Descent Update

Weights

$$
W
=
W-
\alpha
\frac{\partial J}{\partial W}
$$

Bias

$$
b
=
b-
\alpha
\frac{\partial J}{\partial b}
$$

---

# Final Result

$$
\boxed{
dw=
\frac{2}{n}
X^T(\hat y-y)
}
$$

$$
\boxed{
db=
\frac{2}{n}
\sum(\hat y-y)
}
$$
