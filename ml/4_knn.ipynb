# 1. Dataset

````python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# -------------------- Dataset --------------------

# X shape = (8,2)
X = np.array([
    [1,2],
    [2,1],
    [2,3],
    [3,2],
    [6,5],
    [7,6],
    [8,7],
    [9,8]
], dtype=float)

# y shape = (8,)
y = np.array([0,0,0,0,1,1,1,1])

print("X Shape:", X.shape)
print("y Shape:", y.shape)
````

# 2. Sklearn

````python
sk_model = KNeighborsClassifier(n_neighbors=3)

sk_model.fit(X, y)

sk_pred = sk_model.predict(X)

print("Prediction:")
print(sk_pred)

print("Accuracy:", sk_model.score(X,y))
````

# 3. From Scratch

````python
import numpy as np

class KNN:

    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):

        predictions = []

        for x in X:

            # Euclidean Distance
            distances = np.sqrt(np.sum((self.X_train - x)**2, axis=1))

            # k nearest neighbors
            k_idx = np.argsort(distances)[:self.k]

            k_labels = self.y_train[k_idx]

            # Majority Voting
            prediction = np.bincount(k_labels).argmax()

            predictions.append(prediction)

        return np.array(predictions)


my_model = KNN(k=3)

my_model.fit(X,y)

my_pred = my_model.predict(X)

print("Prediction:")
print(my_pred)
````

# 4. Accuracy

````python
from sklearn.metrics import accuracy_score

print("\n===== Sklearn =====")
print("Accuracy :", accuracy_score(y, sk_pred))

print("\n===== Scratch =====")
print("Accuracy :", accuracy_score(y, my_pred))
````

# 5. K-Nearest Neighbors (KNN)

## Step 1: Store Training Data

Unlike Linear or Logistic Regression,

KNN performs **no training**.

It simply stores

$$
X_{train}, y_{train}
$$

---

## Step 2: Compute Distance

For every test sample,

calculate Euclidean distance to every training sample.

$$
d(x_i,x_j)
=
\sqrt{
\sum_{k=1}^{m}
(x_{ik}-x_{jk})^2
}
$$

where

- $m$ = number of features

Code

```python
distances = np.sqrt(np.sum((self.X_train-x)**2,axis=1))
```

---

## Step 3: Sort Distances

Sort distances in ascending order

$$
d_1<d_2<d_3<\cdots
$$

Code

```python
k_idx=np.argsort(distances)
```

---

## Step 4: Select k Nearest Neighbors

Take first

$$
k
$$

neighbors.

Code

```python
k_idx=np.argsort(distances)[:self.k]
```

---

## Step 5: Majority Voting

Choose the class occurring most frequently.

$$
\hat y
=
\operatorname{mode}
(y_1,y_2,\ldots,y_k)
$$

Code

```python
prediction=np.bincount(k_labels).argmax()
```

---

## Final Prediction

Repeat for every test sample.
