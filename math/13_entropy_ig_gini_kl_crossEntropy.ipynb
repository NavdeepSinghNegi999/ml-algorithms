# 1. Entropy

## What is Entropy?

Entropy measures the **uncertainty or impurity** in a set of data.

In a classification problem:

- If all samples belong to one class → entropy is **0**
- If classes are evenly mixed → entropy is **maximum**

For a classification problem with classes $1,\ldots,C$:

$$
\boxed{
H(S)
=
-\sum_{i=1}^{C}p_i\log_2(p_i)
}
$$

where $p_i$ is the proportion of class $i$.

---

## Why Do We Need Entropy?

Suppose a node contains:

$$
[10\ Cats,\ 0\ Dogs]
$$

There is no uncertainty.

We already know that a random sample from this node is a Cat.

Therefore:

$$
H=0
$$

Now suppose:

$$
[5\ Cats,\ 5\ Dogs]
$$

We don't know which class a random sample belongs to.

There is maximum uncertainty.

Therefore entropy is maximum.

So:

$$
\boxed{
More\ mixed
\rightarrow
Higher\ Entropy
}
$$

$$
\boxed{
More\ pure
\rightarrow
Lower\ Entropy
}
$$

---

# 2. Entropy Calculation

Suppose a node contains:

$$
6\ Positive,\quad4\ Negative
$$

Total:

$$
10
$$

Therefore:

$$
p_{positive}=0.6
$$

$$
p_{negative}=0.4
$$

Entropy:

$$
H(S)
=
-[0.6\log_2(0.6)+0.4\log_2(0.4)]
$$

Using:

$$
\log_2(0.6)\approx-0.737
$$

and:

$$
\log_2(0.4)\approx-1.322
$$

Therefore:

$$
H(S)
=
-[0.6(-0.737)+0.4(-1.322)]
$$

$$
=
-[ -0.4422-0.5288]
$$

$$
\boxed{
H(S)\approx0.971
}
$$

---

# 3. Important Entropy Cases

For binary classification:

### Pure Node

$$
p=1,\quad1-p=0
$$

$$
H=0
$$

### Completely Mixed Node

$$
p=0.5,\quad1-p=0.5
$$

$$
H=1
$$

Therefore, for binary classification:

$$
\boxed{
0\leq H\leq1
}
$$

---

# 4. Why Does Entropy Use a Logarithm?

The logarithm gives more importance to uncertainty and converts multiplication of probabilities into addition.

The negative sign is used because:

$$
0<p<1
$$

gives:

$$
\log(p)<0
$$

Without the negative sign, entropy would be negative.

Therefore:

$$
\boxed{
H=-\sum p_i\log(p_i)
}
$$

---

# 5. Information Gain

## What?

Information Gain measures **how much uncertainty is reduced after splitting a dataset**.

Decision trees use Information Gain to choose a good split.

The basic idea is:

$$
\boxed{
Information\ Gain
=
Parent\ Entropy
-
Weighted\ Child\ Entropy
}
$$

Formula:

$$
\boxed{
IG(S,A)
=
H(S)
-
\sum_{v}
\frac{|S_v|}{|S|}
H(S_v)
}
$$

where:

- $S$ = parent dataset
- $A$ = feature used for splitting
- $S_v$ = child dataset
- $\frac{|S_v|}{|S|}$ = weight of that child

---

# 6. Why Do We Use Weighted Entropy?

Suppose a split creates two children:

- Child 1 → 90 samples
- Child 2 → 10 samples

We shouldn't treat both children equally.

Child 1 contains much more data.

Therefore:

$$
Weighted\ Entropy
=
\frac{90}{100}H(Child_1)
+
\frac{10}{100}H(Child_2)
$$

This gives larger children more influence.

---

# 7. Information Gain Example

Suppose the parent contains:

$$
5\ Positive,\quad5\ Negative
$$

Therefore:

$$
H(parent)=1
$$

Suppose a split creates:

### Left child

$$
4\ Positive,\quad0\ Negative
$$

This is pure:

$$
H(left)=0
$$

### Right child

$$
1\ Positive,\quad5\ Negative
$$

Its entropy is approximately:

$$
H(right)\approx0.65
$$

Weighted child entropy:

$$
\frac{4}{10}(0)
+
\frac{6}{10}(0.65)
$$

$$
=0.39
$$

Therefore:

$$
IG
=
1-0.39
$$

$$
\boxed{
IG\approx0.61
}
$$

Higher Information Gain means the split reduces uncertainty more.

---

# 8. How Does a Decision Tree Choose a Split?

Suppose we have three possible features:

| Feature | Information Gain |
|---|---:|
| Age | 0.20 |
| Income | 0.45 |
| Experience | 0.30 |

The tree chooses:

$$
\boxed{
Income
}
$$

because:

$$
0.45
$$

is the highest Information Gain.

So:

$$
\boxed{
Choose\ split
=
Maximum\ Information\ Gain
}
$$

# 9. Gini Impurity

## What?

Gini Impurity also measures how **impure or mixed** a node is.

Formula:

$$
\boxed{
Gini
=
1-\sum_{i=1}^{C}p_i^2
}
$$

For binary classification:

$$
\boxed{
Gini
=
1-(p^2+(1-p)^2)
}
$$

---

# 10. Gini Calculation

Suppose:

$$
6\ Positive,\quad4\ Negative
$$

Then:

$$
p_1=0.6
$$

$$
p_2=0.4
$$

Therefore:

$$
Gini
=
1-(0.6^2+0.4^2)
$$

$$
=
1-(0.36+0.16)
$$

$$
=
1-0.52
$$

$$
\boxed{
Gini=0.48
}
$$

---

# 11. Entropy vs Gini

Both measure impurity.

| | Entropy | Gini |
|---|---|---|
| Formula | $-\sum p_i\log_2p_i$ | $1-\sum p_i^2$ |
| Pure node | $0$ | $0$ |
| Binary maximum | $1$ | $0.5$ |
| Calculation | Slightly more expensive | Simpler |
| Used in | Decision Trees | Decision Trees |
| Goal | Lower impurity | Lower impurity |

For a binary problem:

$$
\boxed{
Entropy_{max}=1
}
$$

while:

$$
\boxed{
Gini_{max}=0.5
}
$$

The maximum values are different because the formulas are different.

---

# 12. Which One Should We Use?

Usually, both produce similar tree splits.

Gini is often preferred because it is slightly simpler computationally.

Entropy is useful when you want to think in terms of:

> Information and uncertainty.

Gini is useful when you want to think in terms of:

> Impurity.

There is no universal rule that one is always better.

# 13. KL Divergence

## What?

KL Divergence measures how different one probability distribution is from another.

It is also called:

**Kullback-Leibler Divergence**

For discrete distributions:

$$
\boxed{
D_{KL}(P\parallel Q)
=
\sum_xP(x)
\log
\frac{P(x)}{Q(x)}
}
$$

It measures how much information is lost when we use $Q$ to represent $P$.

---

# 14. KL Divergence Intuition

Suppose the true distribution is:

$$
P=[0.8,0.2]
$$

and our estimated distribution is:

$$
Q=[0.6,0.4]
$$

KL divergence asks:

> How different is $Q$ from the true distribution $P$?

If:

$$
P=Q
$$

then:

$$
\boxed{
D_{KL}(P\parallel Q)=0
}
$$

KL divergence can never be negative:

$$
\boxed{
D_{KL}(P\parallel Q)\geq0
}
$$

---

# 15. Why Is KL Divergence Not Symmetric?

This is a very important interview question.

Generally:

$$
\boxed{
D_{KL}(P\parallel Q)
\neq
D_{KL}(Q\parallel P)
}
$$

For example:

$$
D_{KL}(P\parallel Q)
$$

asks:

> How different is $Q$ from $P$ when $P$ is treated as the reference?

While:

$$
D_{KL}(Q\parallel P)
$$

asks:

> How different is $P$ from $Q$ when $Q$ is treated as the reference?

The reference distribution changes.

Therefore the values can be different.

So KL divergence is **not a true distance metric**.

# 16. Cross Entropy

## What?

Cross Entropy measures how well a predicted probability distribution matches the true distribution.

Formula:

$$
\boxed{
H(P,Q)
=
-\sum_xP(x)\log Q(x)
}
$$

where:

- $P$ = true distribution
- $Q$ = predicted distribution

---

# 17. Binary Cross Entropy

For binary classification:

$$
y\in\{0,1\}
$$

and predicted probability:

$$
\hat y=P(y=1|x)
$$

Binary Cross Entropy is:

$$
\boxed{
BCE
=
-
[
y\log(\hat y)
+
(1-y)\log(1-\hat y)
]
}
$$

### If $y=1$

Then:

$$
BCE=-\log(\hat y)
$$

If:

$$
\hat y=0.99
$$

then loss is very small.

If:

$$
\hat y=0.01
$$

then loss is very large.

This strongly penalizes **confident wrong predictions**.

---

# 18. Multiclass Cross Entropy

Suppose there are $C$ classes.

The true label is represented using one-hot encoding:

$$
y=[1,0,0]
$$

and the model predicts:

$$
\hat y=[0.7,0.2,0.1]
$$

Cross entropy:

$$
\boxed{
CE
=
-\sum_{i=1}^{C}
y_i\log(\hat y_i)
}
$$

For this example:

$$
CE
=
-[1\log(0.7)+0\log(0.2)+0\log(0.1)]
$$

$$
=
-\log(0.7)
$$

---

# 19. Cross Entropy vs KL Divergence

There is an important relationship:

$$
\boxed{
H(P,Q)
=
H(P)
+
D_{KL}(P\parallel Q)
}
$$

where:

$$
H(P)
=
-\sum_xP(x)\log P(x)
$$

is the entropy of the true distribution.

Therefore:

$$
\boxed{
Cross\ Entropy
=
Entropy
+
KL\ Divergence
}
$$

If $P$ is fixed, then $H(P)$ is constant.

Therefore minimizing cross entropy with respect to $Q$ is equivalent to minimizing:

$$
D_{KL}(P\parallel Q)
$$

This is why cross entropy is commonly used as a classification loss.

---

# 20. Why Cross Entropy Penalizes Confident Wrong Predictions

Suppose the true class is:

$$
y=1
$$

### Prediction 1

$$
\hat y=0.9
$$

Loss:

$$
-\log(0.9)\approx0.105
$$

Small loss.

### Prediction 2

$$
\hat y=0.1
$$

Loss:

$$
-\log(0.1)\approx2.303
$$

Large loss.

Therefore:

$$
\boxed{
Confident\ wrong\ prediction
\rightarrow
Large\ Loss
}
$$

This is one reason cross entropy works well for classification.

# 23. Important Interview Questions

### What is entropy?

A measure of uncertainty or impurity in a probability distribution or decision-tree node.

### When is entropy zero?

When the node is completely pure:

$$
p_i=1
$$

for one class.

### When is entropy maximum?

When all classes are equally probable.

### What is Information Gain?

The reduction in entropy caused by a split.

$$
IG
=
H(parent)-H(children)
$$

with the children weighted by their sizes.

### How does a decision tree choose a split?

It generally chooses the split with the highest Information Gain when using entropy.

### What is Gini Impurity?

A measure of the probability that a randomly selected sample would be incorrectly classified if it were labeled according to the class distribution of the node.

$$
Gini=1-\sum p_i^2
$$

### Entropy vs Gini?

Both measure impurity. Gini is computationally simpler, while entropy is based on information/uncertainty.

### What is KL Divergence?

A measure of how one probability distribution differs from another.

### Why is KL divergence not symmetric?

Because:

$$
D_{KL}(P\parallel Q)
\neq
D_{KL}(Q\parallel P)
$$

The first distribution acts as the reference, so reversing them changes the calculation.

### What is cross entropy?

A measure of the difference between the true distribution and predicted distribution, commonly used as a classification loss.

### Why is cross entropy used in classification?

It strongly penalizes incorrect confident predictions and naturally arises from maximum likelihood for classification models.

### What is the relationship between cross entropy and KL divergence?

$$
H(P,Q)=H(P)+D_{KL}(P\parallel Q)
$$

Since $H(P)$ is constant for a fixed true distribution, minimizing cross entropy is equivalent to minimizing KL divergence.
