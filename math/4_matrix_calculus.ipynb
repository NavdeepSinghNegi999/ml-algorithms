# 1. What is Matrix Calculus?

Matrix calculus is calculus applied to **vectors and matrices**.

In normal calculus:

$$
y=f(x)
$$

we calculate:

$$
\frac{dy}{dx}
$$

In ML, we usually have many parameters:

$$
\mathbf{w}=
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_n
\end{bmatrix}
$$

and a loss:

$$
L(\mathbf{w})
$$

So we need to calculate how the loss changes with respect to **every parameter**.

This produces a gradient:

$$
\nabla_{\mathbf{w}}L
=
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots\\
\frac{\partial L}{\partial w_n}
\end{bmatrix}
$$

---

# 2. Why Do We Need Matrix Calculus?

ML models can contain thousands or millions of parameters.

Instead of calculating every derivative separately by hand, matrix calculus gives us compact expressions for the gradients.

The main purpose is:

$$
\boxed{
\text{Parameters}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Parameter Update}
}
$$

This is fundamental to gradient-based optimization and neural-network training.

---

# 3. Basic Matrix Derivative

Suppose:

$$
\mathbf{y}=A\mathbf{x}
$$

where:

$$
A\in\mathbb{R}^{m\times n},
\qquad
\mathbf{x}\in\mathbb{R}^{n}
$$

Each element of $\mathbf{y}$ is:

$$
y_i=\sum_{j=1}^{n}A_{ij}x_j
$$

Therefore:

$$
\frac{\partial y_i}{\partial x_j}=A_{ij}
$$

So the derivative of:

$$
A\mathbf{x}
$$

with respect to $\mathbf{x}$ is related directly to $A$.

The important idea is:

> Matrix calculus allows us to differentiate vector/matrix expressions while keeping track of their dimensions.

---

# 4. Gradient of a Simple Quadratic

Consider:

$$
f(\mathbf{x})=\mathbf{x}^T\mathbf{x}
$$

For:

$$
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}
$$

we have:

$$
f(\mathbf{x})
=
x_1^2+x_2^2
$$

Differentiate:

$$
\frac{\partial f}{\partial x_1}=2x_1
$$

$$
\frac{\partial f}{\partial x_2}=2x_2
$$

Therefore:

$$
\boxed{
\nabla_{\mathbf{x}}f=2\mathbf{x}
}
$$

This simple result appears repeatedly in ML.

# 5. Gradient of Mean Squared Error

Suppose a linear model is:

$$
\hat{\mathbf{y}}=X\mathbf{w}
$$

where:

- $X$ = feature matrix
- $\mathbf{w}$ = model weights
- $\hat{\mathbf{y}}$ = predictions

The error is:

$$
\mathbf{e}=X\mathbf{w}-\mathbf{y}
$$

MSE can be written as:

$$
L(\mathbf{w})
=
\frac{1}{n}
\|X\mathbf{w}-\mathbf{y}\|^2
$$

Using:

$$
\|\mathbf{e}\|^2=\mathbf{e}^T\mathbf{e}
$$

we get:

$$
L(\mathbf{w})
=
\frac{1}{n}
(X\mathbf{w}-\mathbf{y})^T
(X\mathbf{w}-\mathbf{y})
$$

The gradient is:

$$
\boxed{
\nabla_{\mathbf{w}}L
=
\frac{2}{n}X^T(X\mathbf{w}-\mathbf{y})
}
$$

This is one of the most important matrix-calculus formulas for ML.

---

# 6. Why Does $X^T$ Appear?

This is an important interview question.

We have:

$$
X\mathbf{w}-\mathbf{y}
$$

which gives the prediction error for every sample.

To determine how each feature/weight contributed to the error, we multiply by:

$$
X^T
$$

So:

$$
X^T(X\mathbf{w}-\mathbf{y})
$$

combines the errors with the corresponding feature values.

The result has the same shape as $\mathbf{w}$:

$$
X^T(X\mathbf{w}-\mathbf{y})
\in\mathbb{R}^{d}
$$

Therefore, it gives one gradient value for each weight.

---

# 7. MSE Gradient — Simple Derivation

Start with:

$$
L=
\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

For a linear model:

$$
\hat{y}_i=\mathbf{x}_i^T\mathbf{w}
$$

So:

$$
L=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\mathbf{x}_i^T\mathbf{w})^2
$$

Differentiate using the chain rule:

$$
\frac{\partial L}{\partial\mathbf{w}}
=
\frac{2}{n}
\sum_{i=1}^{n}
(\mathbf{x}_i^T\mathbf{w}-y_i)\mathbf{x}_i
$$

Writing all samples together gives:

$$
\boxed{
\nabla_{\mathbf{w}}L
=
\frac{2}{n}X^T(X\mathbf{w}-\mathbf{y})
}
$$

So matrix calculus is essentially allowing us to write many individual derivatives compactly.

---

# 8. Gradient of Logistic Loss

Logistic regression predicts:

$$
p=\sigma(z)
$$

where:

$$
z=\mathbf{x}^T\mathbf{w}+b
$$

and the sigmoid function is:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

For binary classification, the loss for one sample is:

$$
L=
-\left[
y\log(p)+(1-y)\log(1-p)
\right]
$$

The important result is:

$$
\boxed{
\frac{\partial L}{\partial z}=p-y
}
$$

Since:

$$
z=\mathbf{x}^T\mathbf{w}+b
$$

using the chain rule:

$$
\boxed{
\frac{\partial L}{\partial\mathbf{w}}
=
(p-y)\mathbf{x}
}
$$

For all samples:

$$
\boxed{
\nabla_{\mathbf{w}}L
=
\frac{1}{n}X^T(\mathbf{p}-\mathbf{y})
}
$$

and for the bias:

$$
\boxed{
\frac{\partial L}{\partial b}
=
\frac{1}{n}\sum_{i=1}^{n}(p_i-y_i)
}
$$

---

# 9. Why Is Logistic Regression Gradient So Important?

It demonstrates the complete ML chain:

$$
\mathbf{x}
\rightarrow
z=\mathbf{x}^T\mathbf{w}+b
\rightarrow
p=\sigma(z)
\rightarrow
Loss
$$

Then backpropagation applies the chain rule:

$$
\frac{\partial L}{\partial\mathbf{w}}
=
\frac{\partial L}{\partial p}
\frac{\partial p}{\partial z}
\frac{\partial z}{\partial\mathbf{w}}
$$

After simplification:

$$
\boxed{
\frac{\partial L}{\partial\mathbf{w}}
=
(p-y)\mathbf{x}
}
$$

This is a very useful formula to understand rather than memorize.

---

# 10. Matrix Dimensions — Important for Interviews

Suppose:

$$
X\in\mathbb{R}^{n\times d}
$$

where:

- $n$ = samples
- $d$ = features

and:

$$
\mathbf{w}\in\mathbb{R}^{d}
$$

Then:

$$
X\mathbf{w}\in\mathbb{R}^{n}
$$

and:

$$
X^T\in\mathbb{R}^{d\times n}
$$

Therefore:

$$
X^T(X\mathbf{w}-\mathbf{y})
\in\mathbb{R}^{d}
$$

which matches the shape of:

$$
\nabla_{\mathbf{w}}L
$$

Always check dimensions when doing matrix calculus.

---

# 11. Matrix Calculus and Gradient Descent

Once we calculate the gradient:

$$
\nabla_{\mathbf{w}}L
$$

we update the weights:

$$
\boxed{
\mathbf{w}_{new}
=
\mathbf{w}_{old}
-
\eta\nabla_{\mathbf{w}}L
}
$$

So the complete process is:

$$
\text{Forward Pass}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Update}
$$

Matrix calculus provides the mathematical machinery for calculating the gradient.

---

# 12. Common Mistakes

### Mistake 1: Forgetting the transpose

For MSE:

$$
\nabla_{\mathbf{w}}L
=
\frac{2}{n}X^T(X\mathbf{w}-\mathbf{y})
$$

not:

$$
X(X\mathbf{w}-\mathbf{y})
$$

because the dimensions would generally not match.

### Mistake 2: Confusing gradient with loss

The loss is a scalar:

$$
L\in\mathbb{R}
$$

The gradient with respect to $d$ weights is:

$$
\nabla_{\mathbf{w}}L\in\mathbb{R}^{d}
$$

### Mistake 3: Forgetting the chain rule

For nested functions, derivatives must propagate through every operation.

# 13. Important Interview Questions

### What is matrix calculus?

It is the extension of differentiation to vectors and matrices, allowing us to efficiently calculate gradients of functions involving many variables.

### Why is matrix calculus important in ML?

Because ML models contain many parameters, and optimization requires gradients with respect to those parameters.

### What is the gradient of MSE for a linear model?

$$
\boxed{
\nabla_{\mathbf{w}}L
=
\frac{2}{n}X^T(X\mathbf{w}-\mathbf{y})
}
$$

### What is the gradient of logistic loss?

$$
\boxed{
\nabla_{\mathbf{w}}L
=
\frac{1}{n}X^T(\mathbf{p}-\mathbf{y})
}
$$

### Why does $X^T$ appear?

It combines the error from each sample with the corresponding feature values and produces one gradient value for each weight.

### Why should we check matrix dimensions?

Because matrix operations are only valid when their dimensions are compatible, and the final gradient must have the same shape as the parameters being optimized.
