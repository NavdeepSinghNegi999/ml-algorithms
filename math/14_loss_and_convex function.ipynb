# 1. Loss Function

A loss function tells us **how wrong a model's prediction is**.

For actual value $y$ and prediction $\hat{y}$:

$$
L(y,\hat{y})
$$

A smaller loss means the prediction is closer to what we want.

The model tries to minimize the loss:

$$
\boxed{
\min_{\theta} L(\theta)
}
$$

where $\theta$ represents the model parameters.

### Loss vs Cost

Loss usually refers to one training example:

$$
L(y_i,\hat{y}_i)
$$

Cost is usually the average loss over the complete dataset:

$$
\boxed{
J(\theta)
=
\frac{1}{n}
\sum_{i=1}^{n}
L(y_i,\hat{y}_i)
}
$$

# 2. Mean Squared Error (MSE)

MSE is mainly used for regression.

$$
\boxed{
MSE=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

## Why do we square the error?

Suppose:

$$
y=10,\qquad\hat{y}=7
$$

Error:

$$
10-7=3
$$

Another prediction:

$$
y=10,\qquad\hat{y}=13
$$

Error:

$$
10-13=-3
$$

If we simply average errors:

$$
\frac{3+(-3)}{2}=0
$$

The positive and negative errors cancel.

So we square them:

$$
3^2=9
$$

$$
(-3)^2=9
$$

Now both errors contribute positively.

## Why does MSE punish large errors strongly?

Because the error is squared.

For example:

$$
2^2=4
$$

but:

$$
10^2=100
$$

Therefore:

$$
\boxed{
MSE\text{ is highly sensitive to large errors and outliers}
}
$$

## Advantages

- Smooth and differentiable
- Easy to optimize
- Strongly penalizes large errors
- Common for regression

## Disadvantages

- Sensitive to outliers
- Large errors can dominate the loss

---

# 3. MAE

MAE stands for **Mean Absolute Error**.

$$
\boxed{
MAE=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
}
$$

Instead of squaring the error, we take its absolute value.

Example:

$$
y=[2,4,6]
$$

$$
\hat{y}=[3,5,5]
$$

Errors:

$$
[-1,-1,1]
$$

Absolute errors:

$$
[1,1,1]
$$

Therefore:

$$
MAE=
\frac{1+1+1}{3}
$$

$$
\boxed{
MAE=1
}
$$

## Why is MAE more robust to outliers?

Suppose the error is:

$$
10
$$

MSE contribution:

$$
10^2=100
$$

MAE contribution:

$$
|10|=10
$$

Therefore MSE gives the large error a much bigger penalty.

---

# 4. MSE vs MAE

| | MSE | MAE |
|---|---|---|
| Error | Squared | Absolute |
| Outlier sensitivity | High | Lower |
| Large errors | Strongly penalized | Linear penalty |
| Smooth | Yes | Not differentiable at $0$ |
| Common use | Regression | Regression |

### When should we use MSE?

Use MSE when large errors should be heavily penalized.

### When should we use MAE?

Use MAE when you want more robustness to outliers.

---

# 5. Hinge Loss

Hinge loss is mainly associated with **Support Vector Machines (SVM)**.

For binary classification:

$$
y\in\{-1,+1\}
$$

Hinge loss is:

$$
\boxed{
L=\max(0,1-y\hat{y})
}
$$

where $\hat{y}$ is the model's score.

## Why $y\hat{y}$?

If the prediction is correct, $y$ and $\hat{y}$ have the same sign:

$$
y\hat{y}>0
$$

If the prediction is incorrect:

$$
y\hat{y}<0
$$

SVM does not only want the correct class.

It also wants the prediction to have a sufficient **margin**.

The desired condition is:

$$
\boxed{
y\hat{y}\geq1
}
$$

---

# 6. Hinge Loss Examples

### Correct prediction with large margin

$$
y=1,\qquad\hat{y}=2
$$

$$
L=
\max(0,1-(1)(2))
$$

$$
=
\max(0,-1)
$$

$$
\boxed{
L=0
}
$$

### Correct prediction but small margin

$$
y=1,\qquad\hat{y}=0.5
$$

$$
L=
\max(0,1-0.5)
$$

$$
\boxed{
L=0.5
}
$$

### Wrong prediction

$$
y=1,\qquad\hat{y}=-2
$$

$$
L=
\max(0,1-(-2))
$$

$$
\boxed{
L=3
}
$$

Therefore:

$$
\boxed{
Hinge\ Loss
\rightarrow
Correctness+Margin
}
$$

---

# 7. Cross Entropy

Cross entropy is mainly used for classification.

For binary classification:

$$
y\in\{0,1\}
$$

and predicted probability:

$$
\hat{y}=P(y=1|x)
$$

Binary Cross Entropy is:

$$
\boxed{
BCE=
-
\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
}
$$

---

# 8. Why Does Cross Entropy Penalize Confident Wrong Predictions?

Suppose:

$$
y=1
$$

Then:

$$
BCE=-\log(\hat{y})
$$

If the model predicts:

$$
\hat{y}=0.99
$$

then:

$$
BCE=-\log(0.99)
\approx0.01
$$

Very small loss.

But if:

$$
\hat{y}=0.01
$$

then:

$$
BCE=-\log(0.01)
\approx4.61
$$

Very large loss.

Therefore:

$$
\boxed{
Confident\ wrong\ prediction
\rightarrow
Large\ Cross\ Entropy
}
$$

---

# 9. Multiclass Cross Entropy

For $C$ classes:

$$
\boxed{
CE=
-\sum_{i=1}^{C}
y_i\log(\hat{y}_i)
}
$$

Suppose the true class is class 1:

$$
y=[1,0,0]
$$

and the model predicts:

$$
\hat{y}=[0.7,0.2,0.1]
$$

Then:

$$
CE=
-[1\log(0.7)+0\log(0.2)+0\log(0.1)]
$$

Therefore:

$$
\boxed{
CE=-\log(0.7)
}
$$

Only the probability assigned to the correct class contributes because the other true-label values are zero.

---

# 10. Why Do We Use Cross Entropy in Classification?

For classification, the model usually predicts probabilities.

Cross entropy measures how well those predicted probabilities match the true class.

It also comes naturally from **Maximum Likelihood Estimation**.

For Bernoulli classification:

$$
Likelihood=
\prod_i
p_i^{y_i}(1-p_i)^{1-y_i}
$$

Taking negative log gives:

$$
-\log(Likelihood)
=
-\sum_i
[
y_i\log(p_i)
+
(1-y_i)\log(1-p_i)
]
$$

which is Binary Cross Entropy.

Therefore:

$$
\boxed{
MLE
\rightarrow
Negative\ Log\ Likelihood
\rightarrow
Cross\ Entropy
}
$$

---

# 11. Loss Function Comparison

| Loss | Main Use | Main Idea |
|---|---|---|
| MSE | Regression | Square the error |
| MAE | Regression | Absolute error |
| Hinge Loss | SVM classification | Correct class + margin |
| Cross Entropy | Classification | Compare predicted probabilities with true labels |

# 12. Convex Functions

A function is convex if it has a **bowl-like shape**.

Simple example:

$$
f(x)=x^2
$$

The important optimization property is:

$$
\boxed{
\text{Every local minimum is a global minimum}
}
$$

This makes convex optimization easier.

---

# 13. Why Is Convex Optimization Easier?

Consider a convex function.

There is no separate bad local minimum that is lower than the global solution.

So if an optimization algorithm reaches a local minimum:

$$
\boxed{
Local\ Minimum=Global\ Minimum
}
$$

We don't have to worry about getting stuck in a worse local minimum.

For a non-convex function, we can have:

- Multiple local minima
- Saddle points
- Local maxima

Therefore optimization can be much harder.

---

# 14. Mathematical Definition of Convexity

A function $f$ is convex if for any $x_1,x_2$ and:

$$
0\leq\lambda\leq1
$$

we have:

$$
\boxed{
f(\lambda x_1+(1-\lambda)x_2)
\leq
\lambda f(x_1)
+
(1-\lambda)f(x_2)
}
$$

In simple English:

> The function value between two points should be below or equal to the straight line connecting those two points.

---

# 15. Checking Convexity Using Derivatives

For a twice-differentiable one-dimensional function:

$$
\boxed{
f''(x)\geq0
}
$$

throughout the domain means the function is convex.

Example:

$$
f(x)=x^2
$$

First derivative:

$$
f'(x)=2x
$$

Second derivative:

$$
f''(x)=2
$$

Since:

$$
2>0
$$

the function is convex.

---

# 16. Convex vs Non-Convex

| Convex | Non-Convex |
|---|---|
| Bowl-like structure | More complicated surface |
| Local minimum is global | Local minima may differ |
| Easier optimization | Harder optimization |
| Easier theoretical guarantees | Fewer guarantees |

In many classical ML problems, the objective can be convex.

Neural-network optimization is generally non-convex.

Therefore:

$$
\boxed{
Classical\ ML
\rightarrow
Many\ convex\ objectives
}
$$

$$
\boxed{
Deep\ Learning
\rightarrow
Generally\ non\text{-}convex
}
$$

# 17. Important Interview Questions

### What is a loss function?

A function that measures how different the model's prediction is from the desired output.

### Why is MSE sensitive to outliers?

Because errors are squared:

$$
10^2=100
$$

### MSE vs MAE?

MSE squares errors and heavily penalizes large errors. MAE uses absolute errors and is more robust to outliers.

### What is hinge loss?

A loss mainly used with SVM that penalizes incorrect predictions and predictions that do not satisfy the desired margin.

$$
L=\max(0,1-y\hat{y})
$$

### Why use cross entropy for classification?

Because classification models predict probabilities, and cross entropy measures how well those probabilities match the true labels while strongly penalizing confident incorrect predictions.

### What is a convex function?

A function where the line between any two points lies above or on the function.

### Why is convex optimization easier?

Because:

$$
\boxed{
Local\ Minimum=Global\ Minimum
}
$$

### Is every ML loss convex?

No.

Many neural-network loss functions are non-convex with respect to their parameters.

# Optimization Terms

## 1. Local Minimum

A **local minimum** is a point where the function has the lowest value compared to the points **near it**.

In simple words:

> Lowest point in its neighborhood.

It does not have to be the lowest point of the entire function.

---

## 2. Global Minimum

A **global minimum** is the **lowest point of the entire function**.

In simple words:

> Lowest point everywhere.

Therefore:

$$
\boxed{
Global\ Minimum
\leq
Every\ Other\ Function\ Value
}
$$

---

## 3. Local Maximum

A **local maximum** is a point where the function has the highest value compared to the points **near it**.

In simple words:

> Highest point in its neighborhood.

---

## 4. Global Maximum

A **global maximum** is the **highest point of the entire function**.

In simple words:

> Highest point everywhere.

---

# 5. Convex Function

A convex function roughly looks like a **U-shaped bowl**.

Example:

$$
f(x)=x^2
$$

The important property for ML is:

$$
\boxed{
Every\ Local\ Minimum
=
Global\ Minimum
}
$$

This makes optimization easier because if we find a local minimum, we know it is also the global minimum.

For a twice-differentiable function:

$$
\boxed{
f''(x)\geq0
\Rightarrow
f(x)\text{ is convex}
}
$$

---

# 6. Concave Function

A concave function is roughly an **upside-down U**.

Example:

$$
f(x)=-x^2
$$

For a concave function:

$$
\boxed{
Every\ Local\ Maximum
=
Global\ Maximum
}
$$

For a twice-differentiable function:

$$
\boxed{
f''(x)\leq0
\Rightarrow
f(x)\text{ is concave}
}
$$

---

# 7. Saddle Point

A **saddle point** is a point that behaves like a minimum in one direction and a maximum in another direction.

Example:

$$
f(x,y)=x^2-y^2
$$

Along the $x$ direction:

$$
f(x,0)=x^2
$$

which has a minimum at $x=0$.

Along the $y$ direction:

$$
f(0,y)=-y^2
$$

which has a maximum at $y=0$.

Therefore $(0,0)$ is a saddle point.

$$
\boxed{
Saddle\ Point
\neq
Minimum
\neq
Maximum
}
$$

Saddle points are important in neural-network optimization.

---

# 8. Stationary Point / Critical Point

A stationary point is a point where the derivative is zero.

For one variable:

$$
\boxed{
f'(x)=0
}
$$

For multiple variables:

$$
\boxed{
\nabla f(x)=0
}
$$

However:

$$
\boxed{
Derivative=0
\not\Rightarrow
Minimum
}
$$

A stationary point can be:

- Local minimum
- Local maximum
- Saddle point

So a zero derivative only tells us that the function is **flat at that point**.

---

# 9. Gradient

For a function with multiple variables, we use the **gradient**.

For:

$$
f(x,y)
$$

the gradient is:

$$
\boxed{
\nabla f
=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
}
$$

The gradient points in the direction of **steepest increase**.

Therefore:

$$
\boxed{
-\nabla f
=
Direction\ of\ Steepest\ Decrease
}
$$

This is why Gradient Descent uses:

$$
\boxed{
\theta_{new}
=
\theta_{old}
-
\eta\nabla J(\theta)
}
$$

where:

- $\theta$ = model parameters
- $\eta$ = learning rate
- $\nabla J(\theta)$ = gradient of the loss

---

# 10. Convex vs Concave

| Term | Shape | Important Property |
|---|---|---|
| Convex | U-shaped | Local minimum = Global minimum |
| Concave | ∩-shaped | Local maximum = Global maximum |

---

# 11. Quick Comparison

| Term | Meaning |
|---|---|
| **Local Minimum** | Lowest point nearby |
| **Global Minimum** | Lowest point everywhere |
| **Local Maximum** | Highest point nearby |
| **Global Maximum** | Highest point everywhere |
| **Convex** | Bowl-shaped function |
| **Concave** | Upside-down bowl |
| **Saddle Point** | Neither minimum nor maximum |
| **Stationary Point** | Derivative/gradient is zero |
| **Gradient** | Direction of steepest increase |
