# 1. Probability

## What?

Probability measures how likely an event is to happen.

$$
0\leq P(A)\leq1
$$

- $P(A)=0$ → impossible
- $P(A)=1$ → certain
- $0<P(A)<1$ → possible

For equally likely outcomes:

$$
P(A)=\frac{\text{Favorable outcomes}}{\text{Total outcomes}}
$$

### Example

For a fair die:

$$
P(\text{rolling }6)=\frac{1}{6}
$$

---

# 2. Basic Probability Rules

## Complement

If $A$ is an event, then $A^c$ means "A does not happen".

$$
\boxed{
P(A^c)=1-P(A)
}
$$

## Union

$A\cup B$ means **A or B**.

$$
\boxed{
P(A\cup B)
=
P(A)+P(B)-P(A\cap B)
}
$$

We subtract $P(A\cap B)$ because the overlap was counted twice.

## Intersection

$A\cap B$ means **A and B happen together**.

For independent events:

$$
\boxed{
P(A\cap B)=P(A)P(B)
}
$$

---

# 3. Conditional Probability

## What?

Conditional probability tells us:

> What is the probability of $A$ given that we already know $B$ happened?

Notation:

$$
P(A|B)
$$

Read as:

> Probability of A given B.

## Formula

$$
\boxed{
P(A|B)=
\frac{P(A\cap B)}{P(B)}
}
$$

provided:

$$
P(B)>0
$$

## Why?

Knowing that $B$ happened changes our possible outcomes.

Instead of considering the entire sample space, we consider only the cases where $B$ happened.

### Example

Suppose:

- 100 students
- 60 study ML
- 30 study ML and Python

Probability that a student studies Python **given that they study ML**:

$$
P(Python|ML)
=
\frac{P(Python\cap ML)}
{P(ML)}
$$

$$
=
\frac{30/100}{60/100}
=
\frac{30}{60}
=
0.5
$$

Therefore:

$$
P(Python|ML)=0.5
$$

---

# 4. Multiplication Rule

Starting from conditional probability:

$$
P(A|B)=
\frac{P(A\cap B)}{P(B)}
$$

Multiply by $P(B)$:

$$
\boxed{
P(A\cap B)=P(A|B)P(B)
}
$$

Similarly:

$$
\boxed{
P(A\cap B)=P(B|A)P(A)
}
$$

This rule is extremely important for Bayes theorem.

---

# 5. Bayes Theorem

## What?

Bayes theorem allows us to **reverse conditional probabilities**.

It tells us:

$$
P(A|B)
$$

using:

$$
P(B|A)
$$

## Formula

$$
\boxed{
P(A|B)
=
\frac{P(B|A)P(A)}
{P(B)}
}
$$

Where:

- $P(A|B)$ = posterior
- $P(B|A)$ = likelihood
- $P(A)$ = prior
- $P(B)$ = evidence

## Why?

Sometimes we know:

$$
P(B|A)
$$

but want:

$$
P(A|B)
$$

Bayes theorem lets us reverse the direction.

---

# 6. Bayes Example

Suppose:

$$
P(D)=0.01
$$

1% of people have a disease.

A test has:

$$
P(+|D)=0.90
$$

90% sensitivity.

And:

$$
P(+|D^c)=0.05
$$

5% false-positive rate.

We want:

$$
P(D|+)
$$

Using Bayes:

$$
P(D|+)
=
\frac{P(+|D)P(D)}
{P(+)}
$$

First calculate $P(+)$:

$$
P(+)
=
P(+|D)P(D)
+
P(+|D^c)P(D^c)
$$

$$
=
(0.90)(0.01)
+
(0.05)(0.99)
$$

$$
=0.009+0.0495
$$

$$
=0.0585
$$

Therefore:

$$
P(D|+)
=
\frac{(0.90)(0.01)}
{0.0585}
$$

$$
\approx0.154
$$

So:

$$
\boxed{
P(D|+)\approx15.4\%
}
$$

Even though the test is 90% sensitive, a positive result does **not** mean there is a 90% probability of disease.

The low prior probability matters.

# 7. Independence

## What?

Two events are independent if knowing one event happened does not change the probability of the other.

Mathematically:

$$
\boxed{
P(A|B)=P(A)
}
$$

and:

$$
\boxed{
P(B|A)=P(B)
}
$$

Equivalent multiplication rule:

$$
\boxed{
P(A\cap B)=P(A)P(B)
}
$$

## Example

For two independent coin flips:

$$
P(H_1)=\frac12
$$

$$
P(H_2)=\frac12
$$

Therefore:

$$
P(H_1\cap H_2)
=
\frac12\times\frac12
=
\frac14
$$

---

# 8. Independent vs Mutually Exclusive

These are different concepts.

### Independent

One event does not affect the probability of the other.

$$
P(A\cap B)=P(A)P(B)
$$

### Mutually Exclusive

They cannot happen together.

$$
P(A\cap B)=0
$$

For example, when rolling one die:

$$
A=\text{roll 2}
$$

$$
B=\text{roll 5}
$$

They are mutually exclusive.

They are **not independent**, because:

$$
P(A\cap B)=0
$$

but:

$$
P(A)P(B)=\frac16\times\frac16=\frac1{36}
$$

which is not zero.

# 9. Expectation

## What?

Expectation is the **probability-weighted average** of possible outcomes.

For a discrete random variable:

$$
\boxed{
E[X]=\sum_x xP(X=x)
}
$$

## Why?

A normal mean treats every observation equally.

Expectation weights each possible outcome according to its probability.

### Example

Suppose:

$$
X=
\begin{cases}
10 & P=0.2\\
20 & P=0.5\\
30 & P=0.3
\end{cases}
$$

Then:

$$
E[X]
=
10(0.2)+20(0.5)+30(0.3)
$$

$$
=2+10+9
$$

$$
\boxed{
E[X]=21
}
$$

---

# 10. Expectation of a Function

Sometimes we want:

$$
E[g(X)]
$$

For a discrete variable:

$$
\boxed{
E[g(X)]
=
\sum_x g(x)P(X=x)
}
$$

For example:

$$
E[X^2]
=
\sum_x x^2P(X=x)
$$

This is important because variance can be written using expectation.

---

# 11. Variance Using Expectation

Variance is:

$$
Var(X)
=
E[(X-\mu)^2]
$$

It can also be written as:

$$
\boxed{
Var(X)=E[X^2]-(E[X])^2
}
$$

Since:

$$
E[X]=\mu
$$

we get:

$$
Var(X)=E[X^2]-\mu^2
$$

This form is frequently useful in probability and ML.

---

# 12. Continuous Random Variables

For a continuous random variable with probability density function $f(x)$:

$$
\boxed{
E[X]
=
\int_{-\infty}^{\infty}xf(x)\,dx
}
$$

Probability over an interval:

$$
\boxed{
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
}
$$

Variance:

$$
\boxed{
Var(X)
=
\int_{-\infty}^{\infty}
(x-\mu)^2f(x)\,dx
}
$$

This is where **integration connects to probability**.

# 13. Probability Connection to ML

## Naive Bayes

Uses:

$$
P(A|B)
=
\frac{P(B|A)P(A)}{P(B)}
$$

## Logistic Regression

Outputs a probability:

$$
P(Y=1|X)
$$

using the sigmoid function:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

## Generative Models

Probability distributions are used to model how data is generated.

## Bayesian ML

Uses:

$$
Posterior
\propto
Likelihood\times Prior
$$

or:

$$
P(\theta|D)
\propto
P(D|\theta)P(\theta)
$$

# 14. Important Interview Questions

### What is conditional probability?

The probability of an event given that another event has already occurred.

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

### What is Bayes theorem?

A method for calculating a reversed conditional probability.

$$
P(A|B)=
\frac{P(B|A)P(A)}{P(B)}
$$

### What does independence mean?

The occurrence of one event does not change the probability of the other.

$$
P(A\cap B)=P(A)P(B)
$$

### Are mutually exclusive events independent?

Generally no, unless one event has probability zero.

### What is expectation?

The probability-weighted average of possible outcomes.

$$
E[X]=\sum_xxP(X=x)
$$

### What is the difference between probability and conditional probability?

Probability considers the full sample space, while conditional probability restricts the sample space based on known information.

### Why is Bayes theorem important in ML?

It allows us to update our belief about a hypothesis or parameter after observing new evidence.

# 9. Expectation

Expectation means:

> **The probability-weighted average of possible outcomes.**

For a discrete random variable:

$$
\boxed{E[X]=\sum_x xP(X=x)}
$$

### Example

Suppose a fair die is rolled.

Possible values:

$$
X=1,2,3,4,5,6
$$

Each has probability:

$$
P(X=x)=\frac{1}{6}
$$

Therefore:

$$
E[X]
=1\frac{1}{6}+2\frac{1}{6}+3\frac{1}{6}+4\frac{1}{6}+5\frac{1}{6}+6\frac{1}{6}
$$

$$
=\frac{1+2+3+4+5+6}{6}
=\frac{21}{6}
=3.5
$$

So:

$$
\boxed{E[X]=3.5}
$$

**Important:** 3.5 doesn't have to be an actual outcome. It is the **long-run average**.

---

# 10. Expectation of a Function

Sometimes we want the expectation of a function of $X$.

For example:

$$
E[X^2]
$$

Here, the function is:

$$
g(X)=X^2
$$

The formula is:

$$
\boxed{E[g(X)]=\sum_x g(x)P(X=x)}
$$

Therefore:

$$
E[X^2]=\sum_x x^2P(X=x)
$$

### Same die example

$$
E[X^2]
=
1^2\frac{1}{6}
+2^2\frac{1}{6}
+3^2\frac{1}{6}
+4^2\frac{1}{6}
+5^2\frac{1}{6}
+6^2\frac{1}{6}
$$

$$
=\frac{1+4+9+16+25+36}{6}
$$

$$
=\frac{91}{6}
\approx 15.17
$$

So:

$$
\boxed{E[X^2]\approx15.17}
$$

**Remember:** In $E[X^2]$, **square first, then take the expectation**.

---

# 11. Variance Using Expectation

Variance tells us:

> **How much the values are spread around their mean.**

The basic formula is:

$$
\boxed{Var(X)=E[(X-E[X])^2]}
$$

Meaning:

1. Find the mean.
2. Find the difference from the mean.
3. Square the difference.
4. Take the expectation.

There is also a useful shortcut:

$$
\boxed{Var(X)=E[X^2]-(E[X])^2}
$$

For our die:

$$
E[X]=3.5
$$

and:

$$
E[X^2]\approx15.17
$$

Therefore:

$$
Var(X)=15.17-(3.5)^2
$$

$$
=15.17-12.25
$$

$$
\boxed{Var(X)\approx2.92}
$$

---

# 12. Variance Using the Statistical Formula

We can calculate the **same variance** using the normal statistical formula:

$$
\boxed{Var(X)=\frac{\sum (x-\mu)^2}{N}}
$$

For our die:

$$
\mu=3.5,\qquad N=6
$$

Therefore:

$$
Var(X)=
\frac{
(1-3.5)^2+
(2-3.5)^2+
(3-3.5)^2+
(4-3.5)^2+
(5-3.5)^2+
(6-3.5)^2
}{6}
$$

$$
=
\frac{
6.25+2.25+0.25+0.25+2.25+6.25
}{6}
$$

$$
=\frac{17.5}{6}
$$

$$
\boxed{Var(X)=2.9167}
$$

And using expectation:

$$
\boxed{Var(X)=E[X^2]-(E[X])^2=2.9167}
$$

### Both give the same result

$$
\boxed{
\text{Statistical Variance}
=
\text{Expectation-based Variance}
=
2.9167
}
$$

The formulas are simply **two different ways of calculating the same variance**.

### Easy way to remember

$$
\boxed{E[X]=\text{mean}}
$$

$$
\boxed{E[X^2]=\text{square first, then average}}
$$

$$
\boxed{Var(X)=E[X^2]-(E[X])^2}
$$

And:

$$
\boxed{\text{Variance}=\text{average squared distance from the mean}}
$$
