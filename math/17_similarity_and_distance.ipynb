# 1. Similarity vs Distance

Both are used to compare two data points, vectors, or objects.

### Similarity

Similarity tells us:

> **How similar are two objects?**

Higher value usually means:

$$
\boxed{
More\ Similar
}
$$

### Distance

Distance tells us:

> **How far apart are two objects?**

Lower value usually means:

$$
\boxed{
More\ Similar
}
$$

So the basic idea is:

$$
\boxed{
Similarity\uparrow
\Rightarrow
More\ Similar
}
$$

$$
\boxed{
Distance\downarrow
\Rightarrow
More\ Similar
}
$$

---

# 2. Why Do We Need Them in ML?

Many ML algorithms need to know whether two data points are close or similar.

Examples:

- K-Nearest Neighbors
- K-Means
- Clustering
- Recommendation Systems
- NLP
- Semantic Search
- Embeddings

For example, in KNN:

> Find the training points closest to the new point.

Therefore we need a **distance metric**.

For text embeddings:

> Find the documents with similar meaning.

Therefore we often use **cosine similarity**.

---

# 3. Euclidean Distance

Euclidean distance is the normal **straight-line distance** between two points.

For two vectors:

$$
A=(A_1,A_2,\ldots,A_n)
$$

and:

$$
B=(B_1,B_2,\ldots,B_n)
$$

the formula is:

$$
\boxed{
d(A,B)
=
\sqrt{
\sum_{i=1}^{n}(A_i-B_i)^2
}
}
$$

---

# 4. Why Does Euclidean Distance Have This Formula?

For two dimensions, suppose:

$$
A=(1,2)
$$

and:

$$
B=(4,6)
$$

Difference in $x$:

$$
4-1=3
$$

Difference in $y$:

$$
6-2=4
$$

These form a right triangle.

Using Pythagoras:

$$
d^2=3^2+4^2
$$

Therefore:

$$
d=\sqrt{3^2+4^2}
$$

$$
=\sqrt{25}
$$

$$
\boxed{
d=5
}
$$

The $n$-dimensional formula is simply the same idea extended to more dimensions.

---

# 5. Euclidean Distance Example

Suppose:

$$
A=[1,2,3]
$$

$$
B=[4,6,3]
$$

Then:

$$
d(A,B)
=
\sqrt{
(1-4)^2+
(2-6)^2+
(3-3)^2
}
$$

$$
=
\sqrt{
9+16+0
}
$$

$$
=\sqrt{25}
$$

$$
\boxed{
d(A,B)=5
}
$$

---

# 6. Important Property of Euclidean Distance

If:

$$
A=B
$$

then:

$$
\boxed{
d(A,B)=0
}
$$

As the points become farther apart:

$$
d(A,B)\uparrow
$$

So:

$$
\boxed{
Smaller\ Euclidean\ Distance
\rightarrow
More\ Similar\ Points
}
$$

---

# 7. Manhattan Distance

Manhattan distance adds the **absolute differences** between coordinates.

$$
\boxed{
d(A,B)
=
\sum_{i=1}^{n}|A_i-B_i|
}
$$

---

# 8. Why Is It Called Manhattan Distance?

Imagine you are in a city with a grid of streets.

You cannot travel diagonally through buildings.

You have to travel:

- Left/right
- Up/down

So you add the distance traveled along each direction.

For:

$$
A=(1,2)
$$

and:

$$
B=(4,6)
$$

we get:

$$
d(A,B)
=
|1-4|+|2-6|
$$

$$
=3+4
$$

$$
\boxed{
d(A,B)=7
}
$$

---

# 9. Euclidean vs Manhattan

For the same points:

$$
A=(1,2)
$$

$$
B=(4,6)
$$

Euclidean:

$$
\boxed{
d_E=5
}
$$

Manhattan:

$$
\boxed{
d_M=7
}
$$

Why?

Euclidean allows the straight-line path.

Manhattan adds movement along each dimension separately.

---

# 10. Minkowski Distance

Minkowski distance is a **general form** of several distance metrics.

$$
\boxed{
d(A,B)
=
\left(
\sum_{i=1}^{n}
|A_i-B_i|^p
\right)^{1/p}
}
$$

The value of $p$ determines the type of distance.

### When $p=1$

$$
d(A,B)
=
\sum_i|A_i-B_i|
$$

Therefore:

$$
\boxed{
p=1
\Rightarrow
Manhattan
}
$$

### When $p=2$

$$
d(A,B)
=
\sqrt{\sum_i(A_i-B_i)^2}
$$

Therefore:

$$
\boxed{
p=2
\Rightarrow
Euclidean
}
$$

So:

$$
\boxed{
Minkowski
\rightarrow
Generalized\ Distance
}
$$

---

# 11. Why Does Distance Scaling Matter?

Suppose we have two features:

$$
Age=25
$$

and:

$$
Salary=100000
$$

Suppose we calculate Euclidean distance:

$$
d=
\sqrt{
(Age_1-Age_2)^2+
(Salary_1-Salary_2)^2
}
$$

The salary difference can be much larger numerically than the age difference.

Therefore salary may dominate the distance.

This can make the distance misleading.

So distance-based algorithms often need **feature scaling**.

Examples:

$$
\boxed{
KNN,\ KMeans,\ Clustering
}
$$

---

# 12. Cosine Similarity

Cosine similarity measures the **angle between two vectors**.

The formula is:

$$
\boxed{
Cosine(A,B)
=
\frac{A\cdot B}
{\|A\|\|B\|}
}
$$

where:

$$
A\cdot B
=
\sum_{i=1}^{n}A_iB_i
$$

and:

$$
\|A\|
=
\sqrt{
\sum_{i=1}^{n}A_i^2
}
$$

---

# 13. Why Does Cosine Similarity Use the Angle?

Imagine two vectors:

$$
A
$$

and:

$$
B
$$

Cosine similarity asks:

> **Are these vectors pointing in the same direction?**

It does not primarily care about their magnitude.

Therefore it is especially useful for vectors such as:

- Text embeddings
- Word embeddings
- Document vectors
- Sentence embeddings

---

# 14. Cosine Similarity Example

Suppose:

$$
A=[1,2]
$$

$$
B=[2,4]
$$

### Step 1: Dot product

$$
A\cdot B
=
(1)(2)+(2)(4)
$$

$$
=2+8
$$

$$
=10
$$

### Step 2: Magnitude of $A$

$$
\|A\|
=
\sqrt{1^2+2^2}
$$

$$
=\sqrt5
$$

### Step 3: Magnitude of $B$

$$
\|B\|
=
\sqrt{2^2+4^2}
$$

$$
=\sqrt{20}
$$

### Step 4: Cosine similarity

$$
Cosine(A,B)
=
\frac{10}
{\sqrt5\sqrt{20}}
$$

Since:

$$
\sqrt5\sqrt{20}=10
$$

we get:

$$
\boxed{
Cosine(A,B)=1
}
$$

The vectors point in exactly the same direction.

---

# 15. Cosine Similarity Range

For real-valued vectors:

$$
\boxed{
-1\leq Cosine(A,B)\leq1
}
$$

### Cosine = 1

Vectors point in the same direction.

$$
\boxed{
\theta=0^\circ
}
$$

### Cosine = 0

Vectors are perpendicular.

$$
\boxed{
\theta=90^\circ
}
$$

### Cosine = -1

Vectors point in opposite directions.

$$
\boxed{
\theta=180^\circ
}
$$

---

# 16. Why Is Cosine Similarity Useful for NLP?

Suppose two documents have similar meaning but different lengths.

For example:

$$
A=[1,2,3]
$$

$$
B=[2,4,6]
$$

They have different magnitudes:

$$
\|A\|\neq\|B\|
$$

but:

$$
B=2A
$$

Therefore they point in exactly the same direction:

$$
\boxed{
Cosine(A,B)=1
}
$$

This is useful for embeddings because we often care more about the **direction/semantic pattern** than the magnitude.

---

# 17. Cosine Similarity vs Euclidean Distance

Consider:

$$
A=[1,1]
$$

$$
B=[2,2]
$$

### Cosine Similarity

Because they point in the same direction:

$$
\boxed{
Cosine(A,B)=1
}
$$

### Euclidean Distance

$$
d(A,B)
=
\sqrt{
(2-1)^2+(2-1)^2
}
$$

$$
=\sqrt2
$$

Therefore:

$$
\boxed{
Cosine=1
}
$$

but:

$$
\boxed{
Euclidean=\sqrt2
}
$$

This shows the important difference:

> Euclidean cares about **distance/magnitude**.

> Cosine cares about **direction**.

---

# 18. Jaccard Similarity

Jaccard similarity is mainly used for **sets**.

It measures how much two sets overlap.

$$
\boxed{
J(A,B)
=
\frac{|A\cap B|}
{|A\cup B|}
}
$$

where:

$$
A\cap B
=
\text{elements common to both sets}
$$

and:

$$
A\cup B
=
\text{all unique elements from both sets}
$$

---

# 19. Jaccard Example

Suppose:

$$
A=\{1,2,3,4\}
$$

$$
B=\{3,4,5,6\}
$$

Intersection:

$$
A\cap B=\{3,4\}
$$

Therefore:

$$
|A\cap B|=2
$$

Union:

$$
A\cup B
=
\{1,2,3,4,5,6\}
$$

Therefore:

$$
|A\cup B|=6
$$

Jaccard similarity:

$$
J(A,B)
=
\frac{2}{6}
$$

$$
\boxed{
J(A,B)=0.333
}
$$

---

# 20. Jaccard Range

$$
\boxed{
0\leq J(A,B)\leq1
}
$$

### $J=1$

The sets are identical.

$$
A=B
$$

### $J=0$

The sets have no common elements.

$$
A\cap B=\emptyset
$$

---

# 21. Similarity vs Distance

| Method | Type | Main Idea |
|---|---|---|
| Cosine | Similarity | Direction of vectors |
| Jaccard | Similarity | Set overlap |
| Euclidean | Distance | Straight-line distance |
| Manhattan | Distance | Sum of coordinate differences |
| Minkowski | Distance | Generalized distance |

---

# 22. Which One Should I Use?

### Use Euclidean Distance when:

You care about actual geometric distance.

Common examples:

$$
KNN,\ KMeans
$$

### Use Manhattan Distance when:

You want the absolute difference along each dimension and the grid-like distance makes sense.

### Use Cosine Similarity when:

You care about the direction of vectors.

Very common for:

$$
NLP,\ Embeddings,\ Semantic\ Search
$$

### Use Jaccard Similarity when:

You are comparing **sets or binary presence/absence information**.

Example:

$$
\text{Words in Document A}
$$

vs:

$$
\text{Words in Document B}
$$

# 23. Important Interview Questions

### What is the difference between similarity and distance?

Similarity becomes larger when objects are more alike:

$$
Similarity\uparrow
\Rightarrow
More\ Similar
$$

Distance becomes smaller when objects are more alike:

$$
Distance\downarrow
\Rightarrow
More\ Similar
$$

### Euclidean vs Cosine?

Euclidean measures geometric distance and is affected by magnitude.

Cosine measures the angle between vectors and focuses on direction.

### Why is cosine similarity useful for embeddings?

Because two embeddings can have different magnitudes but similar directions, and cosine similarity focuses on their direction.

### What is Minkowski distance?

It is a generalized distance:

$$
d(A,B)
=
\left(
\sum_i|A_i-B_i|^p
\right)^{1/p}
$$

with:

$$
p=1\Rightarrow Manhattan
$$

and:

$$
p=2\Rightarrow Euclidean
$$

### What is Jaccard similarity?

It measures the overlap between two sets:

$$
J(A,B)
=
\frac{|A\cap B|}
{|A\cup B|}
$$

### Why is feature scaling important for distance-based algorithms?

Because features with much larger numerical ranges can dominate the distance calculation.
