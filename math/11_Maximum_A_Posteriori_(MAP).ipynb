# 1. What is MAP?

MAP stands for **Maximum A Posteriori**.

MAP estimates the parameter that is **most probable after seeing the data**.

$$
\boxed{
\hat{\theta}_{MAP}
=
\arg\max_\theta P(\theta|D)
}
$$

where:

- $\theta$ = model parameters
- $D$ = observed data
- $P(\theta|D)$ = posterior probability

---

# 2. Bayes Theorem

Using Bayes theorem:

$$
P(\theta|D)
=
\frac{P(D|\theta)P(\theta)}
{P(D)}
$$

For optimization, $P(D)$ is constant with respect to $\theta$.

Therefore:

$$
\boxed{
\hat{\theta}_{MAP}
=
\arg\max_\theta
P(D|\theta)P(\theta)
}
$$

This is the key MAP formula.

---

# 3. What Does Each Part Mean?

$$
P(\theta|D)
\propto
P(D|\theta)P(\theta)
$$

### Likelihood

$$
P(D|\theta)
$$

How well do the parameters explain the observed data?

### Prior

$$
P(\theta)
$$

What did we believe about the parameters **before seeing the data**?

### Posterior

$$
P(\theta|D)
$$

What do we believe about the parameters **after seeing the data**?

Therefore:

$$
\boxed{
Posterior
\propto
Likelihood\times Prior
}
$$

---

# 4. MAP vs MLE

## MLE

MLE only considers the likelihood:

$$
\boxed{
\hat{\theta}_{MLE}
=
\arg\max_\theta P(D|\theta)
}
$$

It asks:

> Which parameter makes the observed data most likely?

## MAP

MAP considers both likelihood and prior:

$$
\boxed{
\hat{\theta}_{MAP}
=
\arg\max_\theta
P(D|\theta)P(\theta)
}
$$

It asks:

> Which parameter is most probable after considering both the data and our prior belief?

---

# 5. Main Difference

$$
\boxed{
MLE=Likelihood
}
$$

$$
\boxed{
MAP=Likelihood+Prior
}
$$

Or:

| MLE | MAP |
|---|---|
| Uses likelihood | Uses likelihood + prior |
| No prior information | Uses prior information |
| $\arg\max P(D|\theta)$ | $\arg\max P(D|\theta)P(\theta)$ |
| Data-driven | Data + prior-driven |

---

# 6. Log Form of MAP

Products are easier to optimize using logarithms.

Starting with:

$$
\hat{\theta}_{MAP}
=
\arg\max_\theta
P(D|\theta)P(\theta)
$$

Take log:

$$
\boxed{
\hat{\theta}_{MAP}
=
\arg\max_\theta
[
\log P(D|\theta)
+
\log P(\theta)
]
}
$$

Therefore:

$$
\boxed{
\text{MAP}
=
\text{Log-Likelihood}
+
\text{Log-Prior}
}
$$

---

# 7. MAP and Regularization

This is an important ML connection.

Suppose we assume a Gaussian prior on the weights:

$$
w\sim\mathcal{N}(0,\sigma^2)
$$

The Gaussian prior is:

$$
P(w)
\propto
e^{-\frac{w^2}{2\sigma^2}}
$$

Take negative log:

$$
-\log P(w)
\propto
\frac{w^2}{2\sigma^2}
$$

Therefore, MAP optimization adds a squared-weight penalty:

$$
\boxed{
Loss_{MAP}
=
Loss+\lambda w^2
}
$$

This is the connection between:

$$
\boxed{
Gaussian\ Prior
\rightarrow
L2\ Regularization
}
$$

Similarly, a Laplace prior leads to:

$$
\boxed{
Laplace\ Prior
\rightarrow
L1\ Regularization
}
$$

---

# 8. Why Does a Prior Act Like Regularization?

Suppose we have limited data.

MLE might choose parameters that fit the training data extremely well.

MAP says:

> Fit the data, but also prefer parameters that are plausible according to our prior.

Therefore, the prior can prevent excessively large or unusual parameter values.

This can reduce overfitting.

---

# 9. Simple Example

Suppose we want to estimate a coin's probability of heads.

We observe:

$$
1\ Head
$$

from one toss.

### MLE

The MLE estimate is:

$$
\hat p_{MLE}=1
$$

because the observed proportion of heads is $1$.

But with only one observation, this estimate is extreme.

### MAP

Suppose our prior belief says that probabilities near $0.5$ are more likely.

MAP combines:

$$
Likelihood+Prior
$$

so the estimate will generally be pulled away from the extreme value $1$ and toward the prior belief.

The exact MAP value depends on the chosen prior.

---

# 10. When Does MAP Become Similar to MLE?

When we have a very large amount of data, the likelihood usually becomes much stronger than the prior.

Therefore:

$$
\boxed{
Large\ Data
\Rightarrow
MAP\approx MLE
}
$$

The prior has less influence as the amount of data increases.

---

# 11. Important Interview Questions

### What is MAP?

MAP estimates the parameter with the highest posterior probability after considering both observed data and prior beliefs.

$$
\hat{\theta}_{MAP}
=
\arg\max_\theta P(\theta|D)
$$

### What is the difference between MLE and MAP?

MLE maximizes likelihood:

$$
P(D|\theta)
$$

MAP maximizes posterior:

$$
P(\theta|D)
\propto
P(D|\theta)P(\theta)
$$

### What is a prior?

The probability distribution representing our belief about the parameter before observing the data.

### What is a posterior?

Our updated belief about the parameter after observing the data.

### How is MAP related to regularization?

Choosing a prior on model parameters produces a penalty term in the MAP objective.

For example:

$$
Gaussian\ Prior
\rightarrow
L2\ Regularization
$$

and:

$$
Laplace\ Prior
\rightarrow
L1\ Regularization
$$

### Does MAP always outperform MLE?

No.

A poor or inappropriate prior can hurt the estimate.
