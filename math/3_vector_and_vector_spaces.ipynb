# 1. Cosine Similarity

## What?

Cosine similarity measures how similar two vectors are based on the **angle between them**, rather than their magnitude.

$$
\cos(\theta)=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

Where:

- $\mathbf{a}\cdot\mathbf{b}$ = dot product
- $\|\mathbf{a}\|$ = magnitude of $\mathbf{a}$
- $\|\mathbf{b}\|$ = magnitude of $\mathbf{b}$
- $\theta$ = angle between the vectors

## Why?

Two vectors can have very different magnitudes but point in the same direction.

For example:

$$
\mathbf{a}=[1,2]
$$

$$
\mathbf{b}=[10,20]
$$

They point in exactly the same direction, so:

$$
\cos(\theta)=1
$$

Therefore, cosine similarity focuses on **direction**, not size.

## Range

$$
-1\leq\cos(\theta)\leq1
$$

- $1$ → same direction
- $0$ → perpendicular
- $-1$ → opposite direction

## Example

Let:

$$
\mathbf{a}=[1,0]
$$

$$
\mathbf{b}=[1,1]
$$

Dot product:

$$
\mathbf{a}\cdot\mathbf{b}
=(1)(1)+(0)(1)=1
$$

Norms:

$$
\|\mathbf{a}\|=1
$$

$$
\|\mathbf{b}\|=\sqrt{1^2+1^2}=\sqrt{2}
$$

Therefore:

$$
\cos(\theta)
=
\frac{1}{\sqrt{2}}
\approx0.707
$$

## ML Applications

Cosine similarity is heavily used in:

- Text embeddings
- Semantic search
- Recommendation systems
- RAG
- Document similarity
- Image embeddings

For example, if two sentences have embedding vectors $\mathbf{e}_1$ and $\mathbf{e}_2$:

$$
Similarity=
\frac{\mathbf{e}_1\cdot\mathbf{e}_2}
{\|\mathbf{e}_1\|\|\mathbf{e}_2\|}
$$

A higher value generally means their embedding directions are more similar.

# 2. Projection

## What?

Projection tells us **how much of one vector lies in the direction of another vector**.

Suppose we want to project $\mathbf{a}$ onto $\mathbf{b}$.

The projection is:

$$
\operatorname{proj}_{\mathbf{b}}(\mathbf{a})
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{b}\|^2}\mathbf{b}
$$

## Why?

A vector can contain components in different directions.

Projection allows us to extract the component of $\mathbf{a}$ that lies along $\mathbf{b}$.

Think:

$$
\mathbf{a}
=
\text{parallel component}
+
\text{perpendicular component}
$$

## Example

Let:

$$
\mathbf{a}=
\begin{bmatrix}
3\\
2
\end{bmatrix},
\qquad
\mathbf{b}=
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

Dot product:

$$
\mathbf{a}\cdot\mathbf{b}
=(3)(1)+(2)(0)=3
$$

Norm squared:

$$
\|\mathbf{b}\|^2=1^2+0^2=1
$$

Therefore:

$$
\operatorname{proj}_{\mathbf{b}}(\mathbf{a})
=
\frac{3}{1}
\begin{bmatrix}
1\\
0
\end{bmatrix}
=
\begin{bmatrix}
3\\
0
\end{bmatrix}
$$

So the part of $\mathbf{a}$ pointing in the direction of $\mathbf{b}$ is:

$$
\begin{bmatrix}
3\\
0
\end{bmatrix}
$$

The remaining component is perpendicular:

$$
\mathbf{a}-\operatorname{proj}_{\mathbf{b}}(\mathbf{a})
=
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$

# 3. Norm

## What?

A norm measures the **size or magnitude of a vector**.

It is written as:

$$
\|\mathbf{x}\|
$$

Different norms measure vector size in different ways.

The two most important in ML are **L1 and L2 norms**.

---

# 4. L1 Norm

## Formula

For:

$$
\mathbf{x}=[x_1,x_2,\ldots,x_n]
$$

the L1 norm is:

$$
\|\mathbf{x}\|_1
=
\sum_{i=1}^{n}|x_i|
$$

It adds the absolute values.

## Example

$$
\mathbf{x}=[-3,4,-2]
$$

$$
\|\mathbf{x}\|_1
=
|-3|+|4|+|-2|
$$

$$
=3+4+2=9
$$

## Why?

L1 measures the total absolute magnitude of the vector.

## ML Connection: L1 Regularization

L1 regularization adds a penalty:

$$
Loss_{L1}
=
Loss+\lambda\sum_i|w_i|
$$

It tends to push some weights exactly to zero.

Therefore, L1 regularization can perform **feature selection**.

---

# 5. L2 Norm

## Formula

The L2 norm is:

$$
\|\mathbf{x}\|_2
=
\sqrt{\sum_{i=1}^{n}x_i^2}
$$

It is the usual Euclidean length of a vector.

## Example

$$
\mathbf{x}=[3,4]
$$

$$
\|\mathbf{x}\|_2
=
\sqrt{3^2+4^2}
$$

$$
=\sqrt{9+16}
$$

$$
=5
$$

This is the familiar Pythagorean theorem.

## ML Connection: L2 Regularization

L2 regularization adds:

$$
Loss_{L2}
=
Loss+\lambda\sum_iw_i^2
$$

It discourages very large weights and generally makes weights smaller rather than forcing many of them exactly to zero.

---

# 6. L1 vs L2

| Property | L1 | L2 |
|---|---|---|
| Formula | $\sum|x_i|$ | $\sqrt{\sum x_i^2}$ |
| Penalty | Absolute values | Squared values |
| Effect on weights | Can make weights exactly $0$ | Usually makes weights small |
| Common use | Feature selection | Weight regularization |
| Geometry | Diamond | Circle/Sphere |

Important:

$$
L1\rightarrow Sparsity
$$

$$
L2\rightarrow Small\ weights
$$

# 7. Orthogonality

## What?

Two vectors are orthogonal when they are **perpendicular**.

Mathematically:

$$
\mathbf{a}\cdot\mathbf{b}=0
$$

## Why?

Using:

$$
\mathbf{a}\cdot\mathbf{b}
=
\|\mathbf{a}\|\|\mathbf{b}\|\cos(\theta)
$$

For perpendicular vectors:

$$
\theta=90^\circ
$$

and:

$$
\cos(90^\circ)=0
$$

Therefore:

$$
\mathbf{a}\cdot\mathbf{b}=0
$$

## Example

$$
\mathbf{a}=
\begin{bmatrix}
1\\
0
\end{bmatrix},
\qquad
\mathbf{b}=
\begin{bmatrix}
0\\
1
\end{bmatrix}
$$

Then:

$$
\mathbf{a}\cdot\mathbf{b}
=(1)(0)+(0)(1)=0
$$

Therefore:

$$
\mathbf{a}\perp\mathbf{b}
$$

---

# 8. Connection Between All Four Concepts

These concepts are strongly connected.

### Dot Product

$$
\mathbf{a}\cdot\mathbf{b}
=
\|\mathbf{a}\|\|\mathbf{b}\|\cos(\theta)
$$

From this we get:

### Cosine Similarity

$$
\cos(\theta)
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

### Orthogonality

If:

$$
\mathbf{a}\cdot\mathbf{b}=0
$$

then:

$$
\mathbf{a}\perp\mathbf{b}
$$

### Projection

$$
\operatorname{proj}_{\mathbf{b}}(\mathbf{a})
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{b}\|^2}\mathbf{b}
$$

### Norm

Measures vector magnitude:

$$
\|\mathbf{x}\|_1=\sum_i|x_i|
$$

$$
\|\mathbf{x}\|_2=\sqrt{\sum_i x_i^2}
$$

---

# 9. Important Comparisons

## Cosine Similarity vs Euclidean Distance

**Cosine similarity** focuses on direction:

$$
\cos(\theta)
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

**Euclidean distance** focuses on actual distance:

$$
d(\mathbf{a},\mathbf{b})
=
\|\mathbf{a}-\mathbf{b}\|_2
$$

For text embeddings, cosine similarity is often useful because the direction of the embedding can matter more than its magnitude.

---

## L1 vs L2 Regularization

$$
L1:
\quad
Loss+\lambda\sum_i|w_i|
$$

$$
L2:
\quad
Loss+\lambda\sum_iw_i^2
$$

L1 tends toward sparse weights:

$$
w_i=0
$$

L2 tends toward smaller weights:

$$
|w_i|\rightarrow\text{small}
$$

# 10. Quick Interview Questions

### What does cosine similarity measure?

The cosine of the angle between two vectors, measuring their directional similarity.

### Why is cosine similarity useful for embeddings?

Because it compares vector direction while largely ignoring magnitude.

### When is cosine similarity equal to $1$?

When two non-zero vectors point in the same direction.

### When is cosine similarity $0$?

When the vectors are orthogonal.

### What is projection?

The component of one vector along the direction of another vector.

### What is L1 norm?

$$
\|\mathbf{x}\|_1=\sum_i|x_i|
$$

### What is L2 norm?

$$
\|\mathbf{x}\|_2=\sqrt{\sum_i x_i^2}
$$

### Why does L1 regularization produce sparsity?

The absolute-value penalty encourages some weights to become exactly zero.

### Why does L2 not usually produce exact zeros?

The squared penalty smoothly pushes weights toward zero rather than creating the strong zero-producing behavior of L1.

### How do you know two vectors are orthogonal?

Calculate their dot product:

$$
\mathbf{a}\cdot\mathbf{b}=0
$$
