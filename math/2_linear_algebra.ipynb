# 1. Vector

## What?

A vector is an ordered collection of numbers.

$$
\mathbf{x}=
\begin{bmatrix}
2\\
3\\
4
\end{bmatrix}
$$

This is a **3-dimensional vector** because it contains 3 values.

In ML, a vector can represent one data point:

$$
\mathbf{x}=
\begin{bmatrix}
Age\\
Salary\\
Experience
\end{bmatrix}
$$

## Why?

Vectors allow us to represent:

- Data points
- Features
- Weights
- Embeddings
- Directions
- Model parameters

## Important

A vector with $n$ values lives in an $n$-dimensional space.

$$
\mathbf{x}\in\mathbb{R}^{n}
$$

For example:

$$
\mathbf{x}\in\mathbb{R}^{5}
$$

means the vector has 5 components.

# 2. Dot Product

## What?

The dot product multiplies corresponding elements and then adds them.

For:

$$
\mathbf{a}=
\begin{bmatrix}
a_1\\
a_2\\
a_3
\end{bmatrix},
\qquad
\mathbf{b}=
\begin{bmatrix}
b_1\\
b_2\\
b_3
\end{bmatrix}
$$

we get:

$$
\mathbf{a}\cdot\mathbf{b}
=
a_1b_1+a_2b_2+a_3b_3
$$

## Example

$$
\mathbf{a}=
\begin{bmatrix}
2\\
3\\
4
\end{bmatrix},
\qquad
\mathbf{b}=
\begin{bmatrix}
5\\
2\\
1
\end{bmatrix}
$$

$$
\mathbf{a}\cdot\mathbf{b}
=
(2)(5)+(3)(2)+(4)(1)
$$

$$
=10+6+4=20
$$

## Why?

Geometrically:

$$
\mathbf{a}\cdot\mathbf{b}
=
\|\mathbf{a}\|\|\mathbf{b}\|\cos\theta
$$

So the dot product tells us how strongly two vectors point in the same direction.

- Positive → similar direction
- $0$ → perpendicular
- Negative → opposite tendency

## ML Connection

Dot products appear everywhere:

$$
\hat{y}=\mathbf{w}^{T}\mathbf{x}+b
$$

This is the basic calculation behind a linear model and neural-network layers.

# 3. Matrix

A matrix is a rectangular arrangement of numbers.

$$
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
$$

This matrix has:

$$
2\times3
$$

shape:

- 2 rows
- 3 columns

## Why?

Matrices allow us to store and transform large amounts of data efficiently.

A dataset can be represented as:

$$
X\in\mathbb{R}^{n\times d}
$$

where:

- $n$ = number of samples
- $d$ = number of features

---

# 4. Matrix Multiplication

## What?

Matrix multiplication combines rows of the first matrix with columns of the second matrix using dot products.

If:

$$
A\in\mathbb{R}^{m\times n}
$$

and

$$
B\in\mathbb{R}^{n\times p}
$$

then:

$$
AB\in\mathbb{R}^{m\times p}
$$

The **inner dimensions must match**.

$$
(m\times n)(n\times p)=(m\times p)
$$

## Example

$$
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
$$

$$
B=
\begin{bmatrix}
5&6\\
7&8
\end{bmatrix}
$$

Then:

$$
AB=
\begin{bmatrix}
1(5)+2(7)&1(6)+2(8)\\
3(5)+4(7)&3(6)+4(8)
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
19&22\\
43&50
\end{bmatrix}
$$

Each output element is a **row × column dot product**.

## Important Interview Point

Matrix multiplication is generally **not commutative**:

$$
AB\neq BA
$$

Even when both products are mathematically possible.

## ML Connection

Neural-network layers use matrix multiplication:

$$
Z=XW+b
$$

A large part of deep learning is essentially highly optimized matrix operations.

# 5. Transpose

## What?

Transpose changes rows into columns and columns into rows.

If:

$$
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
$$

then:

$$
A^T=
\begin{bmatrix}
1&4\\
2&5\\
3&6
\end{bmatrix}
$$

Therefore:

$$
(m\times n)^T=n\times m
$$

## Why?

Transpose is useful for:

- Changing vector orientation
- Matrix multiplication
- Dot products
- Covariance matrices
- Linear regression
- Neural networks

For example:

$$
\mathbf{x}^{T}\mathbf{y}
$$

produces the dot product of two vectors.

Important property:

$$
(AB)^T=B^TA^T
$$

# 6. Matrix Inverse

## What?

The inverse of a square matrix $A$ is a matrix $A^{-1}$ such that:

$$
AA^{-1}=A^{-1}A=I
$$

where $I$ is the identity matrix.

Similar to:

$$
5\times\frac{1}{5}=1
$$

For matrices:

$$
A\times A^{-1}=I
$$

## Example

For:

$$
A=
\begin{bmatrix}
a&b\\
c&d
\end{bmatrix}
$$

if:

$$
ad-bc\neq0
$$

then:

$$
A^{-1}
=
\frac{1}{ad-bc}
\begin{bmatrix}
d&-b\\
-c&a
\end{bmatrix}
$$

## Important

A matrix has an inverse only if:

$$
\det(A)\neq0
$$

If:

$$
\det(A)=0
$$

the matrix is **singular** and has no inverse.

## ML Connection

Inverse appears in the closed-form solution of linear regression:

$$
\mathbf{w}
=
(X^TX)^{-1}X^T\mathbf{y}
$$

In practical ML, we often avoid explicitly calculating the inverse because it can be computationally expensive and numerically unstable.

# 7. Rank

## What?

The rank of a matrix is the number of **linearly independent rows or columns**.

Example:

$$
A=
\begin{bmatrix}
1&2\\
2&4
\end{bmatrix}
$$

The second row is:

$$
2[1,2]=[2,4]
$$

So the second row does not provide new information.

Therefore:

$$
Rank(A)=1
$$

For:

$$
B=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
$$

the rows are independent:

$$
Rank(B)=2
$$

## Why?

Rank tells us how much **independent information** exists in a matrix.

## ML Connection

Rank is important in:

- Dimensionality reduction
- PCA
- SVD
- Multicollinearity
- Feature analysis

A low-rank matrix contains redundant information.

# 8. Determinant

## What?

The determinant is a scalar value calculated from a **square matrix**.

For:

$$
A=
\begin{bmatrix}
a&b\\
c&d
\end{bmatrix}
$$

the determinant is:

$$
\det(A)=ad-bc
$$

## Example

$$
A=
\begin{bmatrix}
2&3\\
1&4
\end{bmatrix}
$$

$$
\det(A)=(2)(4)-(3)(1)
$$

$$
=8-3=5
$$

## Why?

Geometrically, the absolute value of the determinant tells us how a transformation changes **area** in 2D or **volume** in higher dimensions.

Most importantly:

$$
\det(A)=0
$$

means the transformation collapses dimensions and the matrix is singular.

Therefore:

$$
\det(A)=0
\Rightarrow A^{-1}\text{ does not exist}
$$

For a square matrix:

$$
\det(A)\neq0
\iff
A\text{ is invertible}
$$

---

# 9. Rank vs Determinant

For an $n\times n$ square matrix:

$$
\det(A)\neq0
\Rightarrow Rank(A)=n
$$

and:

$$
\det(A)=0
\Rightarrow Rank(A)<n
$$

So determinant can help tell whether a square matrix has **full rank**.

# 10. Eigenvalues and Eigenvectors

## What?

For a matrix $A$, an eigenvector is a non-zero vector whose **direction does not change** after multiplication by $A$.

It may only be stretched or compressed.

$$
A\mathbf{v}=\lambda\mathbf{v}
$$

Where:

- $A$ = matrix
- $\mathbf{v}$ = eigenvector
- $\lambda$ = eigenvalue

## Intuition

Normally:

$$
A\mathbf{v}
$$

can change both the direction and magnitude of $\mathbf{v}$.

But for an eigenvector:

$$
A\mathbf{v}=\lambda\mathbf{v}
$$

Only its magnitude changes.

- $\lambda>1$ → stretched
- $0<\lambda<1$ → compressed
- $\lambda<0$ → direction flips
- $\lambda=0$ → mapped to zero


::contentReference[oaicite:0]{index=0}


## How to Find Eigenvalues?

Start with:

$$
A\mathbf{v}=\lambda\mathbf{v}
$$

Move everything to one side:

$$
(A-\lambda I)\mathbf{v}=0
$$

For a non-zero $\mathbf{v}$ to exist:

$$
\det(A-\lambda I)=0
$$

Solve this equation to obtain the eigenvalues $\lambda$.

Then substitute each eigenvalue back into:

$$
(A-\lambda I)\mathbf{v}=0
$$

to find its eigenvectors.

## ML Connection

Eigenvalues/eigenvectors are important in:

- PCA
- Covariance analysis
- Dimensionality reduction
- Spectral methods

In PCA, eigenvectors of the covariance matrix give the principal directions, while the corresponding eigenvalues represent the variance along those directions.

# 11. SVD — Singular Value Decomposition

## What?

SVD decomposes a matrix into three matrices:

$$
A=U\Sigma V^T
$$

Where:

- $U$ = left singular vectors
- $\Sigma$ = singular values
- $V^T$ = transpose of right singular vectors

## Intuition

Think of SVD as:

$$
\text{Original transformation}
\rightarrow
\text{Rotate}
\rightarrow
\text{Scale}
\rightarrow
\text{Rotate}
$$

So a complicated matrix transformation can be broken into simpler operations.

## Why?

SVD is useful because it reveals the important structure/information contained in a matrix.

Applications include:

- Dimensionality reduction
- PCA
- Data compression
- Noise reduction
- Recommender systems
- Latent representations

## Singular Values

The diagonal values in $\Sigma$ are the singular values:

$$
\Sigma=
\begin{bmatrix}
\sigma_1&0&0\\
0&\sigma_2&0\\
0&0&\sigma_3
\end{bmatrix}
$$

Usually:

$$
\sigma_1\geq\sigma_2\geq\sigma_3\geq0
$$

Large singular values represent directions containing more important information.

## Low-Rank Approximation

We can keep only the largest singular values:

$$
A\approx U_k\Sigma_kV_k^T
$$

This produces a smaller approximation while retaining most of the important information.

This is the basic intuition behind using SVD for **dimensionality reduction and compression**.

# 13. Quick Interview Revision

- **Vector:** Ordered collection of values representing a point, direction, feature set, etc.
- **Dot product:** Multiply corresponding elements and add; measures directional alignment.
- **Matrix multiplication:** Row-by-column dot products; inner dimensions must match.
- **Transpose:** Converts rows into columns.
- **Inverse:** Matrix that produces identity when multiplied by the original matrix.
- **Rank:** Number of linearly independent rows/columns.
- **Determinant:** Scalar describing properties such as area/volume scaling; zero means singular.
- **Eigenvector:** Direction that remains unchanged under a matrix transformation.
- **Eigenvalue:** Amount by which its eigenvector is scaled.
- **SVD:** Decomposes a matrix as $A=U\Sigma V^T$.

## Must-Know Relationships

$$
(AB)^T=B^TA^T
$$

$$
AA^{-1}=I
$$

$$
\det(A)\neq0\iff A^{-1}\text{ exists}
$$

$$
A\mathbf{v}=\lambda\mathbf{v}
$$

$$
\det(A-\lambda I)=0
$$

$$
A=U\Sigma V^T
$$
