# 1. What is Information Theory?

Information theory studies **how information and uncertainty are measured**.

In ML, the most important idea is:

$$
\boxed{
\text{How uncertain are we about an outcome?}
}
$$

Entropy measures this uncertainty.

---

# 2. What is Entropy?

Entropy measures the **uncertainty or randomness** of a probability distribution.

For a discrete random variable:

$$
\boxed{
H(X)
=
-\sum_{i=1}^{n}
P(x_i)\log P(x_i)
}
$$

where:

- $P(x_i)$ = probability of outcome $x_i$
- $\log$ is usually base $2$ in information theory

If base $2$ is used, entropy is measured in **bits**.

---

# 3. Why Do We Use the Negative Sign?

Because:

$$
0<P(x_i)\leq1
$$

and therefore:

$$
\log P(x_i)\leq0
$$

So:

$$
-P(x_i)\log P(x_i)\geq0
$$

This makes entropy non-negative:

$$
\boxed{
H(X)\geq0
}
$$

---

# 4. Simple Entropy Example

Suppose a coin has:

$$
P(Head)=0.5
$$

and:

$$
P(Tail)=0.5
$$

Entropy is:

$$
H(X)
=
-\left[
0.5\log_2(0.5)
+
0.5\log_2(0.5)
\right]
$$

Since:

$$
\log_2(0.5)=-1
$$

we get:

$$
H(X)
=
-\left[
0.5(-1)+0.5(-1)
\right]
$$

$$
=
-[-1]
$$

$$
\boxed{
H(X)=1\text{ bit}
}
$$

The uncertainty is maximum because both outcomes are equally likely.

---

# 5. What Happens When There Is No Uncertainty?

Suppose:

$$
P(Head)=1
$$

and:

$$
P(Tail)=0
$$

Then:

$$
H(X)
=
-\left[
1\log_2(1)
+
0\log_2(0)
\right]
$$

Since:

$$
\log_2(1)=0
$$

and:

$$
0\log(0)=0
$$

we get:

$$
\boxed{
H(X)=0
}
$$

There is no uncertainty because we already know the outcome.

Therefore:

$$
\boxed{
Certain\ Outcome
\rightarrow
Entropy=0
}
$$

---

# 6. Important Entropy Intuition

For a binary variable:

$$
H(X)
=
-p\log_2p
-
(1-p)\log_2(1-p)
$$

Entropy is highest when:

$$
p=0.5
$$

and lowest when:

$$
p=0
$$

or:

$$
p=1
$$

Therefore:

$$
\boxed{
Maximum\ Entropy
\rightarrow
Maximum\ Uncertainty
}
$$

$$
\boxed{
Minimum\ Entropy
\rightarrow
Minimum\ Uncertainty
}
$$

---

# 7. Why is Entropy Used in Decision Trees?

Suppose a node contains:

$$
50\ Positive
$$

and:

$$
50\ Negative
$$

The node is very mixed.

Therefore uncertainty is high:

$$
\boxed{
High\ Entropy
}
$$

Now suppose a node contains:

$$
100\ Positive
$$

and:

$$
0\ Negative
$$

The node is completely pure.

Therefore:

$$
\boxed{
Entropy=0
}
$$

Decision trees try to create splits that reduce this uncertainty.

This leads to:

$$
\boxed{
Information\ Gain
}
$$

---

# 8. Information Gain

Information gain measures:

> **How much uncertainty is reduced after making a split.**

Formula:

$$
\boxed{
IG(Y,X)
=
H(Y)-H(Y|X)
}
$$

where:

- $H(Y)$ = entropy before the split
- $H(Y|X)$ = remaining entropy after knowing $X$

Therefore:

$$
\boxed{
Information\ Gain
=
Original\ Uncertainty
-
Remaining\ Uncertainty
}
$$

---

# 9. Simple Information Gain Example

Suppose before splitting:

$$
H(Y)=1
$$

After using feature $X$:

$$
H(Y|X)=0.4
$$

Then:

$$
IG(Y,X)
=
1-0.4
$$

$$
\boxed{
IG(Y,X)=0.6
}
$$

So the feature reduced uncertainty by $0.6$ bits.

A larger information gain means:

$$
\boxed{
Better\ Split
}
$$

---

# 10. Conditional Entropy

Conditional entropy measures the uncertainty remaining in $Y$ after knowing $X$.

It is written as:

$$
\boxed{
H(Y|X)
}
$$

Formula:

$$
\boxed{
H(Y|X)
=
-\sum_xP(x)
\sum_yP(y|x)\log P(y|x)
}
$$

Simple meaning:

> How uncertain am I about $Y$ after I already know $X$?

---

# 11. Example of Conditional Entropy

Suppose:

$$
X=\text{Weather}
$$

and:

$$
Y=\text{Play\ Cricket}
$$

If knowing the weather tells us exactly whether to play cricket, then:

$$
\boxed{
H(Y|X)=0
}
$$

because there is no uncertainty remaining.

If knowing the weather does not help at all, then:

$$
\boxed{
H(Y|X)\approx H(Y)
}
$$

There has been little or no reduction in uncertainty.

---

# 12. Mutual Information

Mutual information measures **how much information one variable gives us about another variable**.

Formula:

$$
\boxed{
I(X;Y)
=
H(Y)-H(Y|X)
}
$$

It can also be written as:

$$
\boxed{
I(X;Y)
=
H(X)-H(X|Y)
}
$$

Another equivalent form is:

$$
\boxed{
I(X;Y)
=
\sum_{x,y}
P(x,y)
\log
\frac{P(x,y)}
{P(x)P(y)}
}
$$

---

# 13. Mutual Information Intuition

Suppose:

$$
X=\text{Feature}
$$

and:

$$
Y=\text{Target}
$$

If knowing $X$ tells us a lot about $Y$:

$$
\boxed{
I(X;Y)\text{ is high}
}
$$

If knowing $X$ tells us almost nothing about $Y$:

$$
\boxed{
I(X;Y)\text{ is low}
}
$$

Therefore:

$$
\boxed{
High\ Mutual\ Information
\rightarrow
Strong\ Dependency/Information\ Relationship
}
$$

$$
\boxed{
Low\ Mutual\ Information
\rightarrow
Little\ Information\ About\ Each\ Other
}
$$

---

# 14. Mutual Information and Independence

If $X$ and $Y$ are completely independent:

$$
P(x,y)=P(x)P(y)
$$

Then:

$$
\frac{P(x,y)}
{P(x)P(y)}
=
1
$$

and:

$$
\log(1)=0
$$

Therefore:

$$
\boxed{
I(X;Y)=0
}
$$

So:

$$
\boxed{
X\perp Y
\Rightarrow
I(X;Y)=0
}
$$

For dependent variables:

$$
\boxed{
I(X;Y)>0
}
$$

---

# 15. Entropy vs Mutual Information

| Concept | Meaning |
|---|---|
| Entropy $H(X)$ | Uncertainty in $X$ |
| Conditional Entropy $H(Y\|X)$ | Uncertainty in $Y$ after knowing $X$ |
| Mutual Information $I(X;Y)$ | How much information $X$ and $Y$ share |
| Information Gain | Reduction in target uncertainty after a split |

The key relationship is:

$$
\boxed{
I(X;Y)
=
H(Y)-H(Y|X)
}
$$

This is mathematically the same form as information gain:

$$
\boxed{
IG(Y,X)
=
H(Y)-H(Y|X)
}
$$

Therefore, in many ML contexts:

$$
\boxed{
Information\ Gain
=
Mutual\ Information
}
$$

when discussing the reduction in uncertainty of the target due to a feature.

---

# 16. Entropy vs Gini Impurity

Both can measure impurity in decision trees.

Entropy:

$$
\boxed{
H(X)
=
-\sum_iP(x_i)\log P(x_i)
}
$$

Gini impurity:

$$
\boxed{
Gini
=
1-\sum_iP(x_i)^2
}
$$

Both are:

$$
0
$$

when a node is completely pure.

Both become larger when the classes are more mixed.

# 18. Important Interview Questions

### What is entropy?

Entropy measures the uncertainty or randomness of a probability distribution.

$$
\boxed{
H(X)
=
-\sum_iP(x_i)\log P(x_i)
}
$$

### When is entropy maximum?

For a binary variable, entropy is maximum when:

$$
\boxed{
P(0)=P(1)=0.5
}
$$

because uncertainty is highest.

### When is entropy zero?

When one outcome is certain:

$$
\boxed{
P(x)=1
}
$$

### What is information gain?

It is the reduction in uncertainty after observing a feature or performing a split.

$$
\boxed{
IG(Y,X)
=
H(Y)-H(Y|X)
}
$$

### What is conditional entropy?

It measures the uncertainty remaining in $Y$ after knowing $X$.

$$
\boxed{
H(Y|X)
}
$$

### What is mutual information?

It measures how much information two variables share.

$$
\boxed{
I(X;Y)
=
H(Y)-H(Y|X)
}
$$

### What is mutual information when variables are independent?

$$
\boxed{
I(X;Y)=0
}
$$

### What is the difference between entropy and mutual information?

Entropy measures uncertainty in one variable.

Mutual information measures how much knowing one variable reduces uncertainty about another.
