# 1. What is SVM?

SVM stands for **Support Vector Machine**.

SVM is mainly used for classification.

Its main idea is:

$$
\boxed{
\text{Find a decision boundary that separates the classes with the maximum possible margin}
}
$$

For two features, the decision boundary is a line:

$$
w_1x_1+w_2x_2+b=0
$$

For more features, it becomes a hyperplane:

$$
\boxed{
w^Tx+b=0
}
$$

---

# 2. What is a Hyperplane?

A hyperplane is the decision boundary that separates the classes.

The equation is:

$$
\boxed{
w^Tx+b=0
}
$$

The model uses the value:

$$
w^Tx+b
$$

to determine which side of the boundary the point lies on.

For binary classification:

$$
w^Tx+b>0
\Rightarrow
Class=+1
$$

$$
w^Tx+b<0
\Rightarrow
Class=-1
$$

Therefore:

$$
\boxed{
w^Tx+b=0
\rightarrow
Decision\ Boundary
}
$$

---

# 3. What are $w$ and $b$?

The equation is:

$$
w^Tx+b=0
$$

where:

- $w$ = weight vector
- $b$ = bias/intercept
- $x$ = input vector

The weight vector controls the **orientation** of the hyperplane.

The bias controls the **position** of the hyperplane.

Therefore:

$$
\boxed{
w\rightarrow Orientation
}
$$

$$
\boxed{
b\rightarrow Position
}
$$

An important geometric property is:

$$
\boxed{
w\text{ is perpendicular to the hyperplane}
}
$$

---

# 4. Why Doesn't SVM Choose Any Separating Boundary?

Suppose many different lines can separate the positive and negative classes.

SVM does not simply choose any separating line.

It chooses the boundary that has the **largest margin**.

The idea is:

$$
\boxed{
Larger\ Margin
\rightarrow
Better\ Separation
}
$$

So SVM tries to keep the decision boundary as far as possible from the closest training points.

---

# 5. What is Margin?

The margin is the distance between the decision boundary and the closest training points.

The decision boundary is:

$$
\boxed{
w^Tx+b=0
}
$$

The two margin boundaries are:

$$
\boxed{
w^Tx+b=1
}
$$

and:

$$
\boxed{
w^Tx+b=-1
}
$$

The training points touching these boundaries are called **support vectors**.

---

# 6. What are Support Vectors?

Support vectors are the training points that are closest to the decision boundary.

For the standard hard-margin SVM:

$$
\boxed{
y_i(w^Tx_i+b)=1
}
$$

These points are important because they determine the position of the maximum-margin boundary.

Points far away from the boundary generally have much less influence on the final boundary.

Therefore:

$$
\boxed{
Support\ Vectors
\rightarrow
Closest\ Important\ Training\ Points
}
$$

---

# 7. How is the Margin Calculated?

The distance from a point $x$ to the hyperplane:

$$
w^Tx+b=0
$$

is:

$$
\boxed{
Distance=
\frac{|w^Tx+b|}
{\|w\|}
}
$$

For the positive margin boundary:

$$
w^Tx+b=1
$$

the distance from the decision boundary is:

$$
\frac{1}{\|w\|}
$$

For the negative margin boundary:

$$
w^Tx+b=-1
$$

the distance is also:

$$
\frac{1}{\|w\|}
$$

Therefore the total margin is:

$$
\frac{1}{\|w\|}
+
\frac{1}{\|w\|}
$$

$$
\boxed{
Margin=
\frac{2}{\|w\|}
}
$$

---

# 8. Why Does SVM Maximize the Margin?

We have:

$$
Margin=
\frac{2}{\|w\|}
$$

To make the margin large, we need:

$$
\|w\|\rightarrow\text{small}
$$

Therefore maximizing the margin is equivalent to minimizing:

$$
\|w\|
$$

SVM commonly minimizes:

$$
\boxed{
\frac{1}{2}\|w\|^2
}
$$

The $\frac{1}{2}$ is mainly used because it makes differentiation simpler:

$$
\frac{d}{dw}
\left(
\frac{1}{2}w^2
\right)
=
w
$$

Therefore the basic hard-margin SVM optimization problem is:

$$
\boxed{
\min_{w,b}
\frac{1}{2}\|w\|^2
}
$$

subject to:

$$
\boxed{
y_i(w^Tx_i+b)\geq1
}
$$

---

# 9. Why Do We Use $y_i(w^Tx_i+b)$?

For SVM:

$$
y_i\in\{-1,+1\}
$$

Suppose:

$$
y_i=+1
$$

We want:

$$
w^Tx_i+b\geq1
$$

Now suppose:

$$
y_i=-1
$$

We want:

$$
w^Tx_i+b\leq-1
$$

Both conditions can be written using one equation:

$$
\boxed{
y_i(w^Tx_i+b)\geq1
}
$$

This is one of the most important SVM formulas.

---

# 10. Hinge Loss

Real-world data is often not perfectly separable.

Therefore, some points may be:

- On the wrong side of the boundary
- Inside the margin
- Outside the desired margin

Hinge loss measures this violation.

The formula is:

$$
\boxed{
L_i=
\max
\left(
0,
1-y_i(w^Tx_i+b)
\right)
}
$$

Define:

$$
m_i=y_i(w^Tx_i+b)
$$

Then:

$$
\boxed{
L_i=\max(0,1-m_i)
}
$$

---

# 11. Understanding Hinge Loss

## Case 1: Correct Prediction with Enough Margin

Suppose:

$$
y_i(w^Tx_i+b)=2
$$

Then:

$$
L_i=
\max(0,1-2)
$$

$$
=
\max(0,-1)
$$

$$
\boxed{
L_i=0
}
$$

The point is correctly classified and is outside the margin.

No loss is produced.

---

## Case 2: Correct Prediction but Inside the Margin

Suppose:

$$
y_i(w^Tx_i+b)=0.5
$$

Then:

$$
L_i=
\max(0,1-0.5)
$$

$$
\boxed{
L_i=0.5
}
$$

The point is correctly classified because:

$$
y_i(w^Tx_i+b)>0
$$

but it is too close to the decision boundary.

Therefore SVM penalizes it.

---

## Case 3: Wrong Prediction

Suppose:

$$
y_i(w^Tx_i+b)=-1
$$

Then:

$$
L_i=
\max(0,1-(-1))
$$

$$
=
\max(0,2)
$$

$$
\boxed{
L_i=2
}
$$

The point is incorrectly classified, so it receives a large penalty.

---

# 12. Hinge Loss Intuition

The important condition is:

$$
\boxed{
y_i(w^Tx_i+b)\geq1
\Rightarrow
Loss=0
}
$$

If:

$$
0<y_i(w^Tx_i+b)<1
$$

then:

$$
\boxed{
Loss>0
}
$$

The prediction is correct but the margin is not large enough.

If:

$$
y_i(w^Tx_i+b)\leq0
$$

then the point is incorrectly classified.

Therefore:

$$
\boxed{
Hinge\ Loss
=
\text{Penalty for wrong prediction or insufficient margin}
}
$$

---

# 13. Hard-Margin SVM

Hard-margin SVM assumes that the classes can be perfectly separated.

The optimization problem is:

$$
\boxed{
\min_{w,b}
\frac{1}{2}\|w\|^2
}
$$

subject to:

$$
\boxed{
y_i(w^Tx_i+b)\geq1
}
$$

This means every training point must be correctly classified and outside the margin.

---

# 14. Why Do We Need Soft-Margin SVM?

Real-world data is usually not perfectly separable.

There may be:

- Noise
- Outliers
- Overlapping classes
- Incorrect labels

Hard-margin SVM cannot handle these situations well.

Therefore we allow some violations.

This gives us:

$$
\boxed{
Soft\text{-}Margin\ SVM
}
$$

---

# 15. Slack Variable

Soft-margin SVM introduces a slack variable:

$$
\xi_i\geq0
$$

The hard constraint:

$$
y_i(w^Tx_i+b)\geq1
$$

becomes:

$$
\boxed{
y_i(w^Tx_i+b)\geq1-\xi_i
}
$$

The slack variable tells us how much the point violates the desired margin.

---

# 16. Understanding $\xi_i$

### If:

$$
\xi_i=0
$$

The point satisfies the margin.

### If:

$$
0<\xi_i<1
$$

The point is correctly classified but lies inside the margin.

### If:

$$
\xi_i=1
$$

The point lies on the decision boundary.

### If:

$$
\xi_i>1
$$

The point is on the wrong side of the decision boundary.

Therefore:

$$
\boxed{
\xi_i
\rightarrow
Amount\ of\ Margin\ Violation
}
$$

---

# 17. Soft-Margin Objective

The soft-margin SVM objective is:

$$
\boxed{
\min_{w,b,\xi}
\frac{1}{2}\|w\|^2
+
C\sum_{i=1}^{n}\xi_i
}
$$

There are two important parts.

### First part

$$
\frac{1}{2}\|w\|^2
$$

encourages a large margin.

### Second part

$$
C\sum_i\xi_i
$$

penalizes margin violations.

Therefore:

$$
\boxed{
SVM
=
Large\ Margin
+
Penalty\ for\ Violations
}
$$

---

# 18. What is $C$?

$C$ controls the tradeoff between:

$$
\boxed{
Large\ Margin
}
$$

and:

$$
\boxed{
Training\ Errors
}
$$

---

## Large $C$

A large $C$ means violations are heavily penalized.

The model tries harder to classify training examples correctly.

Therefore:

$$
\boxed{
Large\ C
\rightarrow
Less\ Training\ Error
}
$$

But if $C$ is too large:

$$
\boxed{
Possible\ Overfitting
}
$$

---

## Small $C$

A small $C$ means violations are less heavily penalized.

The model is more willing to allow some training errors in exchange for a wider margin.

Therefore:

$$
\boxed{
Small\ C
\rightarrow
Wider\ Margin
}
$$

and:

$$
\boxed{
More\ Tolerance\ for\ Training\ Errors
}
$$

---

# 19. Complete Soft-Margin Objective Using Hinge Loss

The soft-margin objective can also be written directly using hinge loss:

$$
\boxed{
\min_{w,b}
\frac{1}{2}\|w\|^2
+
C
\sum_{i=1}^{n}
\max
\left(
0,
1-y_i(w^Tx_i+b)
\right)
}
$$

Here:

$$
\frac{1}{2}\|w\|^2
$$

controls the margin.

And:

$$
C
\sum_i
\max
\left(
0,
1-y_i(w^Tx_i+b)
\right)
$$

controls the penalty for violations.

---

# 20. Why is SVM Related to Regularization?

Look at the objective:

$$
\frac{1}{2}\|w\|^2
+
C\sum_iL_i
$$

The first term:

$$
\frac{1}{2}\|w\|^2
$$

penalizes large weights.

This is an $L_2$ regularization term.

Smaller weights mean:

$$
\|w\|\downarrow
$$

and because:

$$
Margin=\frac{2}{\|w\|}
$$

we get:

$$
Margin\uparrow
$$

Therefore:

$$
\boxed{
L_2\ Regularization
\rightarrow
Smaller\ Weights
\rightarrow
Larger\ Margin
}
$$

---

# 21. SVM vs Logistic Regression

| | SVM | Logistic Regression |
|---|---|---|
| Main idea | Maximum margin | Probability modeling |
| Decision boundary | $w^Tx+b=0$ | $w^Tx+b=0$ |
| Loss | Hinge loss | Log loss / Cross Entropy |
| Output | Decision score | Probability |
| Main focus | Boundary + margin | Probability of class |
| Regularization | Common | Common |

The important difference is:

$$
\boxed{
SVM
\rightarrow
Maximum\ Margin
}
$$

while:

$$
\boxed{
Logistic\ Regression
\rightarrow
Probability\ Estimation
}
$$

# 22. Advantages of SVM

- Works well in high-dimensional spaces.
- Can work well when there are many features.
- The maximum-margin idea can improve generalization.
- Kernel methods allow nonlinear decision boundaries.

---

# 23. Disadvantages of SVM

- Can be computationally expensive for very large datasets.
- Choosing the right kernel and hyperparameters can be difficult.
- Sensitive to feature scaling.
- Less interpretable than simple linear models.
- Does not naturally provide probabilities without additional calibration.

# 24. Important Interview Questions

### What is SVM?

SVM finds a decision boundary that separates classes while maximizing the margin between the boundary and the closest training points.

### What is a hyperplane?

$$
\boxed{
w^Tx+b=0
}
$$

It is the decision boundary separating the classes.

### What is a margin?

The distance between the decision boundary and the closest training points.

$$
\boxed{
Margin=\frac{2}{\|w\|}
}
$$

### Why does SVM maximize the margin?

Because a larger margin generally provides better separation between classes and can improve generalization.

Since:

$$
Margin=\frac{2}{\|w\|}
$$

maximizing the margin means minimizing:

$$
\frac{1}{2}\|w\|^2
$$

### What are support vectors?

They are the training points closest to the decision boundary that determine the maximum-margin hyperplane.

### What is hinge loss?

$$
\boxed{
L=
\max(0,1-y(w^Tx+b))
}
$$

It penalizes incorrect predictions and points that are inside the desired margin.

### Why do we need soft-margin SVM?

Because real-world data may not be perfectly separable due to noise, outliers, and overlapping classes.

### What does $C$ control?

$C$ controls the tradeoff between maximizing the margin and penalizing training/margin violations.

### What happens when $C$ increases?

The model penalizes violations more strongly and tries harder to fit the training data.

### What happens when $C$ decreases?

The model allows more violations and prefers a wider margin.
