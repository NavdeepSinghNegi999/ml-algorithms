# 1. What is MLE?

MLE stands for **Maximum Likelihood Estimation**.

MLE is a method for finding the parameter values that make the **observed data most likely**.

Suppose we have data:

$$
X=\{x_1,x_2,\ldots,x_n\}
$$

and a probability distribution with an unknown parameter $\theta$.

MLE finds:

$$
\boxed{
\hat{\theta}
=
\arg\max_{\theta}L(\theta)
}
$$

where $L(\theta)$ is the likelihood.

### Simple Intuition

Suppose we toss a coin 10 times and get:

$$
7\ Heads,\quad3\ Tails
$$

We want to estimate the probability of heads, $p$.

Intuitively, $p$ should be close to:

$$
0.7
$$

MLE gives us the value of $p$ that makes observing those 7 heads and 3 tails most likely.

---

# 2. Probability vs Likelihood

This is an important distinction.

### Probability

Parameters are fixed, and we ask about the data:

$$
P(Data|\theta)
$$

Example:

> If the probability of heads is $0.7$, what is the probability of getting 7 heads?

### Likelihood

Data is fixed, and we vary the parameter:

$$
L(\theta|Data)
$$

Example:

> We observed 7 heads. Which value of $p$ makes this observation most likely?

The mathematical expression can look the same:

$$
P(Data|\theta)
$$

but the **question being asked is different**.

---

# 3. Likelihood Function

Suppose observations are independent:

$$
x_1,x_2,\ldots,x_n
$$

Then:

$$
P(X_1=x_1,\ldots,X_n=x_n|\theta)
$$

becomes:

$$
\boxed{
L(\theta)
=
\prod_{i=1}^{n}P(x_i|\theta)
}
$$

We multiply because the observations are assumed independent.

---

# 4. Why Do We Use Log-Likelihood?

The likelihood contains multiplication:

$$
L(\theta)=
\prod_{i=1}^{n}P(x_i|\theta)
$$

For many observations, multiplying many small probabilities can cause numerical underflow.

Taking logarithm converts multiplication into addition:

$$
\boxed{
\log L(\theta)
=
\sum_{i=1}^{n}\log P(x_i|\theta)
}
$$

This is much easier to differentiate and optimize.

Because $\log(x)$ is monotonically increasing:

$$
\arg\max_\theta L(\theta)
=
\arg\max_\theta\log L(\theta)
$$

Therefore, maximizing likelihood and maximizing log-likelihood give the **same parameter estimate**.

---

# 5. Bernoulli MLE Example

Suppose:

$$
X_i\sim Bernoulli(p)
$$

For one observation:

$$
P(X_i=x_i)
=
p^{x_i}(1-p)^{1-x_i}
$$

For $n$ independent observations:

$$
L(p)
=
\prod_{i=1}^{n}
p^{x_i}(1-p)^{1-x_i}
$$

Combine powers:

$$
L(p)
=
p^{\sum x_i}
(1-p)^{n-\sum x_i}
$$

Take log:

$$
\ell(p)
=
\log L(p)
$$

$$
=
\left(\sum x_i\right)\log p
+
\left(n-\sum x_i\right)\log(1-p)
$$

Differentiate:

$$
\frac{d\ell}{dp}
=
\frac{\sum x_i}{p}
-
\frac{n-\sum x_i}{1-p}
$$

Set derivative to zero:

$$
\frac{\sum x_i}{p}
=
\frac{n-\sum x_i}{1-p}
$$

Solving:

$$
\boxed{
\hat p=
\frac{\sum x_i}{n}
}
$$

So the MLE estimate of Bernoulli probability is simply the **observed proportion of successes**.

# 6. Logistic Regression and MLE

Logistic regression predicts:

$$
p_i=P(y_i=1|x_i)
$$

using:

$$
\boxed{
p_i=\sigma(z_i)
}
$$

where:

$$
z_i=\mathbf{x}_i^T\mathbf{w}+b
$$

and:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Since:

$$
y_i\in\{0,1\}
$$

we model each target as a Bernoulli random variable.

Therefore:

$$
P(y_i|x_i)
=
p_i^{y_i}(1-p_i)^{1-y_i}
$$

For all observations:

$$
\boxed{
L(\mathbf{w})
=
\prod_{i=1}^{n}
p_i^{y_i}(1-p_i)^{1-y_i}
}
$$

---

# 7. Logistic Regression Log-Likelihood

Take logarithm:

$$
\ell(\mathbf{w})
=
\log L(\mathbf{w})
$$

Therefore:

$$
\boxed{
\ell(\mathbf{w})
=
\sum_{i=1}^{n}
\left[
y_i\log p_i
+
(1-y_i)\log(1-p_i)
\right]
}
$$

MLE wants to maximize this:

$$
\boxed{
\max_{\mathbf{w}}\ell(\mathbf{w})
}
$$

But ML optimization is commonly expressed as **minimization**.

So we minimize the negative log-likelihood:

$$
\boxed{
-\ell(\mathbf{w})
=
-\sum_{i=1}^{n}
\left[
y_i\log p_i
+
(1-y_i)\log(1-p_i)
\right]
}
$$

This is exactly the **Binary Cross-Entropy loss**.

Therefore:

$$
\boxed{
\text{Logistic Regression}
\rightarrow
\text{MLE}
\rightarrow
\text{Negative Log-Likelihood}
\rightarrow
\text{Binary Cross-Entropy}
}
$$

---

# 8. Why Does Logistic Regression Use Log-Likelihood?

Because the target is Bernoulli:

$$
Y\sim Bernoulli(p)
$$

The likelihood naturally becomes:

$$
L=
\prod_i
p_i^{y_i}(1-p_i)^{1-y_i}
$$

Taking the logarithm:

$$
\log L
=
\sum_i
\left[
y_i\log p_i+
(1-y_i)\log(1-p_i)
\right]
$$

The negative of this is Binary Cross-Entropy:

$$
\boxed{
BCE
=
-\sum_i
[y_i\log p_i+(1-y_i)\log(1-p_i)]
}
$$

So BCE is not an arbitrary loss.

It comes directly from **maximum likelihood estimation under a Bernoulli model**.

# 9. Gaussian MLE

Suppose:

$$
X_1,X_2,\ldots,X_n
\sim
\mathcal{N}(\mu,\sigma^2)
$$

The Gaussian PDF is:

$$
f(x|\mu,\sigma^2)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

For independent observations:

$$
L(\mu,\sigma^2)
=
\prod_{i=1}^{n}
f(x_i|\mu,\sigma^2)
$$

Therefore:

$$
L(\mu,\sigma^2)
=
\prod_{i=1}^{n}
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x_i-\mu)^2}{2\sigma^2}}
$$

---

# 10. Gaussian Log-Likelihood

Take the logarithm:

$$
\ell(\mu,\sigma^2)
=
\log L(\mu,\sigma^2)
$$

After simplifying:

$$
\boxed{
\ell(\mu,\sigma^2)
=
-\frac{n}{2}\log(2\pi)
-\frac{n}{2}\log(\sigma^2)
-\frac{1}{2\sigma^2}
\sum_{i=1}^{n}(x_i-\mu)^2
}
$$

Now we maximize this with respect to $\mu$ and $\sigma^2$.

---

# 11. MLE of Gaussian Mean

Differentiate with respect to $\mu$:

$$
\frac{\partial\ell}{\partial\mu}
=
\frac{1}{\sigma^2}
\sum_{i=1}^{n}(x_i-\mu)
$$

Set equal to zero:

$$
\frac{1}{\sigma^2}
\sum_{i=1}^{n}(x_i-\mu)=0
$$

Therefore:

$$
\sum_{i=1}^{n}(x_i-\mu)=0
$$

$$
\sum x_i-n\mu=0
$$

So:

$$
\boxed{
\hat{\mu}
=
\frac{1}{n}\sum_{i=1}^{n}x_i
}
$$

Therefore:

$$
\boxed{
\hat{\mu}_{MLE}=\text{sample mean}
}
$$

This is an important result.

---

# 12. MLE of Gaussian Variance

The MLE estimate of variance is:

$$
\boxed{
\hat{\sigma}^2_{MLE}
=
\frac{1}{n}
\sum_{i=1}^{n}
(x_i-\hat{\mu})^2
}
$$

Notice that MLE uses:

$$
\frac{1}{n}
$$

not:

$$
\frac{1}{n-1}
$$

The reason is that MLE is optimizing the likelihood, while the usual unbiased sample variance uses $n-1$.

Therefore:

$$
\boxed{
MLE\ Variance
=
\frac{1}{n}\sum(x_i-\bar{x})^2
}
$$

while:

$$
\boxed{
Unbiased\ Sample\ Variance
=
\frac{1}{n-1}\sum(x_i-\bar{x})^2
}
$$

---

# 13. MLE vs MAP

MLE finds parameters using only the likelihood:

$$
\boxed{
\hat{\theta}_{MLE}
=
\arg\max_\theta P(D|\theta)
}
$$

MAP also includes a prior:

$$
\boxed{
\hat{\theta}_{MAP}
=
\arg\max_\theta P(\theta|D)
}
$$

Using Bayes:

$$
P(\theta|D)
\propto
P(D|\theta)P(\theta)
$$

Therefore:

$$
\boxed{
MAP
=
Likelihood
+
Prior
}
$$

while:

$$
\boxed{
MLE
=
Likelihood
}
$$

---

# 14. MLE and Optimization

MLE is itself an optimization problem:

$$
\boxed{
\hat{\theta}
=
\arg\max_\theta\ell(\theta)
}
$$

Equivalently, minimize negative log-likelihood:

$$
\boxed{
\hat{\theta}
=
\arg\min_\theta[-\ell(\theta)]
}
$$

Therefore, gradient-based optimization can be used:

$$
\theta_{new}
=
\theta_{old}
-
\eta\nabla[-\ell(\theta)]
$$

This connects MLE directly to:

$$
\text{Calculus}
\rightarrow
\text{Gradients}
\rightarrow
\text{Optimization}
$$

# 15. Important Interview Questions

### What is MLE?

A method for estimating model parameters by finding the values that make the observed data most likely.

### Why do we use log-likelihood?

Because it converts products into sums, improves numerical stability, and makes differentiation easier.

$$
\log\prod_i p_i
=
\sum_i\log p_i
$$

### Does taking log change the MLE?

No.

Because $\log$ is monotonically increasing:

$$
\arg\max L(\theta)
=
\arg\max\log L(\theta)
$$

### Why does logistic regression use log-likelihood?

Because its target is modeled as a Bernoulli random variable, and the Bernoulli likelihood leads directly to the binary cross-entropy loss.

### What is the relationship between BCE and MLE?

BCE is the **negative log-likelihood of the Bernoulli model**.

### What is the Gaussian MLE for the mean?

$$
\boxed{
\hat{\mu}=\frac{1}{n}\sum_i x_i
}
$$

### What is the Gaussian MLE for variance?

$$
\boxed{
\hat{\sigma}^2
=
\frac{1}{n}
\sum_i(x_i-\hat{\mu})^2
}
$$

### Why does Gaussian MLE variance use $n$ instead of $n-1$?

Because MLE maximizes the likelihood. The $n-1$ denominator is used for the unbiased sample variance estimator.
