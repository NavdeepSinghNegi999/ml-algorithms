# 1. What is Bayes Theorem?

Bayes Theorem tells us how to **update the probability of something after getting new evidence**.

In simple English:

> "I had some initial belief. I got new information. How should I update my belief?"

$$
\boxed{
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
}
$$

Where:

- $P(A|B)$ → **Posterior**: probability of $A$ after knowing $B$
- $P(B|A)$ → **Likelihood**: probability of seeing $B$ if $A$ is true
- $P(A)$ → **Prior**: probability of $A$ before seeing $B$
- $P(B)$ → **Evidence**: overall probability of $B$

---

# 2. Why Do We Need Bayes Theorem?

Sometimes we know:

$$
P(B|A)
$$

but we actually want:

$$
P(A|B)
$$

These are **not the same thing**.

For example:

> Probability that a person has a disease given a positive test

is:

$$
P(Disease|Positive)
$$

But a medical test might give us:

$$
P(Positive|Disease)
$$

Bayes theorem lets us go from one to the other.

---

# 3. Derivation of Bayes Theorem

Start with conditional probability:

$$
P(A|B)
=
\frac{P(A\cap B)}{P(B)}
$$

We also know:

$$
P(B|A)
=
\frac{P(A\cap B)}{P(A)}
$$

Therefore:

$$
P(A\cap B)=P(B|A)P(A)
$$

Put this into the first equation:

$$
P(A|B)
=
\frac{P(B|A)P(A)}
{P(B)}
$$

Therefore:

$$
\boxed{
P(A|B)
=
\frac{P(B|A)P(A)}
{P(B)}
}
$$

So Bayes theorem is basically derived from the **definition of conditional probability**.

---

# 4. Main Intuition

$$
\boxed{
Posterior
=
\frac{Likelihood\times Prior}
{Evidence}
}
$$

Or:

$$
\boxed{
New\ Belief
=
\frac{How\ well\ evidence\ fits\ the\ hypothesis
\times
Old\ Belief}
{Overall\ probability\ of\ evidence}
}
$$

# 5. Simple Example

Suppose:

- $1\%$ of people have a disease.
- If someone has the disease, the test is positive $90\%$ of the time.
- If someone does not have the disease, the test is positive $5\%$ of the time.

We want:

$$
P(D|+)
$$

Given:

$$
P(D)=0.01
$$

$$
P(+|D)=0.90
$$

$$
P(+|D^c)=0.05
$$

Therefore:

$$
P(D^c)=0.99
$$

---

## Step 1: Calculate $P(+)$

A positive result can happen in two ways:

$$
P(+)
=
P(+|D)P(D)
+
P(+|D^c)P(D^c)
$$

$$
=
(0.90)(0.01)+(0.05)(0.99)
$$

$$
=0.009+0.0495
$$

$$
=0.0585
$$

---

## Step 2: Apply Bayes

$$
P(D|+)
=
\frac{P(+|D)P(D)}
{P(+)}
$$

$$
=
\frac{(0.90)(0.01)}
{0.0585}
$$

$$
\approx0.154
$$

Therefore:

$$
\boxed{
P(D|+)\approx15.4\%
}
$$

### Important Lesson

Even though the test is $90\%$ sensitive, a positive result means only about $15.4\%$ probability of disease in this example.

Why?

Because the disease is **rare**.

This is called the **base-rate effect**.

---

# 6. Why Is $P(B)$ Important?

A common interview question is:

> Why can't we just calculate $P(A|B)=P(B|A)P(A)$?

Because that gives us only the numerator.

We need to normalize by the probability of observing $B$:

$$
P(B)
$$

For example:

$$
P(+)
=
P(+|D)P(D)
+
P(+|D^c)P(D^c)
$$

This considers **all possible ways** a positive result can occur.

---

# 7. Bayes Theorem in Machine Learning

Bayes theorem is heavily used in **Naive Bayes**.

Suppose we want to classify an email as:

$$
Spam
$$

or:

$$
Not\ Spam
$$

We observe features:

$$
X=\{X_1,X_2,X_3\}
$$

We want:

$$
P(Spam|X)
$$

Using Bayes:

$$
P(Spam|X)
=
\frac{P(X|Spam)P(Spam)}
{P(X)}
$$

Since $P(X)$ is the same normalization factor for each class:

$$
\boxed{
P(Spam|X)
\propto
P(X|Spam)P(Spam)
}
$$

We calculate this for each class and choose the class with the highest posterior probability.

---

# 8. Why Is It Called Naive Bayes?

Suppose our features are:

$$
X_1,X_2,X_3
$$

Naive Bayes assumes the features are **conditionally independent given the class**.

Therefore:

$$
P(X_1,X_2,X_3|C)
$$

is approximated as:

$$
\boxed{
P(X_1|C)P(X_2|C)P(X_3|C)
}
$$

So:

$$
P(C|X_1,X_2,X_3)
\propto
P(C)
P(X_1|C)
P(X_2|C)
P(X_3|C)
$$

Therefore:

$$
\boxed{
P(C|X)
\propto
P(C)\prod_{i=1}^{n}P(X_i|C)
}
$$

The assumption is called "naive" because features are often not actually independent in real-world data.

---

# 9. Bayes Theorem vs Conditional Probability

### Conditional Probability

Tells us:

> What is the probability of $A$ when $B$ is already known?

$$
P(A|B)
=
\frac{P(A\cap B)}
{P(B)}
$$

### Bayes Theorem

Allows us to **reverse the conditional relationship**:

$$
P(B|A)
\rightarrow
P(A|B)
$$

using:

$$
P(A|B)
=
\frac{P(B|A)P(A)}
{P(B)}
$$

---

# 10. Bayes Theorem vs MLE

### MLE

Uses:

$$
P(D|\theta)
$$

It asks:

> Which parameter makes the observed data most likely?

### Bayesian Approach

Uses:

$$
P(\theta|D)
$$

It asks:

> After seeing the data, what is the probability of the parameter?

Bayes connects them:

$$
\boxed{
P(\theta|D)
=
\frac{P(D|\theta)P(\theta)}
{P(D)}
}
$$

Therefore:

$$
\boxed{
Posterior
\propto
Likelihood\times Prior
}
$$

---

# 11. Important Interview Questions

### What is Bayes theorem?

A theorem that calculates a posterior probability using a prior probability and likelihood.

$$
\boxed{
P(A|B)
=
\frac{P(B|A)P(A)}
{P(B)}
}
$$

### What is prior?

Our belief about $A$ before seeing new evidence.

### What is likelihood?

How probable the observed evidence is assuming $A$ is true.

### What is posterior?

Our updated belief about $A$ after seeing the evidence.

### What is evidence?

The overall probability of observing the evidence.

### Why is Bayes theorem important in ML?

It is the mathematical foundation of Bayesian inference and is directly used in algorithms such as Naive Bayes.

### Why is Naive Bayes called naive?

Because it assumes features are conditionally independent given the class.

### Is $P(A|B)$ the same as $P(B|A)$?

No.

$$
\boxed{
P(A|B)\neq P(B|A)
}
$$

For example:

$$
P(Disease|Positive)
\neq
P(Positive|Disease)
$$

---

# 5. Simple Example

Suppose:

- $1\%$ of people have a disease.
- If someone has the disease, the test is positive $90\%$ of the time.
- If someone does not have the disease, the test is positive $5\%$ of the time.

---

## Given

Let:

$$
D = \text{person has the disease}
$$

$$
D^c = \text{person does not have the disease}
$$

$$
+ = \text{test result is positive}
$$

We are given:

### 1. Probability of having the disease

$$
P(D)=0.01
$$

### 2. Probability of a positive test given that the person has the disease

$$
P(+|D)=0.90
$$

### 3. Probability of a positive test given that the person does not have the disease

$$
P(+|D^c)=0.05
$$

---

## Find

We want to find:

$$
\boxed{P(D|+)}
$$

This means:

> Given that the test is positive, what is the probability that the person actually has the disease?

---

# Step 1: Find $P(D^c)$

A person either has the disease or does not have the disease.

Therefore:

$$
P(D^c)=1-P(D)
$$

Substitute:

$$
P(D^c)=1-0.01
$$

$$
\boxed{P(D^c)=0.99}
$$

---

# Step 2: Find $P(D\cap +)$

We want the probability that:

> The person has the disease **AND** the test is positive.

Using the multiplication rule:

$$
P(D\cap +)=P(+|D)P(D)
$$

Substitute:

$$
P(D\cap +)=0.90\times0.01
$$

$$
\boxed{P(D\cap +)=0.009}
$$

---

# Step 3: Find $P(D^c\cap +)$

We want the probability that:

> The person does not have the disease **AND** the test is positive.

Using the multiplication rule:

$$
P(D^c\cap +)=P(+|D^c)P(D^c)
$$

Substitute:

$$
P(D^c\cap +)=0.05\times0.99
$$

$$
\boxed{P(D^c\cap +)=0.0495}
$$

---

# Step 4: Find $P(+)$

A positive test can happen in **two ways**:

1. The person has the disease and tests positive.
2. The person does not have the disease but tests positive.

Therefore:

$$
P(+)=P(D\cap +)+P(D^c\cap +)
$$

Substitute:

$$
P(+)=0.009+0.0495
$$

$$
\boxed{P(+)=0.0585}
$$

---

# Step 5: Apply Bayes' Theorem

Bayes' theorem:

$$
P(D|+)=
\frac{P(+|D)P(D)}{P(+)}
$$

Substitute:

$$
P(D|+)=
\frac{0.90\times0.01}{0.0585}
$$

First calculate the numerator:

$$
0.90\times0.01=0.009
$$

Therefore:

$$
P(D|+)=\frac{0.009}{0.0585}
$$

Now divide:

$$
P(D|+)=0.153846\ldots
$$

Therefore:

$$
\boxed{P(D|+)\approx0.1538}
$$

Convert to percentage:

$$
0.1538\times100=15.38\%
$$

Therefore:

$$
\boxed{P(D|+)\approx15.38\%}
$$

---

# Final Answer

$$
\boxed{P(D|+)\approx15.38\%}
$$

So, if the test is positive, the probability that the person actually has the disease is approximately:

$$
\boxed{15.38\%}
$$

### Important Point

We cannot just use:

$$
P(+|D)P(D)=0.009
$$

because:

$$
P(+|D)P(D)=P(D\cap +)
$$

This means:

> Probability that the person **has the disease AND tests positive**.

But we want:

$$
P(D|+)
$$

which means:

> Among **all people who tested positive**, what fraction actually has the disease?

Therefore:

$$
P(D|+)=
\frac{P(D\cap +)}{P(+)}
$$

and:

$$
\boxed{P(D|+)=15.38\%}
$$
