# Bias-Variance Tradeoff

## 1. Bias

Bias is the error caused by a model being **too simple** and making strong assumptions.

High bias usually means:

$$
\boxed{
High\ Bias\rightarrow Underfitting
}
$$

The model cannot capture the important patterns in the data.

Example:

$$
Training\ Error\ High
$$

$$
Test\ Error\ High
$$

---

## 2. Variance

Variance is the error caused by a model being **too sensitive to the training data**.

High variance usually means:

$$
\boxed{
High\ Variance\rightarrow Overfitting
}
$$

The model learns the training data too closely, including noise.

Example:

$$
Training\ Error\ Low
$$

$$
Test\ Error\ High
$$

---

## 3. Bias vs Variance

| | High Bias | High Variance |
|---|---|---|
| Model | Too simple | Too complex |
| Problem | Underfitting | Overfitting |
| Training error | High | Low |
| Test error | High | High |
| Learns noise? | Usually no | Often yes |

The goal is to find a good balance:

$$
\boxed{
Low\ Bias+Low\ Variance
\rightarrow
Better\ Generalization
}
$$

---

# ROC Curve

## 1. What is ROC?

ROC stands for **Receiver Operating Characteristic**.

It is a graph of:

$$
\boxed{
TPR\text{ vs }FPR
}
$$

where:

$$
TPR=\frac{TP}{TP+FN}
$$

and:

$$
FPR=\frac{FP}{FP+TN}
$$

---

## 2. Why Do We Need Different Thresholds?

A classification model usually gives a probability.

For example:

$$
[0.9,\ 0.7,\ 0.4,\ 0.2]
$$

If the threshold is:

$$
0.5
$$

then:

$$
P\geq0.5\rightarrow Positive
$$

$$
P<0.5\rightarrow Negative
$$

But we can change the threshold:

$$
0.9,\ 0.8,\ 0.7,\ 0.6,\ 0.5,\ldots
$$

Every threshold produces a different confusion matrix.

Therefore, every threshold produces different:

$$
TPR
$$

and:

$$
FPR
$$

These points are plotted to create the ROC curve.

---

# 3. Effect of Changing Threshold

### High threshold

The model is strict about predicting positive.

Usually:

$$
TPR\downarrow
$$

and:

$$
FPR\downarrow
$$

### Low threshold

The model becomes more willing to predict positive.

Usually:

$$
TPR\uparrow
$$

and:

$$
FPR\uparrow
$$

Therefore there is a tradeoff between detecting positives and producing false positives.

---

# 4. What Does a Good ROC Curve Look Like?

We want:

$$
TPR\rightarrow1
$$

and:

$$
FPR\rightarrow0
$$

Therefore, the ideal point is:

$$
\boxed{
(FPR,TPR)=(0,1)
}
$$

The closer the ROC curve is to the **top-left corner**, the better the classifier generally separates the classes.

---

# 5. Random Classifier

A random classifier produces approximately:

$$
TPR=FPR
$$

which creates the diagonal line:

$$
\boxed{
TPR=FPR
}
$$

Its AUC is:

$$
\boxed{
AUC=0.5
}
$$

---

# AUC

## 1. What is AUC?

AUC stands for **Area Under the ROC Curve**.

It summarizes the ROC curve into a single number.

$$
\boxed{
AUC=\text{Area Under the ROC Curve}
}
$$

---

## 2. AUC Range

$$
\boxed{
0\leq AUC\leq1
}
$$

### Perfect classifier

$$
\boxed{
AUC=1
}
$$

### Random classifier

$$
\boxed{
AUC=0.5
}
$$

### Worse than random

$$
\boxed{
AUC<0.5
}
$$

A classifier with AUC below $0.5$ may perform better if its predictions are reversed.

---

# 3. AUC Intuition

A very useful interpretation:

$$
\boxed{
AUC
=
P(\text{random positive gets a higher score than random negative})
}
$$

For example:

$$
AUC=0.90
$$

means there is approximately a:

$$
\boxed{
90\%
}
$$

chance that a randomly selected positive example receives a higher model score than a randomly selected negative example.

---

# 4. How is AUC Calculated?

The basic process is:

$$
\boxed{
Predictions
\rightarrow
Different\ Thresholds
\rightarrow
TPR,FPR
\rightarrow
ROC
\rightarrow
Area
\rightarrow
AUC
}
$$

For neighboring ROC points:

$$
(x_1,y_1)
$$

and:

$$
(x_2,y_2)
$$

the trapezoidal area is:

$$
\boxed{
Area=
\frac{(x_2-x_1)(y_1+y_2)}{2}
}
$$

where:

$$
x=FPR
$$

and:

$$
y=TPR
$$

The areas of all the trapezoids are added:

$$
\boxed{
AUC=\sum_i Area_i
}
$$

---

# ROC vs AUC

| ROC | AUC |
|---|---|
| Curve | Single number |
| Plots TPR vs FPR | Measures area under ROC |
| Shows behavior at different thresholds | Summarizes overall discrimination |
| No single number | Usually between 0 and 1 |

# Important Interview Questions

### What is ROC?

$$
\boxed{
ROC=\text{TPR plotted against FPR at different thresholds}
}
$$

### What is AUC?

$$
\boxed{
AUC=\text{Area Under the ROC Curve}
}
$$

### Why do we change the threshold?

Because each threshold produces different predictions and therefore different TPR and FPR values.

### What does AUC = 0.5 mean?

The model is approximately equivalent to random guessing.

### What does AUC = 1 mean?

The model perfectly separates positive and negative examples.

### Why can ROC-AUC be useful compared with accuracy?

Accuracy depends on one chosen threshold, while ROC-AUC evaluates the model's ranking/separation ability across thresholds.
