# 1. What is PCA?

PCA stands for **Principal Component Analysis**.

PCA is a dimensionality-reduction technique.

Suppose our dataset has:

$$
100\ features
$$

PCA tries to represent the important information using fewer dimensions:

$$
100\ features
\rightarrow
10\ principal\ components
$$

The main idea is:

$$
\boxed{
Find\ directions\ where\ the\ data\ varies\ the\ most
}
$$

These directions are called **principal components**.

---

# 2. Why Do We Need PCA?

Suppose we have two highly correlated features:

$$
Height
$$

and:

$$
Weight
$$

Both may contain similar information.

PCA can combine their information into a new direction:

$$
PC_1
$$

instead of keeping both original features separately.

This can:

- Reduce dimensionality
- Remove redundancy
- Reduce computation
- Help visualization
- Sometimes reduce noise

---

# 3. Step 1: Center the Data

PCA first centers each feature around zero.

Suppose:

$$
X=
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix}
$$

Mean:

$$
\mu=
\frac{1+2+3}{3}
=
2
$$

Subtract the mean from every value:

$$
X_{centered}
=
\begin{bmatrix}
1-2\\
2-2\\
3-2
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
-1\\
0\\
1
\end{bmatrix}
$$

Therefore:

$$
\boxed{
X_{centered}=X-\mu
}
$$

---

# 4. Why Do We Center the Data?

PCA is interested in **variation around the mean**.

Without centering, the location of the data can affect the calculation.

Centering makes:

$$
Mean=0
$$

so PCA focuses on how the data varies around its center.

---

# 5. Step 2: Covariance Matrix

After centering, PCA calculates the covariance matrix.

For centered data matrix $X$:

$$
\boxed{
C=
\frac{1}{n-1}X^TX
}
$$

where:

- $X$ = centered data
- $n$ = number of samples
- $C$ = covariance matrix

The covariance matrix tells us how features vary with each other.

---

# 6. What Does the Covariance Matrix Tell Us?

Suppose:

$$
C=
\begin{bmatrix}
4 & 3\\
3 & 9
\end{bmatrix}
$$

The diagonal contains variance:

$$
C_{11}=4
$$

and:

$$
C_{22}=9
$$

The off-diagonal values contain covariance:

$$
C_{12}=3
$$

and:

$$
C_{21}=3
$$

Therefore:

$$
\boxed{
Diagonal
\rightarrow
Variance
}
$$

$$
\boxed{
Off\text{-}Diagonal
\rightarrow
Covariance
}
$$

---

# 7. Why Does the Covariance Matrix Matter in PCA?

PCA wants to find the directions where the data has the greatest variance.

The covariance matrix contains exactly the information about:

$$
\boxed{
How\ much\ each\ feature\ varies
}
$$

and:

$$
\boxed{
How\ features\ vary\ together
}
$$

So PCA uses the covariance matrix to find the important directions.

---

# 8. Step 3: Eigenvalues and Eigenvectors

Now we perform eigen decomposition of the covariance matrix:

$$
\boxed{
Cv=\lambda v
}
$$

where:

- $C$ = covariance matrix
- $v$ = eigenvector
- $\lambda$ = eigenvalue

This is the key mathematical step of PCA.

---

# 9. What Does the Eigenvector Mean in PCA?

An eigenvector represents a **direction**.

In PCA:

$$
\boxed{
Eigenvector
\rightarrow
Principal\ Component\ Direction
}
$$

For example:

$$
v_1=
\begin{bmatrix}
0.8\\
0.6
\end{bmatrix}
$$

represents one direction in the feature space.

Another eigenvector may represent another perpendicular direction.

---

# 10. What Does the Eigenvalue Mean?

The eigenvalue tells us **how much variance exists along its corresponding eigenvector**.

Therefore:

$$
\boxed{
Large\ Eigenvalue
\rightarrow
Large\ Variance
}
$$

$$
\boxed{
Small\ Eigenvalue
\rightarrow
Small\ Variance
}
$$

This is extremely important for PCA.

---

# 11. How Does PCA Choose Principal Components?

Suppose the covariance matrix gives us:

| Eigenvector | Eigenvalue |
|---|---:|
| $v_1$ | $10$ |
| $v_2$ | $3$ |
| $v_3$ | $1$ |

We sort eigenvalues from largest to smallest:

$$
10>3>1
$$

Therefore:

$$
v_1
$$

is the first principal component.

Then:

$$
v_2
$$

is the second principal component.

Then:

$$
v_3
$$

is the third.

So:

$$
\boxed{
Largest\ Eigenvalue
\rightarrow
Most\ Important\ Principal\ Component
}
$$

---

# 12. Why Does PCA Use the Largest Eigenvalue?

Because the eigenvalue represents variance along that direction.

Suppose:

$$
\lambda_1=10
$$

and:

$$
\lambda_2=2
$$

Then the first direction contains much more variation in the data.

PCA wants to preserve as much information/variation as possible.

Therefore it keeps:

$$
v_1
$$

before:

$$
v_2
$$

---

# 13. Step 4: Projection

After selecting the principal components, we project the original centered data onto them.

Suppose:

$$
X_{centered}
$$

is our centered data and:

$$
W
$$

contains the selected eigenvectors.

Then:

$$
\boxed{
Z=X_{centered}W
}
$$

where:

$$
Z
$$

is the lower-dimensional representation.

---

# 14. Simple Projection Example

Suppose one data point is:

$$
x=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

and the principal direction is:

$$
v=
\begin{bmatrix}
0.6\\
0.8
\end{bmatrix}
$$

The projection value is the dot product:

$$
x\cdot v
=
(3)(0.6)+(4)(0.8)
$$

$$
=1.8+3.2
$$

$$
\boxed{
=5
}
$$

So the original 2D point can be represented by the single value:

$$
5
$$

along that principal direction.

---

}
$$

# 16. Explained Variance

Suppose eigenvalues are:

$$
\lambda_1=6
$$

$$
\lambda_2=3
$$

$$
\lambda_3=1
$$

Total variance:

$$
6+3+1=10
$$

Explained variance ratio of the first component:

$$
\boxed{
EVR_1=
\frac{\lambda_1}
{\sum_i\lambda_i}
}
$$

Therefore:

$$
EVR_1=
\frac{6}{10}
$$

$$
\boxed{
EVR_1=60\%
}
$$

The first component explains $60\%$ of the total variance.

For the first two components:

$$
EVR_{1,2}
=
\frac{6+3}{10}
$$

$$
\boxed{
EVR_{1,2}=90\%
}
$$

Therefore we could reduce:

$$
3\ dimensions
\rightarrow
2\ dimensions
$$

while retaining $90\%$ of the variance.

---

# 17. How Do We Choose $k$?

We usually look at **cumulative explained variance**.

For example:

| Components | Explained Variance |
|---|---:|
| 1 | $60\%$ |
| 2 | $90\%$ |
| 3 | $100\%$ |

If we want to preserve approximately $90\%$ of the variance:

$$
\boxed{
k=2
}
$$

---

# 18. Why Are Eigenvectors Perpendicular?

For a covariance matrix, which is symmetric:

$$
C=C^T
$$

the eigenvectors can be chosen to be orthogonal.

Therefore principal components represent different, non-overlapping directions.

This helps PCA create new axes that are not redundant.

---

# 19. PCA and Correlated Features

Suppose:

$$
Feature_1
$$

and:

$$
Feature_2
$$

are highly correlated.

The data may look roughly like a diagonal line.

PCA finds a direction along that line.

That direction becomes:

$$
PC_1
$$

The second direction is perpendicular to it:

$$
PC_2
$$

If most variation occurs along $PC_1$, then $PC_2$ may contain very little information.

Therefore we can remove $PC_2$.

This is how PCA reduces dimensionality.

---

# 20. PCA Does Not Simply Remove One Original Feature

This is an important point.

PCA creates **new features**.

For example:

$$
PC_1
=
0.8X_1+0.6X_2
$$

and:

$$
PC_2
=
-0.6X_1+0.8X_2
$$

So PCA transforms:

$$
X_1,X_2
$$

into:

$$
PC_1,PC_2
$$

Then we may keep only:

$$
PC_1
$$

Therefore:

$$
\boxed{
PCA
\rightarrow
Creates\ new\ features
}
$$

It does not simply select the best original features.

---

# 21. PCA vs Feature Selection

### Feature Selection

Keeps original features.

Example:

$$
[X_1,X_2,X_3,X_4]
\rightarrow
[X_1,X_3]
$$

### PCA

Creates new features:

$$
[X_1,X_2,X_3,X_4]
\rightarrow
[PC_1,PC_2]
$$

Therefore:

$$
\boxed{
Feature\ Selection
\rightarrow
Keep\ original\ features
}
$$

$$
\boxed{
PCA
\rightarrow
Create\ new\ combinations
}
$$

---

# 22. Advantages of PCA

- Reduces dimensionality
- Reduces redundant information
- Can improve computational efficiency
- Useful for visualization
- Can reduce noise
- Helps with highly correlated features

---

# 23. Disadvantages of PCA

### 1. Loss of interpretability

Original features may have clear meanings.

But:

$$
PC_1=0.7X_1+0.4X_2-0.5X_3
$$

is harder to interpret.

### 2. Information can be lost

If we remove components:

$$
PC_3,PC_4,\ldots
$$

we lose some variance/information.

### 3. Sensitive to scaling

If features have very different scales, large-scale features can dominate the covariance matrix.

Therefore PCA is commonly applied after feature scaling when appropriate.

### 4. PCA is unsupervised

PCA does not use the target variable $y$.

It only looks at the structure of $X$.

---

# 24. Important Interview Questions

### What is PCA?

PCA is a dimensionality-reduction technique that transforms the original features into new orthogonal directions called principal components, ordered by the amount of variance they explain.

### Why do we center data before PCA?

Because PCA analyzes variation around the mean, so centering makes each feature have mean zero.

### Why do we calculate the covariance matrix?

It tells us how features vary individually and how they vary together.

### What does an eigenvector represent in PCA?

$$
\boxed{
Eigenvector
\rightarrow
Direction\ of\ a\ Principal\ Component
}
$$

### What does an eigenvalue represent?

$$
\boxed{
Eigenvalue
\rightarrow
Variance\ captured\ along\ that\ direction
}
$$

### Why do we sort eigenvalues?

To identify the directions containing the most variance.

### Why do we choose the largest eigenvalues?

Because larger eigenvalues correspond to directions containing more variance.

### How do you reduce 100 features to 10 using PCA?

1. Center the data.
2. Calculate the covariance matrix.
3. Calculate eigenvalues and eigenvectors.
4. Sort eigenvalues in descending order.
5. Select the top 10 eigenvectors.
6. Project the data onto those 10 directions.

### Does PCA select existing features?

No.

It creates new features called principal components.

### What is explained variance ratio?

$$
\boxed{
EVR_i=
\frac{\lambda_i}
{\sum_j\lambda_j}
}
$$

It tells us what fraction of the total variance is explained by principal component $i$.
