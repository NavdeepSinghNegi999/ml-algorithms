# 1. What is a Probability Distribution?

A probability distribution describes **how probabilities are assigned to the possible values of a random variable**.

For example, when throwing a die:

$$
X\in\{1,2,3,4,5,6\}
$$

For a fair die:

$$
P(X=x)=\frac{1}{6}
$$

A distribution tells us what values can occur and how likely each value is.

There are two main types:

$$
\boxed{\text{Discrete}}
$$

and

$$
\boxed{\text{Continuous}}
$$

### Discrete

Values are countable.

Examples:

- Bernoulli
- Binomial
- Poisson

### Continuous

Values can take any value within a range.

Examples:

- Gaussian
- Uniform
- Exponential

# 2. Bernoulli Distribution

## What?

Bernoulli distribution models **one trial with only two possible outcomes**.

Usually:

$$
X\in\{0,1\}
$$

For example:

- $1$ = success
- $0$ = failure

Let:

$$
P(X=1)=p
$$

Then:

$$
P(X=0)=1-p
$$

## PMF

$$
\boxed{
P(X=x)=p^x(1-p)^{1-x}
}
$$

where:

$$
x\in\{0,1\}
$$

## Mean

$$
\boxed{
E[X]=p
}
$$

## Variance

$$
\boxed{
Var(X)=p(1-p)
}
$$

## Example

If a model correctly classifies an image with probability:

$$
p=0.9
$$

then:

$$
P(X=1)=0.9
$$

and:

$$
P(X=0)=0.1
$$

## ML Connection

Binary classification is naturally connected to Bernoulli outcomes:

$$
Y\in\{0,1\}
$$

Logistic regression models:

$$
P(Y=1|X)
$$

using the sigmoid function.

---

# 3. Binomial Distribution

## What?

Binomial distribution models the **number of successes in $n$ independent Bernoulli trials**.

Conditions:

1. Fixed number of trials $n$
2. Each trial has two outcomes
3. Trials are independent
4. Probability of success $p$ remains constant

If:

$$
X=\text{number of successes}
$$

then:

$$
X\in\{0,1,\ldots,n\}
$$

## PMF

$$
\boxed{
P(X=k)
=
\binom{n}{k}
p^k(1-p)^{n-k}
}
$$

where:

$$
\binom{n}{k}
=
\frac{n!}{k!(n-k)!}
$$

## Why $\binom{n}{k}$?

Because the $k$ successes can occur in different positions among the $n$ trials.

## Example

Suppose a classifier has:

$$
p=0.8
$$

accuracy on each independent prediction.

For 5 predictions, probability of exactly 4 correct predictions:

$$
P(X=4)
=
\binom54(0.8)^4(0.2)^1
$$

$$
=
5(0.4096)(0.2)
$$

$$
=0.4096
$$

## Mean

$$
\boxed{
E[X]=np
}
$$

## Variance

$$
\boxed{
Var(X)=np(1-p)
}
$$

---

# 4. Bernoulli vs Binomial

| | Bernoulli | Binomial |
|---|---|---|
| Number of trials | $1$ | $n$ |
| Possible output | $0$ or $1$ | $0,1,\ldots,n$ |
| Example | One coin flip | Number of heads in 10 flips |
| Parameter | $p$ | $n,p$ |

Important relationship:

$$
\boxed{
Bernoulli=\text{Binomial with }n=1
}
$$

# 5. Gaussian / Normal Distribution

## What?

Gaussian distribution is a continuous distribution with a characteristic **bell-shaped curve**.

It is defined by:

- Mean $\mu$
- Variance $\sigma^2$

## PDF

$$
\boxed{
f(x)=
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
}
$$

where:

$$
\mu=\text{mean}
$$

and:

$$
\sigma^2=\text{variance}
$$

## What Does $\mu$ Do?

The mean controls the **center** of the distribution.

Increasing $\mu$ shifts the distribution left or right.

## What Does $\sigma$ Do?

Standard deviation controls the **spread**.

Small $\sigma$:

$$
\text{narrow distribution}
$$

Large $\sigma$:

$$
\text{wide distribution}
$$

## Important Property

For a normal distribution approximately:

$$
68\%
$$

of observations lie within:

$$
\mu\pm\sigma
$$

Approximately:

$$
95\%
$$

lie within:

$$
\mu\pm2\sigma
$$

Approximately:

$$
99.7\%
$$

lie within:

$$
\mu\pm3\sigma
$$

This is called the **68-95-99.7 rule**.

---

# 6. Standard Normal Distribution

A standard normal distribution has:

$$
\boxed{
\mu=0,\qquad\sigma=1
}
$$

We convert a normal variable to a standard normal variable using the Z-score:

$$
\boxed{
Z=\frac{X-\mu}{\sigma}
}
$$

This tells us how many standard deviations a value is away from the mean.

---

# 7. Why Gaussian Is Important in ML

Gaussian distributions appear in:

- Statistical modeling
- Linear regression assumptions
- Gaussian Naive Bayes
- Gaussian mixture models
- Noise modeling
- Generative models
- Hypothesis testing
- Confidence intervals

Many real-world measurements are approximately normally distributed, although **not every dataset is Gaussian**.

# 8. Uniform Distribution

## What?

Uniform distribution means every value in a specified range has equal probability density.

For:

$$
X\sim Uniform(a,b)
$$

the PDF is:

$$
\boxed{
f(x)=\frac{1}{b-a}
}
$$

for:

$$
a\leq x\leq b
$$

and:

$$
f(x)=0
$$

otherwise.

## Mean

$$
\boxed{
E[X]=\frac{a+b}{2}
}
$$

## Variance

$$
\boxed{
Var(X)=\frac{(b-a)^2}{12}
}
$$

## Example

For:

$$
X\sim Uniform(0,10)
$$

every value between $0$ and $10$ has equal density.

Mean:

$$
E[X]=\frac{0+10}{2}=5
$$

---

# 9. Exponential Distribution

## What?

Exponential distribution models the **waiting time until an event occurs** when events happen continuously at a constant rate.

For example:

- Time until a customer arrives
- Time until a machine fails
- Time between events

## PDF

For rate $\lambda>0$:

$$
\boxed{
f(x)=\lambda e^{-\lambda x}
}
$$

for:

$$
x\geq0
$$

## Mean

$$
\boxed{
E[X]=\frac{1}{\lambda}
}
$$

## Variance

$$
\boxed{
Var(X)=\frac{1}{\lambda^2}
}
$$

## Important Property: Memorylessness

The exponential distribution is memoryless:

$$
\boxed{
P(X>s+t\mid X>s)=P(X>t)
}
$$

Meaning the probability of waiting another $t$ units does not depend on how long you have already waited.

---

# 10. Poisson Distribution

## What?

Poisson distribution models the **number of events occurring within a fixed interval**, when events occur independently at a constant average rate.

Examples:

- Number of calls received in one hour
- Number of website requests per minute
- Number of defects per meter
- Number of arrivals per hour

Let:

$$
X=\text{number of events}
$$

and:

$$
\lambda=\text{average number of events in the interval}
$$

## PMF

$$
\boxed{
P(X=k)
=
\frac{e^{-\lambda}\lambda^k}{k!}
}
$$

where:

$$
k=0,1,2,\ldots
$$

## Mean

$$
\boxed{
E[X]=\lambda
}
$$

## Variance

$$
\boxed{
Var(X)=\lambda
}
$$

This is a very important property:

$$
\boxed{
Mean=Variance=\lambda
}
$$

---

# 11. Poisson Example

Suppose a server receives an average of:

$$
\lambda=3
$$

requests per minute.

Probability of exactly 2 requests in one minute:

$$
P(X=2)
=
\frac{e^{-3}3^2}{2!}
$$

$$
=
\frac{9e^{-3}}{2}
$$

$$
\approx0.224
$$

So there is approximately a:

$$
22.4\%
$$

chance of exactly 2 requests.

---

# 12. Poisson vs Exponential

These two are closely related.

### Poisson

Counts:

> How many events occur in an interval?

$$
X=\text{number of events}
$$

### Exponential

Measures:

> How long until the next event?

$$
X=\text{waiting time}
$$

Therefore:

$$
\boxed{
Poisson\rightarrow Count
}
$$

$$
\boxed{
Exponential\rightarrow Waiting\ Time
}
$$

# 14. Important Relationships

### Bernoulli → Binomial

A Binomial distribution is the sum of independent Bernoulli trials:

$$
X=X_1+X_2+\cdots+X_n
$$

where:

$$
X_i\sim Bernoulli(p)
$$

Therefore:

$$
\boxed{
X\sim Binomial(n,p)
}
$$

### Poisson → Exponential

If events follow a Poisson process:

$$
\boxed{
Poisson\rightarrow\text{number of events}
}
$$

while:

$$
\boxed{
Exponential\rightarrow\text{time between events}
}
$$

# 15. Discrete vs Continuous

## Discrete

Probability is assigned directly to individual values:

$$
P(X=k)
$$

Examples:

$$
Bernoulli,\ Binomial,\ Poisson
$$

## Continuous

Probability at one exact point is zero:

$$
P(X=x)=0
$$

Instead, probability is calculated over an interval:

$$
\boxed{
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
}
$$

Examples:

$$
Gaussian,\ Uniform,\ Exponential
$$

---

# 16. PMF vs PDF

### PMF

Probability Mass Function is used for **discrete** random variables.

$$
P(X=x)
$$

Example:

$$
P(X=k)
=
\binom nkp^k(1-p)^{n-k}
$$

### PDF

Probability Density Function is used for **continuous** random variables.

$$
f(x)
$$

Probability is obtained using integration:

$$
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
$$

Important:

For a continuous distribution, $f(x)$ itself is **density**, not the probability of exactly $x$.

# 17. Important Interview Questions

### What is the difference between Bernoulli and Binomial?

Bernoulli models one binary trial, while Binomial models the number of successes across multiple independent Bernoulli trials.

### What is the difference between Binomial and Poisson?

Binomial counts successes among a fixed number of trials. Poisson counts events occurring in a fixed interval.

### What is the difference between Poisson and Exponential?

Poisson models the number of events; Exponential models the waiting time between events.

### What are the parameters of a Gaussian distribution?

$$
\mu,\sigma^2
$$

### What is special about a Poisson distribution?

Its mean and variance are both:

$$
\lambda
$$

### What is the standard normal distribution?

A Gaussian distribution with:

$$
\mu=0,\qquad\sigma=1
$$

### What is the 68-95-99.7 rule?

For a Gaussian distribution, approximately:

$$
68\%\rightarrow\mu\pm\sigma
$$

$$
95\%\rightarrow\mu\pm2\sigma
$$

$$
99.7\%\rightarrow\mu\pm3\sigma
$$

### What is the difference between PMF and PDF?

PMF gives probabilities for discrete values, while PDF gives probability density for continuous values.
