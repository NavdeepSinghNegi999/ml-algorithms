# 1. What is a Probability Distribution?

A **probability distribution** tells us:

> **What values a random variable can take and how likely each value is.**

For example, when we throw a die:

$$
X\in\{1,2,3,4,5,6\}
$$

For a fair die, every number has the same probability:

$$
P(X=x)=\frac{1}{6}
$$

So the distribution tells us:

- Possible values → $1,2,3,4,5,6$
- Probability of each value → $\frac{1}{6}$

There are two main types:

$$
\boxed{\text{Discrete}}
$$

and

$$
\boxed{\text{Continuous}}
$$

### Discrete

The possible values are **countable**.

Examples:

- Bernoulli
- Binomial
- Poisson

For example:

$$
X=0,1,2,3,\ldots
$$

### Continuous

The variable can take **any value within a range**.

Examples:

- Gaussian
- Uniform
- Exponential

For example, height could be:

$$
170.2,\ 170.25,\ 170.251,\ldots
$$

There can be infinitely many possible values.

# 2. Bernoulli Distribution

## What?

**Bernoulli = one trial with only two possible outcomes.**

A **trial** simply means one attempt or one experiment.

For example:

- One coin flip → Head or Tail
- One ML prediction → Correct or Wrong
- One email → Spam or Not Spam

Usually we represent the outcomes as:

$$
X\in\{0,1\}
$$

where:

$$
1=\text{success}
$$

$$
0=\text{failure}
$$

If the probability of success is $p$:

$$
P(X=1)=p
$$

Then the probability of failure is:

$$
P(X=0)=1-p
$$

## PMF

**PMF = Probability Mass Function.**

It gives the **probability of each possible value of a discrete random variable**.

For Bernoulli:

$$
\boxed{
P(X=x)=p^x(1-p)^{1-x}
}
$$

where:

$$
x\in\{0,1\}
$$

This formula gives:

$$
P(X=1)=p
$$

and:

$$
P(X=0)=1-p
$$

## Mean

$$
\boxed{E[X]=p}
$$

## Variance

$$
\boxed{Var(X)=p(1-p)}
$$

## Example

Suppose a model correctly classifies an image with probability:

$$
p=0.9
$$

Then:

$$
P(X=1)=0.9
$$

and:

$$
P(X=0)=0.1
$$

So one prediction is a **Bernoulli trial**.

## ML Connection

Binary classification naturally has two outcomes:

$$
Y\in\{0,1\}
$$

Therefore, the target can be modeled using a Bernoulli distribution.

Logistic regression models:

$$
P(Y=1|X)
$$

using the sigmoid function:

$$
\boxed{
p=\frac{1}{1+e^{-z}}
}
$$

So remember:

$$
\boxed{\text{Bernoulli}\rightarrow\text{one binary outcome}}
$$

# 3. Binomial Distribution

## What?

**Binomial = number of successes in multiple Bernoulli trials.**

The easiest way to understand it:

### Bernoulli

One prediction:

> Was it correct or wrong?

### Binomial

Many predictions:

> Out of 10 predictions, how many were correct?

So:

$$
\boxed{\text{Bernoulli = one trial}}
$$

$$
\boxed{\text{Binomial = multiple trials + count successes}}
$$

## Conditions

For a Binomial distribution:

1. Fixed number of trials $n$
2. Each trial has two outcomes
3. Trials are independent
4. Probability of success $p$ remains constant

If:

$$
X=\text{number of successes}
$$

then:

$$
X\in\{0,1,\ldots,n\}
$$

## PMF

The Binomial PMF is:

$$
\boxed{
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
}
$$

where:

- $n$ = total number of trials
- $k$ = number of successes we want
- $p$ = probability of success
- $1-p$ = probability of failure

And:

$$
\binom{n}{k}
=
\frac{n!}{k!(n-k)!}
$$

## Why $\binom{n}{k}$?

Because the $k$ successes can occur in **different positions** among the $n$ trials.

## Example

Suppose a classifier has:

$$
p=0.8
$$

accuracy on each independent prediction.

We make:

$$
n=5
$$

predictions.

What is the probability of getting exactly 4 correct?

Here:

$$
k=4
$$

So:

$$
P(X=4)
=
\binom{5}{4}(0.8)^4(0.2)^1
$$

$$
=5(0.4096)(0.2)
$$

$$
=0.4096
$$

Therefore:

$$
\boxed{P(X=4)=0.4096}
$$

or:

$$
\boxed{40.96\%}
$$

## Mean

$$
\boxed{E[X]=np}
$$

## Variance

$$
\boxed{Var(X)=np(1-p)}
$$

# 4. Bernoulli vs Binomial

| | Bernoulli | Binomial |
|---|---|---|
| Number of trials | $1$ | $n$ |
| Possible output | $0$ or $1$ | $0,1,\ldots,n$ |
| Example | One coin flip | Number of heads in 10 flips |
| Parameter | $p$ | $n,p$ |

### Important Relationship

A Bernoulli distribution is a special case of Binomial:

$$
\boxed{
\text{Bernoulli}=\text{Binomial with }n=1
}
$$

# 5. Gaussian / Normal Distribution

## What?

The **Gaussian or Normal distribution** is a continuous distribution with a **bell-shaped curve**.

The center is the mean, and values near the center are more common.

It is defined by:

- Mean: $\mu$
- Variance: $\sigma^2$

### PDF

**PDF = Probability Density Function.**

For a Normal distribution:

$$
f(x) = \frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

where:

$$
\mu = \text{mean}
$$

and:

$$
\sigma^2 = \text{variance}
$$

### What does $\mu$ do?

The mean controls the **center** of the distribution.

Changing $\mu$ moves the whole curve left or right.

### What does $\sigma$ do?

Standard deviation controls the **spread**.

- Small $\sigma$: narrow distribution
- Large $\sigma$: wide distribution

### Important Property: 68-95-99.7 Rule

For a Normal distribution:

Approximately:

$$
68\%
$$

of observations are within:

$$
\mu \pm \sigma
$$

Approximately:

$$
95\%
$$

are within:

$$
\mu \pm 2\sigma
$$

Approximately:

$$
99.7\%
$$

are within:

$$
\mu \pm 3\sigma
$$

This is called the:

$$
\boxed{68\text{-}95\text{-}99.7\text{ rule}}
$$


# 6. Standard Normal Distribution

A **standard normal distribution** is simply a Normal distribution with:

$$
\mu = 0,\qquad \sigma = 1
$$

We convert a normal variable into a standard normal variable using the **Z-score**:

$$
Z = \frac{X-\mu}{\sigma}
$$

### What does Z-score mean?

It tells us:

> **How many standard deviations a value is away from the mean.**

For example:

$$
Z = 2
$$

means the value is **2 standard deviations above the mean**.


# 7. Why Gaussian Is Important in ML

Gaussian distributions appear in:

- Statistical modeling
- Linear regression assumptions
- Gaussian Naive Bayes
- Gaussian Mixture Models
- Noise modeling
- Generative models
- Hypothesis testing
- Confidence intervals

Many real-world measurements are approximately Normal.

But remember:

> **Not every dataset follows a Gaussian distribution.**


# 8. Uniform Distribution

## What?

Uniform distribution means:

> **Every value in a specified range has equal probability density.**

For:

$$
X \sim \text{Uniform}(a,b)
$$

the PDF is:

$$
f(x) = \frac{1}{b-a}
$$

for:

$$
a \leq x \leq b
$$

and:

$$
f(x) = 0
$$

otherwise.

The shape is flat.

Unlike Normal distribution, there is **no center that is more likely**.

### Mean

$$
E[X] = \frac{a+b}{2}
$$

### Variance

$$
\operatorname{Var}(X) = \frac{(b-a)^2}{12}
$$

### Example

Suppose:

$$
X \sim \text{Uniform}(0,10)
$$

Every value between 0 and 10 has equal density.

Mean:

$$
E[X] = \frac{0+10}{2} = 5
$$

So:

$$
E[X] = 5
$$

### Easy Difference from Normal

- **Normal** $\rightarrow$ values near the center are more likely.
- **Uniform** $\rightarrow$ all values in the range have equal density.


# 9. Exponential Distribution

## What?

Exponential distribution models:

> **The waiting time until the next event happens.**

It is commonly used when events happen continuously at a constant average rate.

Examples:

- Time until a customer arrives
- Time until a machine fails
- Time between events

So remember:

$$
\boxed{\text{Exponential} \rightarrow \text{Waiting Time}}
$$

### PDF

For rate:

$$
\lambda > 0
$$

the PDF is:

$$
f(x) = \lambda e^{-\lambda x}
$$

for:

$$
x \geq 0
$$

### Mean

$$
E[X] = \frac{1}{\lambda}
$$

### Variance

$$
\operatorname{Var}(X) = \frac{1}{\lambda^2}
$$

### Important Property: Memorylessness

Exponential distribution is **memoryless**:

$$
P(X>s+t \mid X>s) = P(X>t)
$$

In simple words:

> **How much longer you have to wait does not depend on how long you have already waited.**


# 10. Poisson Distribution

## What?

Poisson distribution models:

> **The number of events occurring in a fixed interval.**

The interval can be:

- Time $\rightarrow$ per hour/minute
- Distance $\rightarrow$ per meter
- Area $\rightarrow$ per square meter
- etc.

Examples:

- Number of calls received in one hour
- Number of website requests per minute
- Number of defects per meter
- Number of arrivals per hour

Let:

$$
X = \text{number of events}
$$

and:

$$
\lambda = \text{average number of events in the interval}
$$

### PMF

$$
P(X=k) = \frac{e^{-\lambda}\lambda^k}{k!}
$$

where:

$$
k = 0,1,2,\ldots
$$

### Mean

$$
E[X] = \lambda
$$

### Variance

$$
\operatorname{Var}(X) = \lambda
$$

This is a very important property:

$$
\boxed{\text{Mean} = \text{Variance} = \lambda}
$$


# 11. Poisson Example

Suppose a server receives an average of:

$$
\lambda = 3
$$

requests per minute.

What is the probability of exactly 2 requests in one minute?

Here:

$$
k = 2
$$

Use the Poisson PMF:

$$
P(X=2) = \frac{e^{-3}3^2}{2!}
$$

$$
= \frac{9e^{-3}}{2}
$$

$$
\approx 0.224
$$

Therefore:

$$
P(X=2) \approx 22.4\%
$$

So there is approximately a **22.4% chance of exactly 2 requests**.


# 12. Poisson vs Exponential

These two are closely related, but they answer **different questions**.

## Poisson

Asks:

> **How many events occur in an interval?**

$$
X = \text{number of events}
$$

Example:

> How many customers arrive in one hour?

Therefore:

$$
\boxed{\text{Poisson} \rightarrow \text{Count}}
$$

## Exponential

Asks:

> **How long until the next event?**

$$
X = \text{waiting time}
$$

Example:

> How long until the next customer arrives?

Therefore:

$$
\boxed{\text{Exponential} \rightarrow \text{Waiting Time}}
$$

## Easy Memory

> **Poisson = How many?**

> **Exponential = How long?**

# 14. Important Relationships

## Bernoulli $\rightarrow$ Binomial

A Binomial distribution is the **sum of independent Bernoulli trials**:

$$
X = X_1 + X_2 + \cdots + X_n
$$

where:

$$
X_i \sim \text{Bernoulli}(p)
$$

Therefore:

$$
X \sim \text{Binomial}(n,p)
$$

### Simple Example

Suppose each prediction is:

$$
X_i =
\begin{cases}
1 & \text{correct} \\
0 & \text{wrong}
\end{cases}
$$

Each prediction is Bernoulli.

If we add the results of 10 predictions:

$$
X = X_1 + X_2 + \cdots + X_{10}
$$

then $X$ tells us **how many predictions were correct**.

That is Binomial.


## Poisson $\rightarrow$ Exponential

When events follow a Poisson process:

$$
\text{Poisson} \rightarrow \text{number of events}
$$

while:

$$
\text{Exponential} \rightarrow \text{time between events}
$$

So they describe two views of the **same type of event process**:

> **Poisson counts the events, while Exponential measures the waiting time between them.**


# 15. Discrete vs Continuous

This is very important because it explains **PMF vs PDF**.

## Discrete

A discrete variable has **separate/countable values**.

For example:

$$
X = 0,1,2,3,\ldots
$$

We can directly ask:

$$
P(X=k)
$$

Examples:

- Bernoulli
- Binomial
- Poisson

For example:

> What is the probability of exactly 3 website requests?

That can have a direct probability.

---

## Continuous

A continuous variable can take **any value in a range**.

For example:

$$
X = 2.1,\ 2.11,\ 2.111,\ldots
$$

For a continuous random variable:

$$
P(X=x) = 0
$$

for one exact value.

Instead, we calculate the probability over an **interval**:

$$
P(a \leq X \leq b) = \int_a^b f(x)\,dx
$$

Examples:

- Gaussian
- Uniform
- Exponential


# 16. PMF vs PDF

This is an important interview topic.

## PMF

**PMF = Probability Mass Function**

Used for:

**Discrete random variables**

It gives the probability of a particular value:

$$
P(X=x)
$$

For example, Binomial:

$$
P(X=k) =
\binom{n}{k}p^k(1-p)^{n-k}
$$

Here, we can directly say:

$$
P(X=4)=0.4096
$$

because Binomial is discrete.

---

## PDF

**PDF = Probability Density Function**

Used for:

**Continuous random variables**

The PDF is written as:

$$
f(x)
$$

**Important:** $f(x)$ itself is **not the probability of exactly $x$**.

Instead, we use the PDF to calculate probability over an interval:

$$
P(a \leq X \leq b) = \int_a^b f(x)\,dx
$$

So for a continuous variable:

$$
P(X=x)=0
$$

but:

$$
P(a \leq X \leq b)
$$

can be greater than zero.

### Easy Way to Remember PMF vs PDF

$$
\boxed{\text{PMF} \rightarrow \text{Discrete} \rightarrow \text{Direct probability}}
$$

$$
\boxed{\text{PDF} \rightarrow \text{Continuous} \rightarrow \text{Area gives probability}}
$$

```text
                    Probability Distributions
                              |
                ┌─────────────┴─────────────┐
                |                           |
             Discrete                   Continuous
                |                           |
       ┌────────┼────────┐          ┌───────┼────────┐
       |        |        |          |       |        |
   Bernoulli Binomial Poisson    Gaussian Uniform Exponential
       |        |        |          |       |        |
      One     Number    Number     Bell    Equal   Waiting
     binary      of       of      curve   density   time
     trial    successes  events

# 17. Important Interview Questions

## 1. What is the difference between Bernoulli and Binomial?

**Bernoulli** models **one binary trial**, while **Binomial** models the number of successes across multiple independent Bernoulli trials.

### Easy way to remember:

> Bernoulli = **One trial**

> Binomial = **Many trials**

Example:

- Toss a coin once $\rightarrow$ Bernoulli
- Toss a coin 10 times and count heads $\rightarrow$ Binomial


## 2. What is the difference between Binomial and Poisson?

**Binomial** counts successes among a **fixed number of trials**.

**Poisson** counts events occurring in a **fixed interval**.

### Easy way to remember:

> Binomial = **How many successes in $n$ trials?**

> Poisson = **How many events in an interval?**

Example:

- Number of heads in 10 coin tosses $\rightarrow$ Binomial
- Number of website requests in one minute $\rightarrow$ Poisson


## 3. What is the difference between Poisson and Exponential?

**Poisson** models the **number of events**.

**Exponential** models the **waiting time between events**.

### Easy way to remember:

> Poisson = **How many?**

> Exponential = **How long?**

Example:

- Number of customers arriving in one hour $\rightarrow$ Poisson
- Time until the next customer arrives $\rightarrow$ Exponential


## 4. What are the parameters of a Gaussian distribution?

The two parameters are:

$$
\boxed{\mu,\sigma^2}
$$

where:

- $\mu$ = mean
- $\sigma^2$ = variance

The mean controls the **center**.

The standard deviation $\sigma$ controls the **spread**.


## 5. What is special about a Poisson distribution?

For a Poisson distribution:

$$
\boxed{E[X]=Var(X)=\lambda}
$$

So the **mean and variance are both equal to $\lambda$**.


## 6. What is the Standard Normal Distribution?

A Standard Normal distribution is a Gaussian distribution with:

$$
\boxed{\mu=0,\qquad\sigma=1}
$$

We convert a normal variable into a standard normal variable using the Z-score:

$$
Z=\frac{X-\mu}{\sigma}
$$


## 7. What is a Z-score?

A Z-score tells us:

> **How many standard deviations a value is away from the mean.**

For example:

$$
Z=2
$$

means the value is **2 standard deviations above the mean**.

Similarly:

$$
Z=-2
$$

means the value is **2 standard deviations below the mean**.


## 8. What is the 68-95-99.7 rule?

For a Gaussian distribution, approximately:

$$
68\% \rightarrow \mu\pm\sigma
$$

$$
95\% \rightarrow \mu\pm2\sigma
$$

$$
99.7\% \rightarrow \mu\pm3\sigma
$$

This means most values are close to the mean.


## 9. What is the difference between PMF and PDF?

**PMF** is used for **discrete random variables**.

It gives the probability of a particular value:

$$
P(X=x)
$$

**PDF** is used for **continuous random variables**.

It gives probability **density**, not the probability of an exact value.

Probability comes from the area under the PDF:

$$
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
$$

### Easy way to remember:

> PMF $\rightarrow$ Discrete $\rightarrow$ Direct probability

> PDF $\rightarrow$ Continuous $\rightarrow$ Area gives probability


## 10. Why is $P(X=x)=0$ for a continuous random variable?

A continuous variable can take infinitely many possible values.

Therefore, the probability of one exact value is:

$$
\boxed{P(X=x)=0}
$$

But the probability over a range can be greater than zero:

$$
P(a\leq X\leq b)>0
$$

For example, the probability that a person's height is **exactly 170.000000... cm** is theoretically zero.

But the probability that their height is between 169 cm and 171 cm can be greater than zero.


## 11. What is the difference between a parameter and a random variable?

A **random variable** represents the outcome of a random experiment.

Example:

$$
X=\text{number of heads}
$$

A **parameter** describes the distribution.

For example, in a Binomial distribution:

$$
X\sim Binomial(n,p)
$$

Here:

- $X$ = random variable
- $n$ = number of trials
- $p$ = probability of success


## 12. What is the difference between variance and standard deviation?

Variance measures the spread using **squared units**:

$$
Var(X)=\sigma^2
$$

Standard deviation is the square root of variance:

$$
\boxed{\sigma=\sqrt{Var(X)}}
$$

So:

> Variance = squared spread

> Standard deviation = original-unit spread


## 13. What does the mean of a distribution tell us?

The mean tells us the **average or central value** of the distribution.

For example, for a Gaussian distribution:

$$
E[X]=\mu
$$

So $\mu$ represents the center of the distribution.


## 14. What does variance tell us?

Variance tells us how **spread out the values are around the mean**.

Small variance:

> Values are close to the mean.

Large variance:

> Values are more spread out.


## 15. What is the difference between discrete and continuous distributions?

### Discrete

Values are **countable**.

Example:

$$
0,1,2,3,\ldots
$$

Examples:

- Bernoulli
- Binomial
- Poisson

### Continuous

Values can take **any value within a range**.

Examples:

$$
2.1,\ 2.11,\ 2.111,\ldots
$$

Examples:

- Gaussian
- Uniform
- Exponential


## 16. Which distributions are discrete?

The important discrete distributions we studied are:

$$
\boxed{
Bernoulli,\ Binomial,\ Poisson
}
$$

They use **PMF**.


## 17. Which distributions are continuous?

The important continuous distributions we studied are:

$$
\boxed{
Gaussian,\ Uniform,\ Exponential
}
$$

They use **PDF**.


## 18. What is the support of a distribution?

The **support** tells us the possible values a random variable can take.

Examples:

### Bernoulli

$$
X\in\{0,1\}
$$

### Binomial

$$
X\in\{0,1,2,\ldots,n\}
$$

### Poisson

$$
X\in\{0,1,2,\ldots\}
$$

### Exponential

$$
X\geq0
$$

This is useful because it tells us **where the distribution can exist**.


## 19. Why can't a probability be greater than 1?

Probability must always satisfy:

$$
\boxed{0\leq P(X=x)\leq1}
$$

For all possible outcomes, the total probability must be:

$$
\boxed{\sum P(X=x)=1}
$$

For continuous distributions:

$$
\boxed{\int_{-\infty}^{\infty}f(x)\,dx=1}
$$

This means the total area under a PDF is 1.


## 20. Can a PDF value be greater than 1?

**Yes.**

This is an important interview question.

A PDF value $f(x)$ represents **density**, not probability.

Therefore:

$$
f(x)>1
$$

is possible.

But the total area under the PDF must always be:

$$
\int_{-\infty}^{\infty}f(x)\,dx=1
$$

So:

> **Probability cannot be greater than 1, but PDF density can be greater than 1.**


# Quick Revision

## Bernoulli

$$
\boxed{
P(X=x)=p^x(1-p)^{1-x}
}
$$

$$
E[X]=p
$$

$$
Var(X)=p(1-p)
$$

**Remember:**

> One binary trial


## Binomial

$$
\boxed{
P(X=k)=
\binom{n}{k}p^k(1-p)^{n-k}
}
$$

$$
E[X]=np
$$

$$
Var(X)=np(1-p)
$$

**Remember:**

> Number of successes in $n$ trials


## Gaussian

$$
\boxed{
f(x)=
\frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
}
$$

$$
E[X]=\mu
$$

$$
Var(X)=\sigma^2
$$

**Remember:**

> Bell-shaped curve


## Uniform

$$
\boxed{
f(x)=\frac{1}{b-a}
}
$$

$$
E[X]=\frac{a+b}{2}
$$

$$
Var(X)=\frac{(b-a)^2}{12}
$$

**Remember:**

> Equal density over an interval


## Poisson

$$
\boxed{
P(X=k)=
\frac{e^{-\lambda}\lambda^k}{k!}
}
$$

$$
E[X]=Var(X)=\lambda
$$

**Remember:**

> Number of events in an interval


## Exponential

$$
\boxed{
f(x)=\lambda e^{-\lambda x}
}
$$

$$
E[X]=\frac{1}{\lambda}
$$

$$
Var(X)=\frac{1}{\lambda^2}
$$

**Remember:**

> Waiting time between events


# Final Mental Map

$$
\boxed{
\text{Bernoulli}\rightarrow\text{One Binary Trial}
}
$$

$$
\boxed{
\text{Binomial}\rightarrow\text{Number of Successes}
}
$$

$$
\boxed{
\text{Gaussian}\rightarrow\text{Bell Curve}
}
$$

$$
\boxed{
\text{Uniform}\rightarrow\text{Equal Density}
}
$$

$$
\boxed{
\text{Poisson}\rightarrow\text{Number of Events}
}
$$

$$
\boxed{
\text{Exponential}\rightarrow\text{Waiting Time}
}
$$

### Most Important Distinction

> **PMF is for discrete variables and gives probability directly.**

> **PDF is for continuous variables and gives density; probability comes from the area under the PDF.**
