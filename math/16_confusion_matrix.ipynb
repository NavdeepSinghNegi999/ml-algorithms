# 1. Confusion Matrix

A confusion matrix compares:

$$
Actual\ Class
\quad\text{vs}\quad
Predicted\ Class
$$

For binary classification:

| | Actual Positive | Actual Negative |
|---|---:|---:|
| Predicted Positive | $TP$ | $FP$ |
| Predicted Negative | $FN$ | $TN$ |

### TP — True Positive

$$
Actual=Positive,\qquad Prediction=Positive
$$

The model correctly predicts positive.

### TN — True Negative

$$
Actual=Negative,\qquad Prediction=Negative
$$

The model correctly predicts negative.

### FP — False Positive

$$
Actual=Negative,\qquad Prediction=Positive
$$

The model incorrectly predicts positive.

Also called:

$$
\boxed{Type\ I\ Error}
$$

### FN — False Negative

$$
Actual=Positive,\qquad Prediction=Negative
$$

The model misses a positive case.

Also called:

$$
\boxed{Type\ II\ Error}
$$

---

# 2. Accuracy

Accuracy answers:

> Out of all predictions, how many were correct?

$$
\boxed{
Accuracy=
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

---

# 3. Precision

Precision answers:

> Out of everything predicted as positive, how many were actually positive?

$$
\boxed{
Precision=
\frac{TP}
{TP+FP}
}
$$

Precision mainly cares about:

$$
\boxed{FP}
$$

High precision means:

$$
\boxed{
Few\ False\ Positives
}
$$

---

# 4. Recall

Recall answers:

> Out of all actual positive cases, how many did the model correctly find?

$$
\boxed{
Recall=
\frac{TP}
{TP+FN}
}
$$

Recall is also called:

$$
\boxed{
Recall=Sensitivity=TPR
}
$$

Recall mainly cares about:

$$
\boxed{FN}
$$

High recall means:

$$
\boxed{
Few\ False\ Negatives
}
$$

---

# 5. Specificity

Specificity answers:

> Out of all actual negative cases, how many did the model correctly identify as negative?

$$
\boxed{
Specificity=
\frac{TN}
{TN+FP}
}
$$

Specificity is also called:

$$
\boxed{
Specificity=TNR
}
$$

High specificity means:

$$
\boxed{
Few\ False\ Positives
}
$$

---

# 6. False Positive Rate

FPR answers:

> Out of all actual negative cases, how many did the model incorrectly classify as positive?

$$
\boxed{
FPR=
\frac{FP}
{FP+TN}
}
$$

Notice the relationship:

$$
\boxed{
FPR=1-Specificity
}
$$

because:

$$
Specificity=
\frac{TN}{TN+FP}
$$

and:

$$
FPR=
\frac{FP}{TN+FP}
$$

Therefore:

$$
Specificity+FPR=1
$$

---

# 7. False Negative Rate

FNR answers:

> Out of all actual positive cases, how many did the model incorrectly classify as negative?

$$
\boxed{
FNR=
\frac{FN}
{FN+TP}
}
$$

Since:

$$
Recall=
\frac{TP}{TP+FN}
$$

we get:

$$
\boxed{
FNR=1-Recall
}
$$

Therefore:

$$
Recall+FNR=1
$$

---

# 8. Important Relationships

$$
\boxed{
Recall=Sensitivity=TPR
}
$$

$$
\boxed{
Specificity=TNR
}
$$

$$
\boxed{
FPR=1-Specificity
}
$$

$$
\boxed{
FNR=1-Recall
}
$$

---

# 9. F1 Score

F1 combines:

$$
Precision
$$

and:

$$
Recall
$$

using their harmonic mean.

$$
\boxed{
F1=
2
\frac{Precision\times Recall}
{Precision+Recall}
}
$$

Using the confusion-matrix values directly:

$$
\boxed{
F1=
\frac{2TP}
{2TP+FP+FN}
}
$$

F1 is high only when both precision and recall are reasonably high.

---

# 10. Precision vs Recall vs Specificity

| Metric | Main Question | Main Concern |
|---|---|---|
| Precision | When I predict positive, am I correct? | $FP$ |
| Recall | Did I find the actual positives? | $FN$ |
| Specificity | Did I correctly identify the negatives? | $FP$ |

### If False Positives are costly

Focus on:

$$
\boxed{
Precision,\ Specificity
}
$$

### If False Negatives are costly

Focus on:

$$
\boxed{
Recall
}
$$

Suppose:

$$
TP=80
$$

$$

$$

$$

$$

$$

$$

Total:

$$
80+90+10+20=200
$$

$$
Accuracy=
\frac{80+90}{200}
$$

$$
\boxed{
Accuracy=0.85=85\%
}
$$

### Precision

$$
Precision=
\frac{80}{80+10}
$$

$$
\boxed{
Precision\approx88.9\%
}
$$

### Recall

$$
Recall=
\frac{80}{80+20}
$$

$$
\boxed{
Recall=80\%
}
$$

### Specificity

$$
Specificity=
\frac{90}{90+10}
$$

$$
\boxed{
Specificity=90\%
}
$$

### FPR

$$
FPR=
\frac{10}{10+90}
$$

$$
\boxed{
FPR=10\%
}
$$

### FNR

$$
FNR=
\frac{20}{20+80}
$$

$$
\boxed{
FNR=20\%
}
$$

### F1

$$
F1=
2
\frac{(0.889)(0.80)}
{0.889+0.80}
$$

$$
\boxed{
F1\approx84.2\%
}
$$

# 12. Precision-Recall Curve

A Precision-Recall curve shows:

$$
\boxed{
Precision\quad vs\quad Recall
}
$$

at different classification thresholds.

The process is:

$$
Model
\rightarrow
Probability
\rightarrow
Choose\ Threshold
\rightarrow
Predictions
\rightarrow
TP,FP,FN
$$

Then calculate:

$$
Precision
$$

and:

$$
Recall
$$

Change the threshold and repeat.

The resulting points create the PR curve.

---

# 13. Why Does Threshold Affect Precision and Recall?

Suppose the model outputs:

$$
[0.9,\ 0.8,\ 0.6,\ 0.4,\ 0.2]
$$

If:

$$
Threshold=0.8
$$

only high-probability examples are predicted positive.

If we reduce it to:

$$
Threshold=0.3
$$

more examples become positive.

Usually:

$$
Threshold\downarrow
\Rightarrow
Recall\uparrow
$$

because we find more positive examples.

But false positives can also increase:

$$
FP\uparrow
$$

which can reduce precision.

Therefore:

$$
\boxed{
Precision\leftrightarrow Recall
}
$$

is often a tradeoff.

---

# 14. When Is PR Curve Useful?

PR curves are particularly useful when the positive class is rare.

For example:

$$
990\ Negative
$$

and:

$$
10\ Positive
$$

In such a situation, accuracy and even some ROC measures can hide how well the model handles the rare positive class.

PR directly focuses on:

$$
Precision
$$

and:

$$
Recall
$$

Therefore:

$$
\boxed{
Rare\ Positive\ Class
\rightarrow
PR\ Curve\ can\ be\ very\ useful
}
$$

# 15. Interview Questions You Must Know

### What is the difference between precision and recall?

$$
Precision=
\frac{TP}{TP+FP}
$$

asks:

> Of my positive predictions, how many were correct?

$$
Recall=
\frac{TP}{TP+FN}
$$

asks:

> Of all actual positives, how many did I find?

---

### When would you prefer recall?

When:

$$
FN
$$

is more costly.

Example:

$$
Disease\ Detection
$$

---

### When would you prefer precision?

When:

$$
FP
$$

is more costly.

Example:

$$
Spam\ Detection
$$

---

### What is specificity?

$$
\boxed{
Specificity=
\frac{TN}{TN+FP}
}
$$

It measures how well the model identifies actual negatives.

---

### What is the relationship between FPR and specificity?

$$
\boxed{
FPR=1-Specificity
}
$$

---

### What is the relationship between FNR and recall?

$$
\boxed{
FNR=1-Recall
}
$$

---

### Why is F1 better than simply taking the average of precision and recall?

Because F1 uses the harmonic mean:

$$
F1=
2\frac{PR}{P+R}
$$

If either precision or recall is very low, F1 also becomes low.
