# 1. What is Optimization?

Optimization means finding the model parameters that **minimize the loss function**.

$$
\boxed{
\min_{\mathbf{w}}L(\mathbf{w})
}
$$

Example:

$$
L(w)=(w-3)^2
$$

The minimum occurs at:

$$
w=3
$$

In ML, we usually have many parameters:

$$
\mathbf{w}=[w_1,w_2,\ldots,w_n]
$$

so we use gradients to find a direction that reduces the loss.

---

# 2. Gradient Descent

## What?

Gradient Descent is an optimization algorithm that repeatedly moves parameters in the direction that **decreases the loss**.

$$
\boxed{
\mathbf{w}_{new}
=
\mathbf{w}_{old}
-
\eta\nabla L(\mathbf{w})
}
$$

Where:

- $\mathbf{w}$ = parameters
- $\eta$ = learning rate
- $\nabla L$ = gradient of loss

## Why subtract?

The gradient points toward the direction of **steepest increase**.

Therefore:

$$
-\nabla L
$$

points toward the direction of steepest decrease.

## Example

Suppose:

$$
L(w)=w^2
$$

Then:

$$
\frac{dL}{dw}=2w
$$

Let:

$$
w=5,\qquad\eta=0.1
$$

Gradient:

$$
2(5)=10
$$

Update:

$$
w_{new}=5-(0.1)(10)=4
$$

The parameter moves:

$$
5\rightarrow4\rightarrow3.2\rightarrow\cdots\rightarrow0
$$

---

# 3. Batch Gradient Descent

Uses the **entire training dataset** to calculate one gradient update.

$$
\boxed{
\mathbf{w}
\leftarrow
\mathbf{w}
-
\eta\nabla L_{\text{all}}(\mathbf{w})
}
$$

### Advantages

- Stable gradient
- Smooth updates
- Deterministic

### Disadvantages

- Expensive for huge datasets
- Requires processing the entire dataset for every update
- Can be slow

---

# 4. Stochastic Gradient Descent (SGD)

## What?

Instead of using the entire dataset, SGD uses **one training example** at a time.

$$
\boxed{
\mathbf{w}
\leftarrow
\mathbf{w}
-
\eta\nabla L_i(\mathbf{w})
}
$$

where $L_i$ is the loss for one sample.

### Advantages

- Faster updates
- Can work well with large datasets
- Noise can sometimes help escape shallow local regions

### Disadvantages

- Updates are noisy
- Loss may fluctuate
- Can take longer to settle near the minimum

---

# 5. Mini-Batch Gradient Descent

In practice, we commonly use a **small batch** rather than the entire dataset or just one sample.

For batch size $m$:

$$
\boxed{
\mathbf{w}
\leftarrow
\mathbf{w}
-
\eta
\frac{1}{m}
\sum_{i=1}^{m}
\nabla L_i(\mathbf{w})
}
$$

Example:

$$
Batch\ Size=32
$$

means the gradient is calculated using 32 samples before updating the weights.

### Comparison

| Method | Data per Update | Gradient | Speed |
|---|---|---|---|
| Batch GD | Entire dataset | Stable | Slow for large data |
| SGD | 1 sample | Noisy | Fast updates |
| Mini-batch | Small batch | Moderately stable | Common in DL |

# 6. Learning Rate

## What?

The learning rate $\eta$ controls **how large each optimization step is**.

$$
\mathbf{w}_{new}
=
\mathbf{w}_{old}
-
\eta\nabla L
$$

## Why is it important?

Because the gradient tells us the direction, while the learning rate controls **how far we move**.

### Learning Rate Too Small

$$
\eta\rightarrow0
$$

Updates become tiny.

Result:

- Training is very slow
- May take many iterations

### Learning Rate Too Large

Updates can overshoot the minimum.

Possible result:

$$
Loss\uparrow
$$

or oscillation/divergence.

### Good Learning Rate

Moves toward the minimum efficiently without excessive oscillation.

# 7. Momentum

## What?

Momentum adds a **memory of previous gradients**.

Instead of only looking at the current gradient, it also considers previous updates.

A common form is:

$$
\boxed{
v_t=\beta v_{t-1}+(1-\beta)g_t
}
$$

$$
\boxed{
w_t=w_{t-1}-\eta v_t
}
$$

where:

- $g_t=\nabla L(w_t)$
- $v_t$ = accumulated velocity
- $\beta$ = momentum coefficient

## Intuition

Imagine a ball rolling downhill.

Without momentum:

> It reacts mainly to the current slope.

With momentum:

> It keeps some movement from previous steps.

This can help:

- Speed up movement in consistent directions
- Reduce oscillations
- Move through shallow regions faster

---

# 8. Why Does Momentum Help?

Suppose gradients repeatedly point in roughly the same direction:

$$
g_1,g_2,g_3,\ldots
$$

Momentum accumulates that direction.

But if gradients keep changing direction:

$$
+,-,+,-,\ldots
$$

their accumulated effect can partially cancel.

Therefore momentum can:

$$
\boxed{
\text{Accelerate useful directions}
}
$$

and:

$$
\boxed{
\text{Reduce oscillations}
}
$$

# 9. Adam

## What?

Adam stands for:

**Adaptive Moment Estimation**

It combines:

1. Momentum-like first-moment estimation
2. Second-moment estimation of squared gradients

### First Moment

$$
m_t=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

### Second Moment

$$
v_t=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

Because these start at zero, Adam applies bias correction:

$$
\hat{m}_t=
\frac{m_t}{1-\beta_1^t}
$$

$$
\hat{v}_t=
\frac{v_t}{1-\beta_2^t}
$$

Final update:

$$
\boxed{
w_t
=
w_{t-1}
-
\eta
\frac{\hat{m}_t}
{\sqrt{\hat{v}_t}+\epsilon}
}
$$

---

# 10. Adam Intuition

Momentum asks:

> Which direction have we been moving?

Adam additionally asks:

> How large have the gradients typically been?

So Adam adapts the effective step size for different parameters.

Simplified intuition:

$$
\boxed{
Adam
=
Momentum
+
Adaptive\ Learning\ Rates
}
$$

This is why Adam is widely used for neural-network training.

# 11. Gradient Descent vs SGD vs Adam

| | Gradient Descent | SGD | Adam |
|---|---|---|---|
| Data per update | Full dataset | One sample | Usually mini-batch |
| Uses momentum | No | No | Yes |
| Adaptive step size | No | No | Yes |
| Gradient noise | Low | High | Depends on batch |
| Main advantage | Stable | Simple/fast updates | Fast adaptive optimization |

# 12. Optimization Problems

## Local Minimum

A point where the function is lower than nearby points.

## Global Minimum

The lowest point across the entire function.

For a convex function:

$$
\boxed{
Local\ Minimum=Global\ Minimum
}
$$

For non-convex functions, there can be many local minima and saddle points.

---

# 13. What Happens When Gradient Is Zero?

At a stationary point:

$$
\nabla L=0
$$

This does **not necessarily mean** we found a minimum.

It could be:

- Local minimum
- Local maximum
- Saddle point

Example:

$$
f(x)=x^2
$$

At:

$$
x=0
$$

$$
f'(x)=2x=0
$$

and $x=0$ is a minimum.

But for:

$$
f(x)=-x^2
$$

at:

$$
x=0
$$

the derivative is also zero, but it is a maximum.

# 15. Optimization in Neural Networks

The training process is essentially:

$$
\text{Input}
\rightarrow
\text{Forward Pass}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
$$

Then:

$$
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradients}
$$

Then:

$$
\text{Gradients}
\rightarrow
\text{Optimizer}
\rightarrow
\text{Updated Weights}
$$

Repeated over many iterations:

$$
\boxed{
Forward
\rightarrow
Loss
\rightarrow
Backward
\rightarrow
Update
}
$$

# 16. Important Interview Questions

### What is gradient descent?

An iterative optimization algorithm that updates parameters in the opposite direction of the gradient to minimize the loss.

### Why do we subtract the gradient?

Because the gradient points toward the direction of steepest increase.

### What happens if learning rate is too high?

The optimizer may overshoot the minimum, oscillate, or diverge.

### What happens if learning rate is too low?

Training becomes unnecessarily slow.

### Difference between Batch GD and SGD?

Batch GD uses the entire dataset for each update, while SGD uses one sample.

### Why use mini-batches?

They provide a practical balance between computational efficiency and gradient stability.

### What does momentum do?

It accumulates information from previous gradients to accelerate consistent movement and reduce oscillations.

### What is Adam?

An adaptive optimizer that combines first-moment and second-moment estimates of gradients with bias correction.

### Why is Adam popular?

It adapts the effective learning rate for parameters and often converges quickly with little manual tuning.

### Does gradient zero always mean minimum?

No. It can also be a maximum or saddle point.
