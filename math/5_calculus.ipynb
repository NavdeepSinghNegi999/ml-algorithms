# 1. Differentiation

## What?

A derivative tells us the **rate of change** of a function.

Geometrically, it is the **slope of a curve at a particular point**.

$$
\frac{dy}{dx}
$$

means:

> How much does $y$ change when $x$ changes?

## Example

Given:

$$
y=x^2
$$

The derivative is:

$$
\frac{dy}{dx}=2x
$$

At $x=3$:

$$
\frac{dy}{dx}=2(3)=6
$$

So at $x=3$, a small increase in $x$ produces approximately $6$ times that change in $y$.

## Important Rules

### Constant

$$
\frac{d}{dx}(c)=0
$$

### Power Rule

$$
\frac{d}{dx}(x^n)=nx^{n-1}
$$

Example:

$$
\frac{d}{dx}(x^3)=3x^2
$$

### Sum Rule

$$
\frac{d}{dx}[f(x)+g(x)]
=
f'(x)+g'(x)
$$

### Product Rule

$$
\frac{d}{dx}[f(x)g(x)]
=
f'(x)g(x)+f(x)g'(x)
$$

### Common Derivatives

$$
\frac{d}{dx}(e^x)=e^x
$$

$$
\frac{d}{dx}(\ln x)=\frac{1}{x}
$$

$$
\frac{d}{dx}(\sin x)=\cos x
$$

$$
\frac{d}{dx}(\cos x)=-\sin x
$$

---

# 2. Why Is Differentiation Important in ML?

ML models have a **loss function**.

For example:

$$
L(w)=(w-3)^2
$$

We want to find the value of $w$ that minimizes the loss.

Differentiate:

$$
\frac{dL}{dw}=2(w-3)
$$

At the minimum:

$$
\frac{dL}{dw}=0
$$

Therefore:

$$
2(w-3)=0
$$

$$
w=3
$$

So derivatives tell us **which direction we should move parameters to reduce the loss**.

This is the mathematical foundation of optimization.

---

# 3. Partial Derivatives

When a function has multiple variables, we use partial derivatives.

Example:

$$
f(x,y)=x^2+3y^2
$$

Partial derivative with respect to $x$:

$$
\frac{\partial f}{\partial x}=2x
$$

Here, $y$ is treated as constant.

Partial derivative with respect to $y$:

$$
\frac{\partial f}{\partial y}=6y
$$

Here, $x$ is treated as constant.

## Why?

ML models usually have many parameters:

$$
L(w_1,w_2,\ldots,w_n)
$$

We need to know how the loss changes when **one parameter changes while the others stay fixed**.

---

# 4. Chain Rule

## What?

The chain rule is used when one function is inside another function.

If:

$$
y=f(g(x))
$$

then:

$$
\frac{dy}{dx}
=
\frac{dy}{dg}
\frac{dg}{dx}
$$

## Example

Given:

$$
y=(3x+2)^2
$$

Let:

$$
u=3x+2
$$

Then:

$$
y=u^2
$$

First:

$$
\frac{dy}{du}=2u
$$

Second:

$$
\frac{du}{dx}=3
$$

Therefore:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
$$

$$
=2u(3)
$$

Substitute $u=3x+2$:

$$
\boxed{
\frac{dy}{dx}=6(3x+2)
}
$$

## ML Connection

The chain rule is the mathematical foundation of **backpropagation**.

A neural network contains many nested functions:

$$
x
\rightarrow
f_1
\rightarrow
f_2
\rightarrow
f_3
\rightarrow
Loss
$$

The chain rule allows us to calculate how the final loss changes with respect to an earlier weight.

# 5. Gradient

For a function with multiple variables:

$$
f(x_1,x_2,\ldots,x_n)
$$

the gradient is the vector of partial derivatives:

$$
\nabla f=
\begin{bmatrix}
\frac{\partial f}{\partial x_1}\\
\frac{\partial f}{\partial x_2}\\
\vdots\\
\frac{\partial f}{\partial x_n}
\end{bmatrix}
$$

## What Does It Tell Us?

The gradient points in the direction of **steepest increase** of the function.

Therefore:

$$
-\nabla f
$$

points toward the direction of steepest decrease.

This is why gradient descent uses the negative gradient.

---

# 6. Gradient Descent

## Goal

Minimize a loss function:

$$
L(\mathbf{w})
$$

The update rule is:

$$
\mathbf{w}_{new}
=
\mathbf{w}_{old}
-
\eta\nabla L(\mathbf{w})
$$

Where:

- $\mathbf{w}$ = model parameters
- $\eta$ = learning rate
- $\nabla L$ = gradient of the loss

## Why Subtract the Gradient?

The gradient points toward increasing loss.

Therefore, to decrease the loss, we move in the opposite direction:

$$
-\nabla L
$$

## Simple Example

Suppose:

$$
L(w)=w^2
$$

Then:

$$
\frac{dL}{dw}=2w
$$

Let:

$$
w=5,\qquad \eta=0.1
$$

Gradient:

$$
2(5)=10
$$

Update:

$$
w_{new}=5-(0.1)(10)
$$

$$
w_{new}=4
$$

The parameter moved toward the minimum:

$$
w=0
$$

# 7. Second Derivative

The first derivative tells us the **slope**.

The second derivative tells us how the **slope is changing**.

$$
\frac{d^2f}{dx^2}
$$

For:

$$
f(x)=x^2
$$

First derivative:

$$
f'(x)=2x
$$

Second derivative:

$$
f''(x)=2
$$

A positive second derivative generally indicates a locally convex/upward-curving function.

A negative second derivative generally indicates a locally concave/downward-curving function.

In multiple dimensions, second derivatives form the **Hessian matrix**.

# 8. Integration

## What?

Integration is essentially the process of **accumulating small quantities**.

It is also closely related to finding the **area under a curve**.

$$
\int f(x)\,dx
$$

## Example

Given:

$$
f(x)=2x
$$

Its integral is:

$$
\int 2x\,dx=x^2+C
$$

because:

$$
\frac{d}{dx}(x^2+C)=2x
$$

So differentiation and integration are closely related.

---

# 9. Definite Integration

A definite integral calculates accumulation over an interval.

$$
\int_a^b f(x)\,dx
$$

For example:

$$
\int_0^2 2x\,dx
$$

First find the antiderivative:

$$
x^2
$$

Then:

$$
\left[x^2\right]_0^2
=
2^2-0^2
=4
$$

So the area under $f(x)=2x$ from $0$ to $2$ is:

$$
4
$$

---

# 10. Why Is Integration Less Important for Typical ML Interviews?

Differentiation is directly used in:

$$
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Gradient Descent}
\rightarrow
\text{Model Training}
$$

Integration is more commonly encountered in:

- Probability density functions
- Expected values
- Continuous probability distributions
- Mathematical statistics

For most ML interviews, you generally need **basic integration intuition and common rules**, rather than advanced integration techniques.

---

# 11. Differentiation vs Integration

| | Differentiation | Integration |
|---|---|---|
| Main idea | Rate of change | Accumulation |
| Geometric idea | Slope | Area |
| ML importance | Very High | Low |
| Main ML use | Optimization, backpropagation | Probability/statistics |
| Relationship | Breaks down change | Accumulates change |

They are closely related:

$$
\frac{d}{dx}\left(\int f(x)\,dx\right)=f(x)
$$

and, under suitable conditions:

$$
\int_a^b f'(x)\,dx=f(b)-f(a)
$$

# 12. Important Interview Questions

### What does a derivative represent?

The instantaneous rate of change or slope of a function.

### Why are derivatives important in ML?

They tell us how changing model parameters changes the loss, allowing optimization algorithms such as gradient descent to update parameters.

### What is a partial derivative?

The derivative with respect to one variable while treating the other variables as constant.

### Why is the chain rule important in deep learning?

Because neural networks are compositions of functions, and backpropagation uses the chain rule to calculate gradients through these layers.

### What is a gradient?

A vector containing the partial derivatives of a multivariable function.

### Why does gradient descent subtract the gradient?

Because the gradient points toward the direction of greatest increase, so its negative points toward greatest decrease.

### What does the learning rate do?

It controls the size of each parameter update:

$$
\mathbf{w}_{new}
=
\mathbf{w}_{old}
-
\eta\nabla L
$$

### What is integration?

Integration represents accumulation and can be interpreted geometrically as area under a curve.

# Differentiation Formulas & Rules for ML

## Basic Rules

$$
\frac{d}{dx}(c)=0
$$

$$
\frac{d}{dx}(x)=1
$$

$$
\frac{d}{dx}(cx)=c
$$

$$
\frac{d}{dx}[cf(x)]=cf'(x)
$$

---

## Power Rule

$$
\boxed{
\frac{d}{dx}(x^n)=nx^{n-1}
}
$$

Example:

$$
\frac{d}{dx}(x^3)=3x^2
$$

---

## Sum & Difference Rules

$$
\frac{d}{dx}[f(x)+g(x)]
=
f'(x)+g'(x)
$$

$$
\frac{d}{dx}[f(x)-g(x)]
=
f'(x)-g'(x)
$$

---

## Product Rule

$$
\boxed{
\frac{d}{dx}[f(x)g(x)]
=
f'(x)g(x)+f(x)g'(x)
}
$$

---

## Quotient Rule

$$
\boxed{
\frac{d}{dx}\left[\frac{f(x)}{g(x)}\right]
=
\frac{f'(x)g(x)-f(x)g'(x)}
{[g(x)]^2}
}
$$

---

## Chain Rule

$$
\boxed{
\frac{d}{dx}f(g(x))
=
f'(g(x))g'(x)
}
$$

Or:

$$
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
}
$$

---

## Exponential Functions

$$
\frac{d}{dx}(e^x)=e^x
$$

$$
\frac{d}{dx}(e^{f(x)})
=
e^{f(x)}f'(x)
$$

For general base:

$$
\frac{d}{dx}(a^x)
=
a^x\ln(a)
$$

$$
\frac{d}{dx}(a^{f(x)})
=
a^{f(x)}\ln(a)f'(x)
$$

---

## Logarithms

$$
\boxed{
\frac{d}{dx}\ln(x)=\frac{1}{x}
}
$$

$$
\frac{d}{dx}\ln(f(x))
=
\frac{f'(x)}{f(x)}
$$

General logarithm:

$$
\frac{d}{dx}\log_a(x)
=
\frac{1}{x\ln(a)}
$$

---

## Trigonometric Functions

$$
\frac{d}{dx}\sin(x)=\cos(x)
$$

$$
\frac{d}{dx}\cos(x)=-\sin(x)
$$

$$
\frac{d}{dx}\tan(x)=\sec^2(x)
$$

---

## Inverse Trigonometric Functions

$$
\frac{d}{dx}\sin^{-1}(x)
=
\frac{1}{\sqrt{1-x^2}}
$$

$$
\frac{d}{dx}\cos^{-1}(x)
=
-\frac{1}{\sqrt{1-x^2}}
$$

$$
\frac{d}{dx}\tan^{-1}(x)
=
\frac{1}{1+x^2}
$$

---

## Absolute Value

For $x\neq0$:

$$
\frac{d}{dx}|x|
=
\operatorname{sign}(x)
$$

where:

$$
\operatorname{sign}(x)=
\begin{cases}
1,&x>0\\
-1,&x<0
\end{cases}
$$

At $x=0$, $|x|$ is not differentiable.

---

## Common ML Derivatives

### Sigmoid

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

$$
\boxed{
\sigma'(x)=\sigma(x)(1-\sigma(x))
}
$$

---

### ReLU

$$
ReLU(x)=\max(0,x)
$$

$$
ReLU'(x)=
\begin{cases}
1,&x>0\\
0,&x<0
\end{cases}
$$

At $x=0$, the derivative is technically undefined; implementations choose a convention.

---

### Tanh

$$
\boxed{
\frac{d}{dx}\tanh(x)
=
1-\tanh^2(x)
}
$$

---

### Softmax + Cross-Entropy

For logits $z_i$ and one-hot target $y_i$:

$$
\boxed{
\frac{\partial L}{\partial z_i}
=
p_i-y_i
}
$$

where:

$$
p_i=\frac{e^{z_i}}{\sum_j e^{z_j}}
$$

---

## Common Loss Derivatives

### Mean Squared Error

$$
L=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

$$
\boxed{
\frac{\partial L}{\partial \hat{y}_i}
=
\frac{2}{n}(\hat{y}_i-y_i)
}
$$

---

### Binary Cross-Entropy

$$
L=
-\left[y\ln(p)+(1-y)\ln(1-p)\right]
$$

With sigmoid output:

$$
\boxed{
\frac{\partial L}{\partial z}=p-y
}
$$

---

## Important Matrix Derivatives

$$
\boxed{
\nabla_{\mathbf{x}}(\mathbf{a}^T\mathbf{x})=\mathbf{a}
}
$$

$$
\boxed{
\nabla_{\mathbf{x}}(\mathbf{x}^T\mathbf{x})=2\mathbf{x}
}
$$

$$
\boxed{
\nabla_{\mathbf{x}}(\mathbf{x}^TA\mathbf{x})
=
(A+A^T)\mathbf{x}
}
$$

If $A$ is symmetric:

$$
\boxed{
\nabla_{\mathbf{x}}(\mathbf{x}^TA\mathbf{x})
=
2A\mathbf{x}
}
$$

---

## Derivative Rules to Memorize First

$$
\boxed{
Power,\ Product,\ Quotient,\ Chain
}
$$

Then memorize the common ML derivatives:

$$
\boxed{
e^x,\quad \ln x,\quad \sigma(x),\quad ReLU(x),\quad \tanh(x)
}
$$

and:

$$
\boxed{
MSE,\quad Binary\ Cross\text{-}Entropy,\quad Softmax+Cross\text{-}Entropy
}
$$

# Integration Formulas & Rules

## Basic Rules

$$
\int c\,dx=cx+C
$$

$$
\int 0\,dx=C
$$

$$
\int [f(x)+g(x)]\,dx
=
\int f(x)\,dx+\int g(x)\,dx
$$

$$
\int [f(x)-g(x)]\,dx
=
\int f(x)\,dx-\int g(x)\,dx
$$

$$
\int cf(x)\,dx
=
c\int f(x)\,dx
$$

---

## Power Rule

For $n\neq-1$:

$$
\boxed{
\int x^n\,dx
=
\frac{x^{n+1}}{n+1}+C
}
$$

Example:

$$
\int x^3\,dx
=
\frac{x^4}{4}+C
$$

---

## Special Case: $\frac{1}{x}$

The power rule cannot be used for $n=-1$.

Instead:

$$
\boxed{
\int\frac{1}{x}\,dx
=
\ln|x|+C
}
$$

---

## Exponential Functions

$$
\boxed{
\int e^x\,dx=e^x+C
}
$$

$$
\int e^{f(x)}f'(x)\,dx
=
e^{f(x)}+C
$$

For general base:

$$
\boxed{
\int a^x\,dx
=
\frac{a^x}{\ln a}+C
}
$$

---

## Logarithm

$$
\int\frac{1}{x}\,dx
=
\ln|x|+C
$$

More generally:

$$
\int\frac{f'(x)}{f(x)}\,dx
=
\ln|f(x)|+C
$$

---

## Trigonometric Functions

$$
\boxed{
\int\sin x\,dx=-\cos x+C
}
$$

$$
\boxed{
\int\cos x\,dx=\sin x+C
}
$$

$$
\boxed{
\int\sec^2x\,dx=\tan x+C
}
$$

$$
\boxed{
\int\csc^2x\,dx=-\cot x+C
}
$$

$$
\boxed{
\int\sec x\tan x\,dx=\sec x+C
}
$$

$$
\boxed{
\int\csc x\cot x\,dx=-\csc x+C
}
$$

---

## Inverse Trigonometric Forms

$$
\boxed{
\int\frac{1}{\sqrt{1-x^2}}\,dx
=
\sin^{-1}(x)+C
}
$$

$$
\boxed{
\int\frac{-1}{\sqrt{1-x^2}}\,dx
=
\cos^{-1}(x)+C
}
$$

$$
\boxed{
\int\frac{1}{1+x^2}\,dx
=
\tan^{-1}(x)+C
}
$$

---

## Substitution Rule

Used when an integral contains a function and its derivative.

If:

$$
u=g(x)
$$

then:

$$
du=g'(x)\,dx
$$

and:

$$
\boxed{
\int f(g(x))g'(x)\,dx
=
\int f(u)\,du
}
$$

Example:

$$
\int 2x(x^2+1)^3\,dx
$$

Let:

$$
u=x^2+1
$$

Then:

$$
du=2x\,dx
$$

Therefore:

$$
\int u^3\,du
=
\frac{u^4}{4}+C
$$

Substitute back:

$$
\boxed{
\frac{(x^2+1)^4}{4}+C
}
$$

---

## Integration by Parts

Derived from the product rule.

$$
\boxed{
\int u\,dv
=
uv-\int v\,du
}
$$

Useful when the integrand is a product of functions.

Example types:

$$
x e^x,\qquad x\ln x,\qquad x\sin x
$$

---

## Definite Integral

A definite integral has limits:

$$
\boxed{
\int_a^b f(x)\,dx
}
$$

Using the Fundamental Theorem of Calculus:

$$
\boxed{
\int_a^b f(x)\,dx
=
F(b)-F(a)
}
$$

where:

$$
F'(x)=f(x)
$$

Example:

$$
\int_0^2 2x\,dx
$$

Antiderivative:

$$
F(x)=x^2
$$

Therefore:

$$
\left[x^2\right]_0^2
=
2^2-0^2
=
4
$$

---

## Important Properties of Definite Integrals

$$
\int_a^a f(x)\,dx=0
$$

$$
\int_a^b f(x)\,dx
=
-\int_b^a f(x)\,dx
$$

$$
\int_a^b f(x)\,dx
+
\int_b^c f(x)\,dx
=
\int_a^c f(x)\,dx
$$

---

## Integration and Probability

For a continuous probability density function $f(x)$:

$$
\boxed{
\int_{-\infty}^{\infty}f(x)\,dx=1
}
$$

Probability over an interval:

$$
\boxed{
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx
}
$$

Expected value:

$$
\boxed{
E[X]
=
\int_{-\infty}^{\infty}x f(x)\,dx
}
$$

Variance:

$$
\boxed{
Var(X)
=
\int_{-\infty}^{\infty}(x-\mu)^2f(x)\,dx
}
$$

These are the main places integration connects with ML/statistics.
