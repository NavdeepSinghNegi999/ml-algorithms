# 1. Dataset

````python
import numpy as np

# X shape = (8,2)

X = np.array([
    [1,2],
    [2,1],
    [2,3],
    [3,2],
    [6,5],
    [7,6],
    [8,7],
    [9,8]
], dtype=float)

# y shape = (8,)

y = np.array([0,0,0,0,1,1,1,1])

print("X Shape :", X.shape)
print("y Shape :", y.shape)
````

# 2. Sklearn

````python
from sklearn.linear_model import Perceptron

sk_model = Perceptron(max_iter=1000)

sk_model.fit(X,y)

sk_pred = sk_model.predict(X)

print(sk_pred)

print("Accuracy :",sk_model.score(X,y))
````

# 3. From Scratch

````python
import numpy as np

class PerceptronScratch:

    def __init__(self, lr=0.1, epochs=100):

        self.lr = lr
        self.epochs = epochs


    def activation(self, x):

        return np.where(x >= 0, 1, 0)


    def fit(self, X, y):

        # X : (n_samples,n_features)
        # y : (n_samples,)

        n_samples, n_features = X.shape

        # w : (n_features,)
        self.w = np.zeros(n_features)

        # b : Scalar
        self.b = 0

        for _ in range(self.epochs):

            for i in range(n_samples):

                # X[i] : (n_features,)
                # Linear : Scalar
                linear = X[i] @ self.w + self.b

                # Prediction : Scalar
                pred = self.activation(linear)

                # Error : Scalar
                error = y[i] - pred

                # Weight Update
                self.w += self.lr * error * X[i]

                # Bias Update
                self.b += self.lr * error


    def predict(self, X):

        linear = X @ self.w + self.b

        return self.activation(linear)


my_model = PerceptronScratch()

my_model.fit(X,y)

my_pred = my_model.predict(X)

print(my_pred)

print("Weights :",my_model.w)

print("Bias :",my_model.b)
````

````python
from sklearn.metrics import accuracy_score

print("\n===== Sklearn =====")
print(accuracy_score(y, sk_pred))

print("\n===== Scratch =====")
print(accuracy_score(y, my_pred))
````

# Perceptron

## Step 1 : Linear Equation

$$
z=XW+b
$$

---

## Step 2 : Activation Function

$$
\hat y=
\begin{cases}
1,&z\ge0\\
0,&z<0
\end{cases}
$$

---

## Step 3 : Error

$$
Error=y-\hat y
$$

---

## Step 4 : Perceptron Loss

The Perceptron minimizes the Perceptron Criterion.

$$
J(W)
=
-\sum_{i\in M}
y_i(W^Tx_i+b)
$$

where

- $M$ = Misclassified samples

---

## Step 5 : Gradient

The gradient of the Perceptron Criterion is

$$
\frac{\partial J}{\partial W}
=
-
\sum_{i\in M}
y_iX_i
$$

$$
\frac{\partial J}{\partial b}
=
-
\sum_{i\in M}
y_i
$$

---

## Step 6 : Gradient Descent Update

General Gradient Descent Rule

$$
\boxed{
W
=
W
-
\alpha
\frac{\partial J}{\partial W}
}
$$

$$
\boxed{
b
=
b
-
\alpha
\frac{\partial J}{\partial b}
}
$$

---

## Step 7 : Perceptron Update Rule

Substituting the gradient gives

$$
\boxed{
W
=
W
+
\alpha
(y-\hat y)X
}
$$

$$
\boxed{
b
=
b
+
\alpha
(y-\hat y)
}
$$

which is exactly implemented as

```python
self.w += self.lr * error * X[i]
self.b += self.lr * error
```

where

```python
error = y[i] - pred
```

---
